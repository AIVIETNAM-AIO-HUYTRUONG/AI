## **#1: Understanding File Paths**

- Khi chương trình Python của bạn cần đọc dữ liệu từ một file hoặc ghi kết quả, trước hết nó cần biết file đó nằm ở đâu trong **storage (bộ nhớ lưu trữ)** của máy tính, như **hard drive (ổ cứng)** hoặc **SSD**.

- Giống như bạn cần một **address (địa chỉ)** để tìm một ngôi nhà cụ thể, chương trình cũng cần một **file path (đường dẫn tệp)** để tìm một file cụ thể.

- Path thực chất là một **address** của file trong cấu trúc phân cấp của các **folders**, còn gọi là **directories**

- Hãy tưởng tượng **file system (hệ thống tệp)** của máy tính như một **filing cabinet (tủ hồ sơ)** lớn:
    - **Root (Thư mục gốc)**: là tủ chính
    - **main folders (Thư mục chính)** hoặc **drives (ổ đĩa)**: là các ngăn
    - **sub-folders (Thư mục con)**: là các folder nằm bên trong các **main folders**
    - Cuối cùng các tài liệu bên trong là **files**

### **1. Components of a Path**

- Một **file path** thường gồm:
    1. **Directory/Folder Names (Tên thư mục)**: tên các folder mà bạn cần đi qua để đến file.
    2. **Separators (Ký tự phân tách)**: Các ký tự đặc biệt dùng để ngăn cách các **Directory/Folder Names** và **tên file cuối cùng**.
    3. **File Name**: Tên của file, thường có **extension (phần mở rộng)** như `.txt`, `.csv`, `.py`

### **2. Operating System Differences**

- **File paths** được biểu diễn hơi khác nhau tùy bạn đang dùng **Windows, macOS**, hay **Linux**.

#### **2.1 Windows**
- **Paths** thường bắt đầu bằng một **ký tự ổ đĩa (drive letter)** như `C:` hoặc `D:`
- Các **directory** được ngăn cách bằng **backslash (`\`)**:
    - VD: `C:\Users\YourUsername\Documents\report.txt`

    - Note: Windows quy định ký tự phân tách là **backslash** `\` . Tuy nhiên, Python trên Windows được thiết kế thông minh để chấp nhận cả dấu gạch chéo xuôi (`/`), vốn là chuẩn của **Linux** và **macOS**:
        - Thay vì phải viết: `path = "C:\\Users\\Name\\Documents\\data.csv"` (lưu ý phải dùng `\\` vì `\` là ký tự đặc biệt trong Python) - bạn hoàn toàn có thể viết: `path = "C:/Users/Name/Documents/data.csv"`

#### **2.2 MacOS and Linux (Unix-like systems)**

- Các hệ điều hành này dùng **một root directory** duy nhất, được ký hiệu bằng **forward slash `/`**
- Các **directory** luôn được ngăn cách bằng **forward slash**. Không có **drive letter** như trên **Windows**:
    - macOS: `/Users/yourusername/Documents/report.txt`
    - Linux: `/home/yourusername/Documents/report.txt`

### **3. Absolute Paths**

- **absolute path (Đường dẫn tuyệt đối)** cung cấp vị trí đầy đủ của một file hoặc directory, bắt đầu từ cấp cao nhất của file system.
    - Trên **Windows**, nó bắt đầu bằng **drive letter** (ví dụ: `C:\...`) - `C:\Program Files\Python310\python.exe`
    - Trên **macOS/Linux**, nó bắt đầu bằng **root slash** (`/...`):
        - macOS: `/Applications/Calculator.app`
        - Linux: `/usr/bin/python3`
- Absolute paths là **unambiguous (không mơ hồ)**. Chúng chỉ tới một vị trí cụ thể, bất kể script của bạn đang chạy ở đâu.

- Tuy nhiên, việc **hardcoding** absolute paths vào script có thể làm code kém **portable (linh động)**. Nếu bạn chuyển project sang vị trí khác hoặc chia sẻ cho người khác, absolute path có thể không còn đúng nữa.

### **4. Relative Paths**
- Relative Paths Là chỉ dẫn dựa trên **current working directory (CWD – thư mục làm việc hiện tại)** của bạn.

- **Ký hiệu đặc biệt** được dùng trong **relative paths**:
    - `.` : Biểu thị **current directory (thư mục hiện tại)**
    - `..` : Biểu thị **parent directory (thư mục cha, lên một cấp)**
- VD: Script Python đang chạy từ **thư mục `project/`** -> **CWD**:
```
Users/
└── alice/
    ├── scripts/
    │   └── utility.py
    └── project/          ← 📍 Bạn đang đứng ở đây (CWD)
        ├── script.py     ← File script đang chạy
        ├── data.txt
        └── input/
            └── config.json
```

- Để chương trình có thể :
    - Tìm được `data.txt`: Tìm file ngay tại chỗ tôi đang đứng **CWD (thư mục project hiện tại)** -> `/Users/alice/project/data.txt`.
    - Tìm được `input/config.json`: Bước vào thư mục `input` bên trong **folder** đó, rồi bắt đầu tìm file `config.json` trong **folder** đó. Đường dẫn đầy đủ sẽ là `/Users/alice/project/input/config.json`.
    - Tìm `utility.py`: Dùng `..` để bước ra khỏi **CWD**, lên thư mục `alice/`, rồi vào `scripts/`, lấy `utility.py`" → `/Users/alice/scripts/utility.py`

- Nếu bạn tổ chức các file dữ liệu bên trong thư mục project, bạn có thể dùng relative paths, và script vẫn chạy đúng ngay cả khi toàn bộ thư mục project được di chuyển sang nơi khác.

### **5. Summary**

- **File paths (Đường dẫn tệp)** là địa chỉ duy nhất giúp Python định vị và truy cập các tệp trên hệ thống lưu trữ của máy tính. Hiểu rõ về chúng là nền tảng để làm việc với bất kỳ dữ liệu nào được lưu trữ bên ngoài bộ nhớ chương trình.

        - Ưu điểm: **Unambiguous (Không mơ hồ)** – luôn chỉ đến cùng một vị trí, bất kể chương trình đang chạy từ đâu.

    - **Relative paths (Đường dẫn tương đối)**:

        - Ưu điểm: Cực kỳ **portable (linh động)**. Nếu bạn tổ chức các tệp dữ liệu trong cùng một cấu trúc thư mục với project, relative path sẽ hoạt động đúng ngay cả khi bạn di chuyển toàn bộ project sang vị trí khác.

## **#2: Opening and Closing Files**

- Để đọc từ hoặc ghi vào một file trên ổ đĩa của máy tính, trước hết bạn cần thiết lập một kết nối đến file đó từ `Python script`. Quá trình này được gọi là **opening the file (mở file)**.

- Python cung cấp một **built-in function (hàm tích hợp)**: `open()` để có thể làm việc này.

### **1. Using the `open()` Function**

- Hàm `open()` khá đơn giản để sử dụng. Ở mức cơ bản nhất, nó cần **filename (tên file)** và có thể thêm **path (đường dẫn)** của file bạn muốn làm việc, cùng với một **mode (chế độ)** để chỉ ra bạn muốn làm gì với file đó, chẳng hạn như đọc hoặc ghi.

    ```python
    file_object = open('filename', 'mode')
    ```

- **filename (tên file)**:
    - Là một chuỗi string biểu diễn tên file bạn muốn mở. Nếu file nằm cùng thư mục với Python script, bạn chỉ cần cung cấp tên file, ví dụ `'notes.txt'`.
    - Nếu file nằm ở nơi khác, bạn phải cung cấp `path` đến file đó. Cách viết `path` phụ thuộc vào **operating system (hệ điều hành)**, như đã nói ở phần **Understanding File Paths (Hiểu về đường dẫn tệp)**. - VD:
        - Dùng `../data/config.txt` cho `relative path (đường dẫn tương đối)`
        - Dùng `C:/Users/username/Documents/report.txt` trên **Windows** hoặc  `'/home/user/documents/report.txt'` trên **Linux/macOS** cho `absolute path (đường dẫn tuyệt đối)`

- **mode (chế độ)**:
    - Là một chuỗi chỉ ra cách file sẽ được sử dụng.
    - **Mode** quyết định những thao tác bạn được phép thực hiện, như đọc hay ghi, và file sẽ được xử lý như thế nào nếu nó đã tồn tại hoặc chưa tồn tại
    - Ba **mode** phổ biến nhất cho **text files (tệp văn bản)** là:
        - `r` (**Read Mode** – **chế độ đọc**):
            - Đây là mode mặc định nếu bạn không chỉ định gì thêm. Nó cho phép đọc dữ liệu từ một file đã tồn tại.
            - Nếu file không tồn tại, Python sẽ **báo lỗi FileNotFoundError**
                ```python
                config_file = open('config.txt', 'r')
                # Or simply, since 'r' is the default:
                # config_file = open('config.txt')
                ```

        - `w` (**Write Mode** – **chế độ ghi**):
            - Mode này cho phép ghi dữ liệu vào file
            - Nếu file đã tồn tại, nội dung hiện tại của nó sẽ bị xóa hoàn toàn trước khi bắt đầu ghi.
            - Nếu file chưa tồn tại, **Python** sẽ tạo một file mới rỗng.
            - Hãy dùng **mode** này cẩn thận vì nó có thể ghi đè dữ liệu cũ.
                ```python
                log_file = open('log.txt', 'a')
                ```

        - `a` (**Append Mode** – **chế độ nối thêm**):
            - Mode này cho phép thêm dữ liệu mới vào cuối file hiện có.
            - Nếu file đã tồn tại, dữ liệu mới sẽ được nối vào cuối file
            - Nếu file chưa tồn tại, **Python** sẽ tạo file mới, tương tự như `mode w`, nhưng không xóa nội dung cũ nếu file đã có sẵn.
                ```python
                output_file = open('output.txt', 'w')
                ```

### **2. File objects**

- Khi bạn gọi `open()` thành công, nó sẽ trả về một **file object (đối tượng file)**, đôi khi còn được gọi là **file handle**.

- Object này đóng vai trò như liên kết của Python với file thật trên ổ đĩa. Bạn sẽ dùng các **method (phương thức)** của **file object** để thực hiện các thao tác như đọc dữ liệu từ file hoặc ghi dữ liệu vào file.

````python
output_file = open('output.txt', 'w')
print(output_file)
````

- `<_io.TextIOWrapper name='output.txt' mode='w' encoding='utf-8'>`: Đây chính là hình dáng của **file object** khi in ra:
    - `_io.TextIOWrapper`-> kiểu object (text file)
    - `name='output.txt'` → tên file
    - `mode='w'` → tên mode
    - `encoding='utf-8'` → encoding

### **3. The Necessity of Closing Files**

- Sau khi bạn hoàn thành các thao tác trên file như đọc, ghi, hoặc nối thêm, việc `close the file (đóng file)` là rất quan trọng.
- **Đóng file** sẽ báo cho **operating system (hệ điều hành)** biết rằng bạn đã xong việc với file đó, và việc này có một số lợi ích:
    - **frees up system resources (giải phóng tài nguyên hệ thống)**:
        - **Hệ điều hành** có giới hạn số lượng file có thể mở cùng lúc cho một **user** hoặc một **program**.
        - **Đóng những file** không còn dùng nữa sẽ **giải phóng tài nguyên** này.
    - **ensures data is written (đảm bảo dữ liệu được ghi ra)**:
        - Khi bạn ghi dữ liệu vào file, dữ liệu có thể ban đầu được giữ tạm trong một vùng nhớ gọi là **buffer (bộ đệm)** để tăng hiệu quả.
        - **Đóng file** sẽ đảm bảo toàn bộ dữ liệu trong **buffer** được ghi thật sự xuống **disk**. Nếu không **đóng file**, đặc biệt sau khi ghi, dữ liệu có thể bị thiếu hoặc không đầy đủ.
    - **prevents data corruption (ngăn ngừa hỏng dữ liệu)**: để file mở không cần thiết, đặc biệt khi **program** kết thúc bất ngờ, đôi khi có thể dẫn đến **lỗi hỏng dữ liệu**.

### **4. Using the `close()` Method**

- **File objects** trong **Python** có `method` `close()` mà bạn phải gọi khi làm xong việc với file.

````python
# Open the file for writing
output_file = open('output.txt', 'w')

# Perform write operations (covered in the next sections)
output_file.write('This is the first line.\n')
output_file.write('This is the second line.\n')

# Explicitly close the file
output_file.close()
````

- Điều quan trọng là phải nhớ gọi `close()`.

    > Tuy nhiên, có một vấn đề: nếu trong **code** xảy ra **lỗi** sau khi bạn đã mở file nhưng trước khi tới dòng `close()` thì file có thể vẫn bị mở. **Python** cung cấp một cách an toàn hơn để xử lý việc **mở** và **đóng file** bằng **with statement (câu lệnh with / câu lệnh quản lý ngữ cảnh)**, giúp đảm bảo file được đóng tự động ngay cả khi có **lỗi** xảy ra. ## **#3: Reading Data from Files**

- Đọc data từ file là một tác vụ phổ biến trong Python. Sau khi mở file bằng `open()` ở **read mode (`r`)**, bước tiếp theo là trích xuất data.
- Python cung cấp **methods** để đọc từ **file object**. Việc chọn method phụ thuộc vào nhu cầu sử dụng

- Thực hành tốt nhất là sử dụng `with statement` để đảm bảo file được đóng tự động.

- Giả sử có file `greet.txt`:

    ```txt
    Hello, Python learner!
    Start file handling.
    Enjoy reading files.
    ```

### **1. Reading the Entire File Content with `read()`**

- Cách đơn giản nhất để lấy toàn bộ data của file là dùng `read()`.
- Khi gọi mà không truyền đối số nào, nó sẽ đọc từ vị trí hiện tại cho đến cuối file và trả về toàn bộ nội dung dưới dạng **string**.

````python
# Example: Reading the entire content of greet.txt
try:
    with open('../data/greet.txt', 'r') as file:
        content = file.read()
        print("--- File Content (using read()) ---")
        print(f"content: {content}")
        print("--- End of Content ---")
except FileNotFoundError:
    print("Error: greet.txt not found.")
````

- Code này sẽ in ra **toàn bộ content** trong `greet.txt`

#### **1.1 Potential Issue**
- Bạn có thể gặp vấn đề khi dùng `read()` với file lớn. Vì nó nạp toàn bộ nội dung file vào **memory (bộ nhớ)** dưới dạng một string duy nhất, nó tiêu tốn rất nhiều tài nguyên bộ nhớ và thậm chí làm chương trình bị crash nếu file quá lớn.

- Bạn cũng có thể truyền vào một số nguyên tùy chọn cho `read(size)` để chỉ định số **bytes** tối đa cần đọc.

- Với text files, con số này thường tương ứng với số **characters (ký tự)** theo encoding mặc định. Cách này hữu ích khi bạn chỉ cần một phần file hoặc muốn xử lý file theo từng **chunk (khối dữ liệu)**.

````python
# Example: Reading only the first 10 bytes
try:
    with open('../data/greet.txt', 'r') as file:
        partial_content = file.read(10)
        print("--- First 10 Bytes ---")
        print(partial_content)
        print("--- End of Partial Content ---")
except FileNotFoundError:
    print("Error: greet.txt not found.")
````

### **2. Reading One Line at a Time with `readline()`**

- Nếu chúng ta cần xử lý file từng dòng một, thì `readline()` lại rất nhưng ích. Mỗi lần gọi `readline()`, nó sẽ đọc một **complete line (dòng hoàn chỉnh)** từ file, bắt đầu từ vị trí hiện tại cho đến và bao gồm cả ký tự **newline `\n`** ở cuối dòng.
- Nếu tới cuối file và không còn dòng nào nữa, nó sẽ trả về một string rỗng `""`

````python
# Example: Reading lines one by one with readline()
try:
    with open('../data/greet.txt', 'r') as file:
        print("--- Reading lines with readline() ---")
        line1 = file.readline()
        print(f"Line 1: {line1}", end='') # end='' prevents extra newline

        line2 = file.readline()
        print(f"Line 2: {line2}", end='')

        line3 = file.readline()
        print(f"Line 3: {line3}", end='')

        # Calling it again after the last line
        end_of_file = file.readline()
        print(f"End of file check: '{end_of_file}' (Empty string indicates EOF)")
        print("--- Done reading lines ---")
except FileNotFoundError:
    print("Error: greet.txt not found.")
````

- **Lưu ý**:
    - `end=""` trong `print()`. Nó ngăn `print()` tự thêm newline của riêng nó. Vì `readline()` đã bao gồm newline từ file rồi.
    - Dùng `readline()` sẽ **memory-eficient (tiết kiệm bộ nhớ)** hơn `read()` khi xử lý file lớn mà bạn chỉ cần đọc từng dòng.

### **3. Reading All Lines into a List with `readlines()`**

- `readlines` đọc tất cả các dòng còn lại từ vị trí hiện tại trong file và trả về chúng dưới dạng một **list (danh sách)** các string.

- Mỗi string trong list tương ứng với một dòng trong file và, giống như `readline()`, nó có chứa ký tự newline ở cuối.

````python
# Example: Reading all lines into a list
try:
    with open('../data/greet.txt', 'r') as file:
        lines_list = file.readlines()
        print("--- Reading lines with readlines() ---")
        print(f"Type of result: {type(lines_list)}")
        print(f"Number of lines: {len(lines_list)}")
        print("Content of list:")
        print(lines_list)

        # Example of accessing individual lines
        if len(lines_list) > 0:
            print(f"First line from list: {lines_list[0]}", end='')
        print("--- Done reading lines ---")

except FileNotFoundError:
    print("Error: greet.txt not found.")
````

- Method này đọc toàn bộ file vào memory, tương tự như `read()`, nhưng tổ chức nó thành một list các dòng. Điều này có thể tiện nếu bạn cần truy cập ngẫu nhiên vào các dòng khác nhau, nhưng nó cũng có vấn đề về bộ nhớ như `read()` khi file quá lớn.

### **4. The Recommended Approach: Iterating Directly Over the File Object**

- Cách tốt nhất để đọc file từng dòng là lặp trực tiếp trên file object bằng `for`.
- Python sẽ tự xử lý việc đọc dòng một cách hiệu quả, chỉ nạp một dòng hoặc một **buffer (bộ đệm)** nhỏ vào memory tại một thời điểm. Đây là cách được ưu tiên khi đặc biệt xử lý các file lớn.

````python
# Example: Iterating directly over the file object
try:
    with open('../data/greet.txt', 'r') as file:
        print("--- Iterating directly over the file object ---")
        line_number = 1
        for line in file:
            # Process each line here
            # Often useful to strip whitespace/newlines
            processed_line = line.strip()
            print(f"Line {line_number}: '{processed_line}'")
            line_number += 1
        print("--- Done iterating ---")
except FileNotFoundError:
    print("Error: greet.txt not found.")

````

- Cách này kết hợp cả **readability (dễ đọc)** và **efficiency (hiệu quả)**.
- Bên trong vòng lặp, biến `line` chứa dòng hiện tại vừa được đọc từ file, bao gồm cả ký tự newline.
- Ta thường dùng các **string method** như `strip()` hoặc `rstrip()` trong vòng lặp để xóa **whitespace (khoảng trắng)** ở đầu/cuối chuỗi, trước khi xử lý nội dung của từng dòng.

Có bốn phương pháp chính để đọc dữ liệu từ một đối tượng tệp (file object) đã mở:

##### **5.2.1. `file.read()`**

- **Mục đích**: Đọc toàn bộ nội dung của tệp từ vị trí hiện tại đến cuối tệp.
- **Kết quả**: Trả về toàn bộ nội dung dưới dạng một chuỗi (`string`) duy nhất.

- **Lưu ý**: Khi đạt đến cuối tệp và không còn dòng nào để đọc, nó sẽ trả về một chuỗi rỗng (`""`).

##### **5.2.3. `file.readlines()`**

## **#4: Writing Data to Files**

- Python cho phép bạn **write data to files (ghi dữ liệu vào file)**.
- Khả năng này rất quan trọng để **lưu output của chương trình (saving program output)**, **lưu nội dung do người dùng tạo ra (storing user-generated content)**, hoặc **tạo file cấu hình (configuration files)**.

### **1. Opening a File for Writing**

- Để ghi vào file, bạn cần mở file ở một **mode** cho phép ghi (`w` hoặc `a`) của `open()`.

````python
# Open 'output.txt' in write mode
# If 'output.txt' exists, its contents will be erased!
# If it doesn't exist, it will be created.
file_object = open('output.txt', 'w')
# ... perform write operations ...
file_object.close() # Remember to close
````

- **Important**: (Quan trọng)
    - Mở file ở mode `w` có một **side effect (tác dụng phụ)** rất lớn:
        - Nếu file đã tồn tại, toàn bộ nội dung hiện có sẽ bị xóa ngay lập tức.
        - Nếu file chưa tồn tại, Python sẽ tạo file đó cho bạn.
        - Hãy cẩn thận khi dùng `w` để tránh mất dữ liệu ngoài ý muốn.

    - Cũng giống như khi đọc file, dùng `with statement` là cách được khuyến nghị, **vì nó tự động đóng file kể cả khi có lỗi**

````python
# Preferred way: Using 'with'
with open('output.txt', 'w') as f:
    # 'f' is our file object, ready for writing
    pass # We'll add write operations here
# File is automatically closed when exiting the 'with' block
````

### **2. Using the `write()` Method**

- Khi file đã được mở ở mode `w` hoặc `a`, bạn có thể ghi **string data (dữ liệu dạng chuỗi)** vào file bằng method `write()` của file object. Method này nhận vào một **string duy nhất** và **ghi nó vào file ở vị trí hiện tại**.

````python
# Define the text we want to write
message = "Hello from Python!\n" # Note the newline character \n

try:
    # Open the file in write mode using 'with'
    with open('greeting.txt', 'w') as f:
        f.write(message) # Write the string to the file
    print("Successfully wrote to greeting.txt")

    # Optional: Verify by reading the file back
    with open('greeting.txt', 'r') as f:
        content = f.read()
        print("File content:")
        print(content)

except IOError as e:
    print(f"An error occurred: {e}")
````

- Phân tích code chạy:
    - Tạo mới (hoặc ghi đè) file `greeting.txt` trong cùng thư mục với script.
    - Ghi chuỗi `"Hello from Python!\n"` vào file đó.
    - In ra thông báo thành công.
    - Phần tùy chọn sẽ mở lại file ở chế độ đọc và in nội dung để xác nhận thao tác ghi.

- **LƯU Ý**:
    - `\n` ở cuối string `message`. Đây là **newline character (ký tự xuống dòng)**.
    - `write()` **không tự động** thêm newline sau khi ghi chuỗi. Nếu bạn muốn các lần ghi tiếp theo xuất hiện dòng mới, bạn phải thêm `\n` vào string đang ghi.

### **3. Writing Multiple Lines**

- Nếu bạn cần ghi nhiều dòng text, bạn có thể lựa chọn những option sau:

#### **3.1 Multiple `write()` calls**

- Gọi `write()` cho từng dòng, và đảm bảo mỗi chuỗi đều kết thúc bằng `\n`.

````python
lines_to_write = [
    "First line.\n",
    "Second line.\n",
    "Third line.\n"
]

try:
    with open('multiple_lines.txt', 'w') as f:
        for line in lines_to_write:
            f.write(line) # Write each line including its newline
    print("Successfully wrote multiple lines.")

    # Optional: Verify by reading the file back
    with open('multiple_lines.txt', 'r') as f:
        content = f.read()
        print("File content:")
        print(content)
except IOError as e:
    print(f"An error occurred: {e}")
````

#### **3.2. Using `writelines()`**:

- Method này nhận một **list** hoặc bất kỳ **iterable** nào chứa các chuỗi và ghi từng chuỗi vào file.
- Giống như `write()`, nó **không tự động thêm newline**, các chuỗi của bạn phải tự đặt `\n` nếu cần.

````python
lines_to_write = [
    "Report Header\n",
    "--------------\n",
    "Data point 1: Value A\n",
    "Data point 2: Value B\n"
]

try:
    with open('report.txt', 'w') as f:
        f.writelines(lines_to_write) # Write all strings from the list
    print("Successfully wrote report using writelines().")

    # Optional: Verify by reading the file back
    with open('report.txt', 'r') as f:
        content = f.read()
        print("File content:")
        print(content)

except IOError as e:
    print(f"An error occurred: {e}")
````

### **4. Writing Non-String Data**

- `write()` **chỉ chấp nhận string**.
- Nếu bạn có dữ liệu kiểu khác, bạn phải chuyển chúng thành **string** với `str()` trước khi ghi.

````python
item = "Widget"
quantity = 15
price = 29.95
total_cost = quantity * price

try:
    with open('order_summary.txt', 'w') as f:
        f.write("Order Summary\n")
        f.write("=============\n")
        f.write("Item: " + item + "\n")
        f.write("Quantity: " + str(quantity) + "\n") # Convert integer to string
        f.write("Price per item: $" + str(price) + "\n") # Convert float to string
        f.write("Total Cost: $" + str(total_cost) + "\n") # Convert float to string
    print("Successfully wrote order summary.")


    # Optional: Verify by reading the file back
    with open('order_summary.txt', 'r') as f:
        content = f.read()
        print("File content:")
        print(content)

except IOError as e:
    print(f"An error occurred: {e}")
````

#### **5.1. Mở File để Ghi (Opening a File for Writing)**

- **Hàm `open()` và các chế độ (`mode`)**: Để ghi dữ liệu, bạn cần mở file bằng hàm `open()` với một trong các chế độ sau:
    - `w` (**Write Mode - Chế độ ghi**): Nếu file đã tồn tại, toàn bộ nội dung của nó sẽ bị **xóa sạch** trước khi ghi. Nếu file chưa tồn tại, Python sẽ tạo một file mới. Hãy sử dụng chế độ này một cách cẩn thận để tránh mất dữ liệu.
    - `a` (**Append Mode - Chế độ nối thêm**): Nếu file đã tồn tại, dữ liệu mới sẽ được thêm vào cuối file mà không làm mất nội dung cũ. Nếu file chưa tồn tại, Python sẽ tạo một file mới (tương tự `w`, nhưng không xóa nếu file đã có).

- **Sử dụng `with statement`**: Luôn ưu tiên dùng `with open('filename', 'mode') as f:` để đảm bảo file được đóng tự động sau khi kết thúc khối lệnh, ngay cả khi có lỗi xảy ra. Điều này giúp ngăn ngừa mất dữ liệu và giải phóng tài nguyên hệ thống.

#### **5.2. Các Phương Pháp Ghi Dữ Liệu**

- **Ghi chuỗi với `f.write(string)`**:
    - Method `write()` nhận một chuỗi (`string`) làm đối số và ghi chuỗi đó vào file tại vị trí hiện tại của con trỏ file.
    - **Quan trọng**: `write()` **không tự động thêm ký tự xuống dòng (newline character `\n`)**. Nếu bạn muốn mỗi lần ghi là một dòng mới trong file, bạn phải tự thêm `\n` vào cuối chuỗi cần ghi.

- **Ghi nhiều dòng với `f.writelines(iterable_of_strings)`**:
    - Method `writelines()` nhận một đối tượng iterable (thường là một list) chứa các chuỗi và ghi từng chuỗi trong iterable đó vào file.
    - Tương tự như `write()`, `writelines()` cũng **không tự động thêm `\n`**. Mỗi chuỗi trong iterable cần phải có `\n` ở cuối nếu bạn muốn chúng xuất hiện trên các dòng riêng biệt trong file.

#### **5.3. Ghi Dữ Liệu Không Phải Chuỗi (Writing Non-String Data)**

- Method `write()` và `writelines()` **chỉ chấp nhận dữ liệu kiểu chuỗi (`string`)**.
- Nếu bạn có dữ liệu ở các kiểu khác (ví dụ: số nguyên `int`, số thực `float`, boolean `bool`, hoặc các đối tượng khác), bạn phải chuyển đổi chúng thành chuỗi bằng hàm `str()` trước khi truyền cho `write()` hoặc `writelines()`. Ví dụ: `f.write(str(my_number))`.

**Tóm lại**: Khi ghi file, luôn nhớ chọn chế độ mở file phù hợp (`w` để ghi đè, `a` để nối thêm), sử dụng `with statement` để quản lý file an toàn, và đảm bảo mọi dữ liệu được ghi đều là chuỗi, thêm ký tự xuống dòng `\n` thủ công nếu cần.

## **#5: Appending to Files**

- Đôi khi chúng ta không muốn thay đổi toàn bộ nội dung của file, mà chỉ muốn **thêm data mới vào cuối file**. Điều này rất phổ biến trong các tác vụ như:
    - **logging (ghi log)**.
    - **updating records (cập nhật bản ghi)**
    - Thêm data theo thời gian mà không làm mất data cũ.

- Mode `w` không còn phù hợp trong trường hợp này vì nó sẽ **xóa toàn bộ nội dung cũ** và bắt đầu tạo file mới.

- Cách duy nhất chúng ta là dùng **append mode** bằng cách truyền `a` làm **mode argument (đối số chế độ)** vào `open()`.

- Khi mở file ở **append mode**:
    - Nếu file đã tồn tại: **file pointer (con trỏ file)**, tức là vị trí mà lần ghi tiếp theo sẽ diễn ra được đặt ở cuối file. Mọi data bạn ghi sẽ được thêm vào sau nội dung hiện có.
    - Nếu file chưa tồn tại: Python sẽ tạo một file mới rỗng hoàn toàn cho bạn, giống như khi dùng **write mode `w`**

````python
# Open log.txt in append mode ('a')
with open('log.txt', 'a') as f:
    # Write a new line to the end of the file
    # Note: We add '\n' to ensure the new entry starts on a new line
    f.write("Event: User Login\n")
    f.write("Event: Data Processed\n")

# The 'with' block automatically closes the file

# Let's verify the content
with open('log.txt', 'r') as f:
    content = f.read()
    print(content)
````

- Dùng **append mode** với **with statement** là cách chuẩn để thêm thông tin vào cuối file, việc này đảm bảo **data integrity (tính toàn vẹn dữ liệu)** và **proper resource management (quản lý tài nguyên đúng cách)**
- Nó cung cấp một cơ chế đơn giản để mở rộng file mà không bị mất data cũ.

## **#6: Using with for Automatic File Closing**

- Khi làm việc với file trong Python, bạn thường dùng `open()` để mở file, rồi sau đó mới gọi `close` để đóng file. Cách này hoạt động cơ bản tốt với file, nhưng có một vấn đề lớn:
    - Nếu chương trình gặp lỗi sau khi mở file trước khi `close()` được gọi. Trong trường hợp đó, file có thể bị mở ngoài ý muốn: dẫn đến **data corruption (hỏng dữ liệu)** hoặc **làm tiêu tốn tài nguyên hệ thống**
    - Quên đóng file, hoặc gặp lỗi khiến `close()` không gọi được.
    - Vậy nên Python cung cấp một cách xử lý đáng tin cậy hơn, bằng **with statement**.
        - `with` đảm bảo rằng các **resources (tài nguyên)** như file, được quản lý đúng cách.
        - Khi dùng `with` để mở file, Python đảm bảo file sẽ được **automatically close (tự động đóng)**.

### **1. The `with` Statement Syntax**
- Cấu trúc cơ bản để mở file bằng `with` như sau:

    ```python
    with open('path/to/your/file.txt', 'mode') as file_variable:
        # Code to work with the file goes here
        # Use file_variable to read or write
        # e.g., content = file_variable.read()
        pass # Replace pass with your file operations
    
    # Execution continues here after the 'with' block
    # The file is automatically closed by this point
    ```

- Hãy phân tích cấu trúc này:
    1. `with`: từ khóa bắt **đầu khối context management (quản lý ngữ cảnh)**.
    2. `open('path/to/your/file.txt', 'mode')`: Lời gọi hàm `open`, chỉ ra **file path** và **read mode** `'r'`.
    3. `as file_variable`: Hàm `open()` trả về một **file object**. Từ khóa **as** gán object này vào một biến, ở đây là `file_variable`, để bạn dùng bên trong khối `with`.
    4. **Indented block (khối thụt lề)**: Toàn bộ code được thụt vào dưới câu lệnh `with` sẽ chạy trong lúc file đang mở. Bạn dùng `file_variable `để thực hiện các thao tác như đọc hoặc ghi.
    5. **Automatic Closing (đóng tự động)**: Khi code rời khỏi **indented block**, Python sẽ bắt đầu tự động thực hiện các bước dọn dẹp cần thiết cho **file object**, tức là đóng file. Bạn không cần gọi `file_variable.close()`.

### **2. Reading a File with `with`**

- VD đọc nội dung của file `example.txt` bằng `with`:

````python
lines_to_write = ["First line.\n", "Second line.\n", "Third line.\n"]

try:
    with open('output.txt', 'w') as outfile:
        outfile.writelines(lines_to_write)
        print("Data written to output.txt successfully.")
    # File is now closed automatically

    # Let's verify by reading it back (using 'with' again!)
    with open('output.txt', 'r') as infile:
        print("\nVerifying file content:")
        print(infile.read())

except IOError as e:
    print(f"An error occurred during file writing or reading: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
````

- Trong ví dụ này:
    1. `example.txt` được mở ở **read mode (chế độ đọc)** `'r'`.
    2. **File object** được gán vào biến `f`.
    3. Chúng ta đọc nội dung của file ngay trong khối `with`.
    4. Dù thao tác `read()` thành công hay có lỗi khác xảy ra trong block, Python vẫn đảm bảo `f.close()` được gọi ngầm trước khi thoát khỏi khối đó.

### **3. Writing to a File with `with`**

- `with` cũng hoạt động hiệu quả khi ghi file:

````python
lines_to_write = ["First line.\n", "Second line.\n", "Third line.\n"]

try:
    with open('output.txt', 'w') as outfile:
        outfile.writelines(lines_to_write)
        print("Data written to output.txt successfully.")
    # File is now closed automatically

    # Let's verify by reading it back (using 'with' again!)
    with open('output.txt', 'r') as infile:
        print("\nVerifying file content:")
        print(infile.read())

except IOError as e:
    print(f"An error occurred during file writing or reading: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
````

- Ở đây:
    1. `output.txt` được mở ở **write mode** `w`
    2. Chúng ta sẽ dùng `writelines` để ghi nhiều dòng.
    3. Ngay khi code thoát khỏi khối `with`, `output.txt` được đóng, đảm bảo toàn bộ dữ liệu trong **buffer (bộ đệm)** được ghi xuống **disk**.
    4. Sau đó, ta mở lại file an toàn bằng một `with statement` khác để đọc và kiểm tra nội dung.

### **4. Why Use `with`?**

- Dùng `with` là cách chuẩn và được khuyến nghị khi làm việc với file trong Python vì:

    - **Safety (an toàn)**: đảm bảo file được đóng đúng cách ngay cả khi có lỗi, tránh **resource leaks (rò rỉ tài nguyên)** và nguy cơ **data corruption (hỏng dữ liệu)**.
    - **Readability (dễ đọc)**: code sạch hơn và dễ hiểu hơn. Cấu trúc block làm rõ phạm vi mà file đang mở.
    - **Convenience (tiện lợi)**: không cần viết thêm các khối `try...finally` chỉ để đóng file, giúp giảm **boilerplate code (code lặp, code khuôn mẫu)**.

### **5. Summary**

- **Vấn đề khi mở và đóng tệp truyền thống**: Khi làm việc với tệp bằng cách gọi `open()` và sau đó là `close()`, có nguy cơ tệp không được đóng đúng cách nếu có lỗi xảy ra giữa hai thao tác này. Điều này có thể dẫn đến **rò rỉ tài nguyên (resource leaks)** và **hỏng dữ liệu (data corruption)**.

    - Khi sử dụng `with open(...) as file_variable:`, Python đảm bảo rằng tệp sẽ được **tự động đóng (automatically closed)** ngay cả khi có ngoại lệ (lỗi) xảy ra trong khối `with`.

    ```python
    with open('path/to/your/file.txt', 'mode') as file_variable:
        # Code to work with the file goes here

    ```

- **Ưu điểm chính của `with statement`**:
    - **An toàn (Safety)**: Đảm bảo tệp luôn được đóng, ngăn ngừa mất dữ liệu và giải phóng tài nguyên hệ thống một cách hiệu quả.
    - **Dễ đọc (Readability)**: Cấu trúc rõ ràng giúp mã nguồn dễ hiểu hơn, vì nó giới hạn phạm vi mà tệp đang mở.
    - **Tiện lợi (Convenience)**: Giảm thiểu **code lặp (boilerplate code)** bằng cách không cần phải viết khối `try...finally` chỉ để đảm bảo tệp được đóng. Bạn không cần gọi `file_variable.close()` một cách thủ công.

## **#7: Working with Different File Modes**

- Khi mở một file trong Python bằng hàm open(), bạn cần cho Python biết bạn muốn tương tác với file đó như thế nào?
    - Bạn muốn **đọc (read)** từ file
    - Bạn muốn **ghi (write)** nội dung mới vào file, có thể xóa data cũ
    - Bạn muốn **thêm (apppend)** data mới vào cuối file.
- Nếu bạn không truyền mode, Python sẽ mặc định rằng bạn muốn đọc file ở **text mode (chế độ văn bản)**, tương đương với `'rt'`.

- Tuy nhiên, chúng ta nên làm rõ ý định của mình là một cách tốt nhất để tránh lỗi xảy ra

### **1. Primary Modes**

- Các ký tự sau xác định thao tác chính bạn muốn thực hiện:

#### **1.1 `'r'` (Read)**

- Đây là mode mặc định. Nó mở file để đọc. Nếu file không tồn tại, Python sẽ **báo lỗi FileNotFoundError**.
- **File pointer (con trỏ file)**, tức là vị trí hiện tại để đọc hoặc ghi

````python
# Open data.txt for reading (explicitly)
try:
    with open('data.txt', 'r') as f:
        content = f.read()
        print("File content:", content)
except FileNotFoundError:
    print("Error: data.txt not found.")
````

#### **1.2 `'w'` (Write)**

- Mode này mở file để ghi. Hãy cẩn thận: nếu file đã tồn tại, nội dung của nó sẽ bị xóa ngay lập tức **(truncated – bị cắt/xóa nội dung)**.
- Nếu file chưa tồn tại, nó sẽ được tạo mới.
- **File pointer** được đặt ở đầu file.

````python
# Open output.txt for writing. Creates it or erases existing content.
with open('output.txt', 'w') as f:
    f.write("This is the first line.\n")
    f.write("This overwrites anything previously in the file.")

with open('output.txt', 'r') as f:
    content = f.read()
    print("File content:", content)
````

#### **1.3 `'a'` (Append)**

- Mode này mở file để nối thêm dữ liệu. Mọi dữ liệu được ghi vào sẽ được thêm vào cuối file.
- Nếu file chưa tồn tại, nó sẽ được tạo mới.
- File pointer được đặt ở cuối file, nên nội dung cũ không bị ghi đè. Mode này rất hữu ích cho các tác vụ như thêm bản ghi vào **log file (file nhật ký)**.

````python
# Open log.txt for appending. Creates it if it doesn't exist.
with open('log.txt', 'a') as f:
    f.write("New log entry added.\n")

with open('log.txt', 'r') as f:
    content = f.read()
    print("File content:", content)
````

#### **1.4 `'x'` (Exclusive Creation)**

- Mode này dùng để tạo file mới và mở nó để ghi.
- Nếu file đã tồn tại, Python sẽ báo lỗi **FileExistsError**. Điều này giúp tránh vô tình ghi đè lên một file có sẵn khi bạn chỉ muốn tạo file mới.

````python
# Try to create a new config file; fail if it exists.
try:
    with open('config_new.ini', 'x') as f:
        f.write("[Settings]\n")
        f.write("Mode=Test\n")
        print("config_new.ini created successfully.")

    with open('config_new.ini', 'r') as f:
        content = f.read()
        print("File content:", content)

except FileExistsError:
    print("Error: config_new.ini already exists. Cannot overwrite.")
````

### **2. Mode Modifiers**
- Bạn có thể kết hợp với các **primary modes** với các **Mode Modifiers**:

#### **2.1 `'t'` (Text Mode)**

- Modifier này chỉ ra rằng file sẽ được xử lý như **text (văn bản)**. Đây là modifier mặc định nếu không chỉ định gì thêm.
- Trong **text mode**, Python **xử lý encoding (mã hóa)**, tức là **chuyển đổi string sang bytes khi ghi** và **bytes sang string khi đọc**, đồng thời **tự động chuyển đổi các line endings (ký tự xuống dòng theo hệ thống)** đặc thù của từng hệ điều hành.

- `'r'` tương đương với `'rt'` và `'w'` tương đương với `'wt'`

#### **2.2 `'b'` (Binary Mode)**

- Modifier này chỉ ra rằng file sẽ được xử lý như một chuỗi các raw bytes (byte thô). Dùng mode này khi làm việc với file không phải văn bản, như:
    - **Image files (file ảnh)**
    - **Audio files (file âm thanh)**
    - **Executable programs (chương trình thực thi)**
    - hoặc bất kỳ dữ liệu nào cần kiểm soát chính xác ở mức byte.
- Trong binary mode, Python **không thực hiện encoding/decoding** hay **chuyển đổi line ending**.
- Dữ liệu được đọc và ghi dưới dạng **bytes objects (đối tượng bytes)**. Kết hợp nó với **primary mode**, ví dụ:
    - **`'rb'` (read binary – đọc nhị phân)**
    - **`'wb'` (write binary – ghi nhị phân)**
    - **`'ab'` (append binary – nối thêm nhị phân)**
    - **`'xb'` (exclusive creation binary – tạo mới nhị phân độc quyền)**.

````python
# Example: Writing raw bytes to a file
data_to_write = b'\x00\x10\xFF\xEE' # A bytes literal
with open('binary_data.bin', 'wb') as f:
    f.write(data_to_write)

# Example: Reading raw bytes from a file
with open('binary_data.bin', 'rb') as f:
    read_data = f.read()
    print("Read binary data:", read_data)
````

#### **2.3 `'+'` (Update Mode)**

- Modifier này cho phép cả **reading (đọc)** và **writing (ghi)** trên cùng một **file object**. Nó phải được kết hợp với một trong các **primary modes (`'r'`, `'w'`, hoặc `'a'`)**:
    - `'r+'`: mở file để vừa đọc vừa ghi. File pointer ở đầu file. File phải tồn tại, giống như 'r'.
    - `'w+'`: mở file để vừa ghi vừa đọc. Tạo file nếu chưa tồn tại, hoặc xóa nội dung nếu file đã có, giống như 'w'. File pointer ở đầu file.
    - `'a+'`: mở file để vừa nối thêm vừa đọc. Tạo file nếu chưa tồn tại. File pointer ban đầu ở cuối file cho các thao tác ghi. Vị trí đọc phụ thuộc vào nơi bạn dùng seek() để di chuyển con trỏ.

````python
# Example using 'r+' to read and then overwrite part of a file
# Assume file 'data.txt' contains "Line 1\nLine 2\n"
try:
    with open('data.txt', 'r+') as f:
        content = f.read()
        print("Original content:", content)
        # Move pointer back to the beginning to overwrite
        f.seek(0)
        f.write("Overwritten!")
        # Note: The rest of the original content might remain if the new content is shorter
except FileNotFoundError:
    print("Error: data.txt not found.")
````

### **3. Summary of Common Modes**

| Mode | Mô tả | Nếu file tồn tại | Nếu file không tồn tại | Vị trí con trỏ |
|---|---|---|---|---|
| `r` | Read (đọc, mặc định) | Đọc từ đầu | `FileNotFoundError` | Start (đầu) |
| `w` | Write (ghi) | Truncate (xóa nội dung) | Create (tạo mới) | Start |
| `a` | Append (nối thêm) | Append to end (thêm vào cuối) | Create | End (cuối) |
| `x` | Exclusive Create (tạo mới độc quyền) | `FileExistsError` | Create | Start |
| `rb` | Read binary (đọc nhị phân) | Đọc từ đầu | `FileNotFoundError` | Start |
| `wb` | Write binary (ghi nhị phân) | Truncate (xóa nội dung) | Create | Start |
| `ab` | Append binary (nối thêm nhị phân) | Append to end | Create | End |
| `r+` | Read and Write (đọc và ghi) | Đọc/ghi từ đầu | `FileNotFoundError` | Start |
| `w+` | Write and Read (ghi và đọc) | Truncate (xóa nội dung) | Create | Start |
| `a+` | Append and Read (nối thêm và đọc) | Append to end | Create | End (cho thao tác ghi) |
