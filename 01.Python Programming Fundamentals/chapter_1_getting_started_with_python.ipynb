## **#3: Understanding Interpreted vs Compiled Languages**

- Khi bạn viết mã bằng Python hoặc bất kỳ ngôn ngữ lập trình nào, bạn đang viết **các chỉ dẫn ở dạng mà con người có thể hiểu tương đối dễ dàng**. Tuy nhiên, máy tính thực chất chỉ có thể hoạt động ở **mức rất thấp**, thực thi các lệnh được mã hóa dưới dạng **chuỗi nhị phân (0 và 1)**, thường được gọi là **mã máy (machine code)**.

- Để thu hẹp khoảng cách này, **mã nguồn (source code) mà con người viết cần được chuyển đổi thành mã máy**. Có hai cách chính để thực hiện quá trình chuyển đổi này:

    - **Biên dịch (Compilation)**
    - **Thông dịch (Interpretation)**

- Python thường được biết đến là **ngôn ngữ thông dịch (interpreted language)**. Hiểu được sự khác biệt này sẽ giúp bạn hiểu rõ hơn một số đặc điểm của Python

### **Compiled Languages: Translating the Whole Book First**

- Bạn có thể hình dung **ngôn ngữ biên dịch** giống như việc **dịch toàn bộ một cuốn sách sang ngôn ngữ khác trước khi đưa cho người đọc**:

    1. **Viết chương trình (Writing)**: Bạn viết chương trình của mình bằng **ngôn ngữ cấp cao** như c, C++, Go, ... Thì đây chính là **mã nguồn (source code)** của bạn.
    2. **Biên dịch (Compilling)**: Bạn sử dụng một chương trình đặc biệt gọi là **compiler (trình biên dịch)**.
        - Compiler sẽ:
            - Đọc **toàn bộ mã nguồn**
            - Dịch nó thành **machine code** (hoặc đôi khi là **một ngôn ngữ trung gian**) phù hợp với kiến trúc của máy tính: **Inter x86, ARM**
            
            - Nếu **compiler** phát hiện: Lỗi cú pháp **syntax errors**, một số vấn đề logic nhất định - Thì nó sẽ **báo lỗi** và bạn phải **sửa lỗi trước khi quá trình biên dịch có thể hoàn tất**.

    3. **Liên kết (Linking)**: Sau khi biên dịch, compiler có thể tạo ra các **object file trung gian** - Sau đó một chương trình khác gọi là **linker** sẽ:
        - Kết hợp các **object file** này - liên kết chúng với các **thư viện cần thiết** - Đề tạo thành **một file thực thi duy nhất (executable file)**. Vd: `.exe` trên **Win**, **Binary file** trên **Linux/MacOS**.

    4. **Thực thi (Executing)**: Cuối cùng, chúng ta chạy **file thực thi** vừa tạo. Lúc này:
        - **CPU của máy tính** sẽ trực tiếp đọc
        - và thực thi **các lệnh mã máy** trong file đó.

- Ưu điểm chính của **biên dịch (compilation)** thường là **tốc độ thực thi**. Vì quá trình chuyển đổi **machine code** đã được thực hiện từ trước, nên khi chạy chương trình, nó có thẻ hoạt động rất nhanh. Ngoài ra, **compiler** cũng phát hiện nhiều **loại lỗi** trước khi bạn chạy chương trình.

- Tuy nhiên, phương pháp này cũng có một số nhược điểm:

    - Quá trình **biên dịch mất thời gian**.
    - Mã máy tạo ra thường **phụ thuộc vào nền tảng (platform-specific)**. Điều này có nghĩa là bạn cần **biên dịch lại chương trình** nếu muốn chạy nó trên **hệ điều hành hoặc kiến trúc CPU khác**.

### **Interpreted Languages: Translating Sentence by Sentence**

- Bây giờ, hãy xét một **ngôn ngữ thông dịch**. Cách này giống như có một **phiên dịch viên đồng thời** trong một cuộc họp quốc tế, người sẽ dịch **từng đoạn lời nói ngay khi nó được nói ra**:

    1. **Viết chương trình (Writing)**: Bạn viết chương trình bằng một ngôn ngữ cấp cao như: Python, JavaScript, Ruby... Đây là **mã nguồn (source code)** của bạn, ví dụ một tệp `.py`

    2. **Thực thi (Executing)**: Bạn chạy **source code** đó bằng một **interpreter (trình thông dịch)**.
        - Interpreter sẽ:

            - Đọc mã của bạn,
            - Thường là **từng dòng hoặc từng câu lệnh một**,
            - Chuyển phần đó thành lệnh máy (hoặc một dạng trung gian),
            - Rồi **thực thi ngay lập tức** trước khi chuyển sang phần tiếp theo

- Ưu điểm chính của ngôn ngữ thông dịch thường là **tính linh hoạt và sự thuận tiện trong quá trình phát triển**. Bạn không cần một bước biên dịch riêng. Thay vào đó, bạn có thể **viết mã và chạy trực tiếp**, giúp chu trình phát triển diễn ra nhanh hơn: `write, test, debug, repeat`.

- Ngôn ngữ thông dịch cũng thường **dễ di chuyển giữa các nền tảng (portable)**. Chỉ cần hệ thống đích có **interpreter phù hợp được cài đặt**, cùng một mã nguồn thường có thể **chạy mà không cần chỉnh sửa**.

- Tuy nhiên, cách này cũng có một số **đánh đổi**:
    - **Tốc độ thực thi** có thể chậm hơn so với ngôn ngữ biên dịch, vì quá trình dịch diễn ra **trong lúc chương trình đang chạy (runtime)**.

    - Một số lỗi (ví dụ: **lỗi kiểu dữ liệu** hoặc **tên biến chưa được định nghĩa**) có thể **chỉ được phát hiện khi interpreter chạy tới dòng mã chứa lỗi đó**, thay vì được phát hiện trước khi chạy chương trình.

### **Python's Approach: A Blend**

- Python thường được xem là một **ngôn ngữ thông dịch**, đó là lý do bạn chỉ cần gõ: `python your_script.py` và thấy chương trình chạy ngay.Tuy nhiên, phía sau hậu trường, triển khai Python phổ biến nhất gọi là **CPython** - lại sử dụng một bước trung gian để tăng hiệu quả:

    1. **Biên dịch sang Bytecode (Compilation to Bytecode)**: Khi bạn chạy một tệp Python `(.py)`, **CPython trước tiên biên dịch mã nguồn** của bạn thành **một định dạng cấp thấp hơn** và **không phụ thuộc nền tảng** gọi là **bytecode**
        - **Bytecode** này được lưu trong các tệp `.pyc` bên trong thư mục `__pycache__`.
        - **Cấp thấp hơn (Lower-level)**: Mã nguồn Python (.py) mà bạn viết là ngôn ngữ bậc cao, gần với ngôn ngữ con người. Bytecode là tập hợp các chỉ thị (instructions) đơn giản, rút gọn, gần với ngôn ngữ máy hơn nhưng vẫn chưa phải là mã máy (machine code) mà CPU có thể hiểu trực tiếp.

        - **Không phụ thuộc nền tảng (Platform-independent)**: Bytecode không được thiết kế cho một loại CPU hay hệ điều hành cụ thể nào (như Intel x86, ARM, Windows hay Linux). Thay vào đó, nó được thiết kế để chạy trên Python Virtual Machine (PVM). Vì PVM đóng vai trò là "lớp trung gian" giữa bytecode và phần cứng, nên cùng một file bytecode có thể chạy trên bất kỳ máy tính nào đã cài đặt Python
        
        - Quá trình biên dịch này:
            - Diễn ra **tự động**
            - Thường **bị ẩn khỏi người dùng**
            - Và giúp việc xử lý **bytecode** nhanh hơn so với việc thông dịch trực tiếp mã nguồn ban đầu.

        
    2. **Interpretation by PVM**: Sau đó, **bytecode** này được thực thi bởi **Python Virtual Machine (PVM)** — tức **bộ máy chạy (runtime engine)** của Python. **PVM** sẽ:
        - Đọc các lệnh bytecode
        - Thông dịch chúng
        - Và thực thi từng lệnh.
        
- Trong trường hợp chạy trực tiếp lần đầu, **bytecode** sẽ được tạo ra tự động. Nếu file `.pyc` đã tồn tại và còn hợp lệ, Python có thể sử dụng lại nó.

- Hãy tưởng tượng bạn có một dòng code Python đơn giản:
    1. **Mã nguồn `(.py)`**: Bạn viết `x = 1 + 2`.
    2. **Biên dịch sang Bytecode**: Khi bạn chạy, **CPython** dịch dòng này thành các chỉ thị **bytecode** (thường có đuôi `.pyc` trong thư mục `__pycache__`). Các chỉ thị này trông như sau:
        - `LOAD_CONST 1`
        - `LOAD_CONST 2`
        - `BINARY_ADD`
        - `STORE_NAME x`
    3. **Thực thi (PVM)**: Python Virtual Machine đọc các chỉ thị trên và dịch chúng thành mã máy tương ứng với CPU của máy tính bạn đang dùng (ví dụ: máy Windows Intel hoặc máy Mac Apple Silicon) để thực hiện phép tính.

### **Why This Matters for Beginners**

- Hiểu rằng Python về cơ bản hoạt động như **một ngôn ngữ thông dịch** giúp làm rõ nhiều lợi ích, đặc biệt khi bạn mới học lập trình:

    - **Simplicity**: Bạn chỉ cần viết mã rồi chạy. Ban đầu không cần quản lý một quy trình build phức tạp.

    - **Interactive Development**: Bộ thông dịch Python **(REPL)**, cho phép bạn nhập lệnh và xem kết quả ngay lập tức. Đây là hệ quả trực tiếp của việc Python là ngôn ngữ thông dịch.

    - **Readability and Debugging**: Các lỗi lúc chạy thường chỉ thẳng đến dòng mã nguồn nơi vấn đề xảy ra, nên việc debug trở nên dễ hơn.

    - **Cross-Platform**: Các script Python thường có thể chạy trên Windows, macOS và Linux mà không cần thay đổi, miễn là hệ thống đã cài Python interpreter.

## **#4: Using the Python Interpreter (REPL)**

- **Python interactive interpreter**, thường được gọi là **REPL**, cung cấp một cách trực tiếp để tương tác với Python. **REPL** là viết tắt của **Read–Eval–Print Loop**, và cái tên này mô tả đúng chức năng của nó: nó đọc mã Python bạn nhập vào, đánh giá và thực thi mã đó, in ra kết quả nếu có, rồi quay lại để chờ bạn nhập lệnh tiếp theo.

- Đây là một môi trường rất tốt để:
    - Thử nghiệm các đoạn mã nhỏ - Kiểm tra cú pháp - Hoặc khám phá khả năng của Python
    - Mà không cần phải tạo một file.

### **What is the REPL?**

- Bạn có thể xem REPL như một cuộc trò chuyện trực tiếp với Python interpreter. Bạn nhập một lệnh, nhấn Enter, và Python phản hồi ngay lập tức. Vòng phản hồi nhanh này rất hữu ích cho việc học và kiểm tra nhanh.
    - **Read**: đọc dòng mã bạn nhập vào.
    - **Eval**: đánh giá và thực thi mã đó.
    - **Print**: in kết quả ra màn hình.
    - **Loop**: chờ bạn nhập tiếp.

### **Starting the REPL**

- Thông thường, bạn khởi động Python REPL bằng cách mở **terminal** hoặc **command prompt**, sau đó gõ `python` hoặc đôi khi là `python3` tùy theo cách cài đặt và hệ điều hành, rồi nhấn Enter.

    - Trên Windows:
        1. Mở **Command Prompt** (tìm `cmd`) hoặc **PowerShell**.
        2. Gõ `python` rồi nhấn Enter.
    - Trên macOS:
        1. Mở **Terminal** (Applications > Utilities > Terminal hoặc tìm bằng Spotlight).
        2. Gõ `python3` rồi nhấn Enter.
        (Thường macOS dùng `python3` để phân biệt với Python 2 cũ của hệ thống.)
    - Trên Linux:
        1. Mở ứng dụng terminal.
        2. Gõ `python3` hoặc `python` nếu `python3` đã được liên kết với `python`, rồi nhấn Enter.
- Nếu mọi thứ được thiết lập đúng, bạn sẽ thấy kết quả giống như sau (phiên bản và chi tiết có thể khác):

```
Python 3.10.4 (main, Mar 31 2022, 08:41:55) [GCC 7.5.0] on linux
Type "help", "copyright", "credits" or "license" for more information.
>>>
```

- Dấu nhắc `>>>` là **Python prompt**. Nó cho biết interpreter đã sẵn sàng nhận lệnh của bạn.

### **Using the REPL**
- Hãy thử một vài lệnh đơn giản ngay tại dấu nhắc `>>>`

    1. **Basic Arithmetic**: Type a mathematical expression and press Enter. Python evaluates it immediately.
        ```
        >>> 2 + 3
        5
        >>> 100 - 5 * 10
        50
        >>> (50 + 50) * 2 / 4
        50.0
        ```
        - Bạn sẽ thấy Python tính kết quả và hiển thị nó ở dòng tiếp theo.
   
    2. `Print()`: Dùng để hiển thị kết quả ra màn hình.
        ```
        >>> print("Hello, Python!")
        Hello, Python!
        ```
   
    3. **Assigning Variables**: Bạn có thể tạo biến giống như khi viết trong script.

        ```
        >>> message = "Learning Python is fun"
        >>> print(message)
        Learning Python is fun
        >>> count = 10
        >>> count * 5
        50
        ```
        - Lưu ý rằng chỉ gán biến như `message = ...` thì sẽ không tạo ra output. Nhưng nếu bạn nhập chính tên biến đó, **REPL** thường sẽ in ra giá trị của nó:

        ```
        >>> message
        'Learning Python is fun'
        >>> count
        10
        ```

    4. **Multi-line Statements**: Với các cấu trúc như if hoặc vòng lặp, prompt sẽ chuyển thành `...` để cho biết Python đang chờ thêm input để hoàn tất khối lệnh. Khi muốn kết thúc khối, bạn nhấn Enter ở một dòng trống.

        ```
        >>> name = "Alice"
        >>> if name == "Alice":
        ...     print("Hello, Alice!")
        ... else:
        ...     print("Hello, stranger!")
        ...
        Hello, Alice!
        >>>
        ```
        

### **Exiting the REPL**
- Khi dùng xong trình thông dịch tương tác, bạn có thể thoát theo một trong các cách sau:

    - Gõ `exit()` rồi nhấn Enter,
    - Gõ `quit()` rồi nhấn Enter,
    - Nhấn `Ctrl+D` trên **Linux/macOS** Hoặc `Ctrl+Z` rồi nhấn Enter trên **Windows**.

- Sau đó, bạn sẽ quay lại terminal hoặc command prompt bình thường.

### **Why Use the REPL?**
- REPL đặc biệt hữu ích cho:

    - **Thử nhanh (Quick Experiments)**: kiểm tra những đoạn mã nhỏ hoặc cú pháp mà không cần tạo file.
    - **Khám phá đối tượng (Exploring Objects)**: xem các phương thức hoặc thuộc tính có sẵn của một biến hay đối tượng bằng các hàm như `dir()` và `help()`.

    - **Debugging**: chạy nhanh một hoặc hai dòng mã để hiểu cách nó hoạt động.
    - **Learning**: nhận phản hồi ngay lập tức về cú pháp và lệnh Python.

- Mặc dù bạn sẽ viết phần lớn chương trình trong các file script `(.py)`, REPL vẫn là một công cụ rất hữu ích trong quá trình học và phát triển Python. Nó giống như một **sandbox** để bạn thử mọi thứ.
