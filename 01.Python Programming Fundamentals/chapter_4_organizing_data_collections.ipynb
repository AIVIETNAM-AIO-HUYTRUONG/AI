# **Chapter 4: Organizing Data: Collections**

- Chương này giới thiệu các **built-in collection types (kiểu dữ liệu tập hợp dựng sẵn**) của Python, cung cấp cách hiệu quả để lưu trữ và quản lý nhiều item dữ liệu. Bạn sẽ học về:

- **Lists (danh sách)**: các sequence có thứ tự và có thể thay đổi, dùng để lưu collections mà các item có thể được sửa, thêm, hoặc xóa.

- **Sets (tập hợp)**: các collection không có thứ tự, chỉ lưu các item unique (duy nhất), rất hữu ích cho việc kiểm tra sự tồn tại và loại bỏ dữ liệu trùng lặp.

## **#1: Introduction to Sequences: Lists**

- Trong lập trình, chúng ta thường làm việc với **collections of items (tập hợp nhiều phần tử)**. Hãy tưởng tượng, bạn cần theo dõi tên học sinh trong một lớp, hay danh sách còn mua, điểm số cao nhất trong một trò chơi.

- Mặc dù bạn có thể tạo một variabie riêng cho từng item (`item1 = "milk"`, `item2 = eggs`), cách này sẽ nhanh chóng trở nên khó quản lý khi số lương items tăng lên.

- Python cung cấp một số **built-in data structures (cấu trúc dữ liệu dựng sẵn)**, thường được gọi là **collections**, được thiết kế dành riêng cho việc lưu trữ nhiều data. Một trong những câu trúc cơ bản và linh hoạt nhất **list**

- Hãy xem **list** như một **ordered sequence (dãy có thứ tự)** các item, giống như một shopping list (danh sách mua sắm) hoặc một numbered to-do list (danh sách việc cần làm có đánh số). Mỗi item trong list có một vị trí cụ thể.

- Điều quan trọng là list trong Python là **mutable (có thể thay đổi)**, nghĩa là bạn có thể sửa nội dung của nó sau khi tạo ra — bạn có thể thêm item mới, xóa item hiện có, hoặc thay đổi giá trị của item đã nằm trong list.

### **1. Creating a List**

- Để tạo list trong Python bằng cách đặt item phân tách bằng dấu (`,`) bên trong **square brackets (`[]`)**

```python
# An empty list
empty_list = []
print(empty_list)

# A list of integers
numbers = [1, 2, 3, 5, 8]
print(numbers)

# A list of strings
fruits = ["apple", "banana", "cherry"]
print(fruits)

# A list with mixed data types
mixed_list = [10, "hello", 3.14, True]
print(mixed_list)
```

- Như bạn thấy, list rất linh hoạt và có thể chứa các items thuộc nhiều kiểu dữ liệu khác nhau, bao gồm **numbers, strings, booleans** và **thậm chí cả các list khác (nested list)**

### **2. Accessing Items: Indexing**

- Vì list có thứ tự, bạn có thể truy cập từng item riêng lẻ bằng thứ gọi là **index (chỉ mục)**.

- Python dùng **zero-based indexing (đánh số bắt đầu từ 0)**, nghĩa là item đầu tiên ở index `0`, item thứ hai ở index `1`, và tiếp tục như vậy đến item cuối.

- Để truy cập được một item, bằng cần đặt **index vào trong `[]`**


```python
fruits = ["apple", "banana", "cherry", "date"]

# Access the first item (index 0)
first_fruit = fruits[0]
print(first_fruit)  # Output: apple

# Access the third item (index 2)
third_fruit = fruits[2]
print(third_fruit)  # Output: cherry
```

- Nếu bạn cố gắng truy cập một **index** không tồn tại, Python sẽ báo lỗi **IndexError**

### **3. Negative Indexing**

- Python cũng hỗ trợ việc **negative indexing (chỉ mục số âm)**, đây là một cách khác để truy cập item, nhưng nó sẽ tính từ cuối list. Index `-1` là item cuối cùng, `-2` là thứ 2 từ cuối tính lên, và cứ tiếp tục như vậy

```python
fruits = ["apple", "banana", "cherry", "date"]

# Access the last item
last_fruit = fruits[-1]
print(last_fruit)  # Output: date

# Access the second-to-last item
second_last_fruit = fruits[-2]
print(second_last_fruit) # Output: cherry
```

### 4. **Finding the Length of a List**

- Bạn có thể dễ dàng biết có bao nhiêu items trong list bằng cách dùng **built-in functio** đó là `len()`

```python
numbers = [10, 20, 30, 40, 50]
fruits = ["apple", "banana"]
empty_list = []

print(len(numbers))      # Output: 5
print(len(fruits))       # Output: 2
print(len(empty_list))   # Output: 0
```

### 5. Summary

- **List** là một **ordered sequence (dãy có thứ tự)** các item, và là **mutable (có thể thay đổi)**, cho phép thêm, xóa hoặc sửa đổi các item sau khi tạo.
- Bạn tạo một list bằng cách đặt các item phân tách bằng dấu phẩy trong **dấu ngoặc vuông (`[]`)**. List có thể chứa các kiểu dữ liệu khác nhau (số, chuỗi, boolean, thậm chí cả các list khác).
- Bạn truy cập các item trong list bằng **indexing (chỉ mục)**. Python sử dụng **zero-based indexing** (item đầu tiên là `0`).
- **Negative indexing** cho phép truy cập các item từ cuối list (ví dụ: `-1` cho item cuối cùng).
- Hàm `len()` tích hợp sẵn được dùng để tìm **độ dài (số lượng item)** của một list.

## **#2: Modifying Lists: Adding, Removing, Changing Items**

- Một trong những đặc điểm quan trọng của list là tính **mutable (có thể thay đổi)**. Khác với strings và tuples, các items trong list có thể thay đổi, thêm vào, xóa đi sau khi list được khởi tạo. Sự linh hoạt này khiến list trở thành công cụ hữu ích nhất để lưu các data collections có tính thay đổi khi chương trình chạy.

### **1. Changing an Item at a Specific Index**

- Nếu bạn biết index của item muốn thay đổi, bạn có thể gán trực tiếp một giá trị mới vào index đó.

- Hãy nhớ rằng list indices bắt đầu từ 0.

```python
# A list of colors
colors = ["red", "green", "blue"]
print(f"Original list: {colors}")

# Change the item at index 1 (the second item) from "green" to "yellow"
colors[1] = "yellow"
print(f"After changing index 1: {colors}")

# Change the last item using a negative index
colors[-1] = "purple"
print(f"After changing index -1: {colors}")
```

- Việc gán trực tiếp này sẽ thay thế giá trị cũ tại index được chỉ định bằng giá trị mới.

- Nếu bạn dùng một index không tồn tại trong list, Python sẽ báo lỗi **IndexError**.

### **2. Adding Items to a List**

- Python cung cấp một số method để thêm item mới vào list.

#### **2.1 Adding One Item to the End: `append()`**

- Cách phổ biến nhất để thêm một item là dùng `append()`.
- Method này sẽ thêm item mới vào vị trí cuối của list

```python
# A list of tasks for the day
tasks = ["email team", "review report"]
print(f"Tasks to do: {tasks}")

# Add a new task to the end
tasks.append("buy groceries")
print(f"After append: {tasks}")
```

#### **2.2 Adding One Item at a Specific Position: `insert()`**

- Nếu bạn cần thêm item ở một vị trí cụ thể không phải cuối list, hãy dùng `insert()`

- Method này nhận hai đối số:
  1. index nơi bạn muốn chèn item
  2. Là item muốn thêm

```python
# Insert a high-priority task at the beginning (index 0)
tasks.insert(0, "call client")
print(f"After insert at index 0: {tasks}")

# Insert another task before 'buy groceries' (which is now at index 3)
tasks.insert(3, "prepare meeting notes")
print(f"After insert at index 3: {tasks}")
```

- Việc dùng `insert()` có thể kém hiệu quả hơn `append()` đối với list lớn, vì việc dời các phần tử cần thời gian.

#### **2.3 Adding Multiple Items from Another Sequence: `extend()`**

- Để thêm toàn bộ items từ một **iterable (đối tượng có thể lặp)** khác, như list, tuple, hoặc string, vào cuối list hiện tại, hãy dùng `extend()`.

```python
# Add more tasks from another list
more_tasks = ["schedule follow-up", "submit expenses"]
tasks.extend(more_tasks)
print(f"After extend: {tasks}")
```

##### **2.3.1 `extend()` and `append()`**

- Nếu bạn dùng `append()` với một list, toàn bộ list đó sẽ được thêm vào như một phần tử duy nhất. Còn `extend()` sẽ thêm từng phần tử riêng lẻ của iterable đó.

```python
numbers = [1, 2, 3]
extra_numbers = [4, 5]

# Using extend (correct way to add individual items)
numbers.extend(extra_numbers)
print(f"After extend: {numbers}") # Output: [1, 2, 3, 4, 5]

# Reset numbers and try append
numbers = [1, 2, 3]
numbers.append(extra_numbers)
print(f"After append: {numbers}") # Output: [1, 2, 3, [4, 5]] - Note the nested list!
```

### **3. Removing Items from a List**

#### **3.1 Removing the First Occurrence of a Value: `remove()`**

- Nếu bạn biết value của item muốn xóa nhưng không nhất thiết biết index, hãy dùng `remove()`.
- Method này tìm trong list và xóa lần xuất hiện đầu tiên của value đó.
- Nếu value bạn muốn xóa không có trong list, Python sẽ báo lỗi **ValueError**.

```python
# Let's go back to our colors list
colors = ["red", "yellow", "purple", "yellow"]
print(f"Current colors: {colors}")

# Remove the first instance of "yellow"
colors.remove("yellow")
print(f"After remove('yellow'): {colors}")
```

#### **3.2 Removing an Item by Index: `pop()`**

- Nếu bạn muốn xóa item dựa trên index, hãy dùng `pop()`.

- Method này còn trả về item đã bị xóa, điều này rất hữu ích.

- Nếu bạn không chỉ định index, `pop()` sẽ xóa và trả về item cuối cùng trong list.

- Dùng `pop()` với một index không hợp lệ sẽ gây ra **IndexError**.

```python
colors = ["red", "yellow", "purple"]
print(f"Current colors: {colors}") # Should be ['red', 'purple', 'yellow']

# Remove and get the item at index 1 ("purple")
removed_color = colors.pop(1)
print(f"Removed color: {removed_color}")
print(f"After pop(1): {colors}")

# Remove and get the last item (default behavior)
last_color = colors.pop()
print(f"Removed last color: {last_color}")
print(f"After pop(): {colors}")
```

#### **3.3 Removing an Item by Index or Slice: `del` statement**

- `del` statement là cách tổng quát hơn để xóa item hoặc một slice của list bằng index.

- Khác với `pop()`, `del` không trả về giá trị đã bị xóa.

```python
numbers = [10, 20, 30, 40, 50, 60]
print(f"Original numbers: {numbers}")

# Remove the item at index 0
del numbers[0]
print(f"After del numbers[0]: {numbers}")

# Remove items from index 2 up to (but not including) index 4
# This removes the items currently at index 2 and 3 (30 and 40 originally, now 40 and 50)
del numbers[2:4]
print(f"After del numbers[2:4]: {numbers}")

```

```python
# del can also delete the entire list variable, but that's different from clearing it
del numbers
print(numbers) # This would now cause a NameError
```

#### **3.4 Removing All Items: `clear()`**

- Nếu bạn muốn xóa toàn bộ items khỏi list và biến nó thành list rỗng, hãy dùng `clear()`.

```python
numbers = [10, 60] # From the previous example
print(f"Numbers before clear: {numbers}")

numbers.clear()
print(f"Numbers after clear: {numbers}")
```

### 4. Summary

- List là **mutable (có thể thay đổi)**, nghĩa là bạn có thể sửa đổi nội dung của chúng (thêm, xóa, thay đổi item) sau khi khởi tạo.
- Bạn có thể **thay đổi một item** tại một index cụ thể bằng cách gán một giá trị mới (ví dụ: `my_list[index] = new_value`).
- Để **thêm item**:
  - `append(item)`: thêm một item vào cuối list.
  - `insert(index, item)`: thêm một item vào vị trí index cụ thể.
  - `extend(iterable)`: thêm tất cả các item từ một iterable (như list, tuple) vào cuối list.
- Để **xóa item**:
  - `remove(value)`: xóa lần xuất hiện đầu tiên của một value cụ thể. Nếu value không tồn tại sẽ báo `ValueError`.
  - `pop(index)`: xóa item tại index đã cho và trả về item đó. Nếu không có index, nó sẽ xóa và trả về item cuối cùng. Nếu index không hợp lệ sẽ báo `IndexError`.
  - `del my_list[index]` hoặc `del my_list[start:end]`: xóa item hoặc một slice của list bằng index hoặc slice. Không trả về giá trị đã xóa. Cũng có thể xóa toàn bộ biến list.
  - `clear()`: xóa tất cả các item khỏi list, biến nó thành một list rỗng.

## **#3: Immutable Sequences: Tuples**

- Tuples là một kiểu sequence built-in function, nhìn bề ngoài nó khá giống với list. Tuy nhiên, tuples là **immutable (bất biến)**. Điều này nghĩa là sau khi tuples được tạo, nội dung của nó **không thể bị thay đổi, thêm vào, hoặc xóa đi**

### **1. Creating Tuples**

- Bạn tạo tuple bằng dấu `()` thay vì dấu `[]` như list. Các items bên trong được tách bằng dấu phẩy.

```python
# An empty tuple
empty_tuple = ()
print(empty_tuple)

# A tuple with integers
coordinates = (10, 20)
print(coordinates)

# A tuple with mixed data types
person_info = ("Alice", 30, "New York")
print(person_info)

# Parentheses are sometimes optional if the context is clear
another_tuple = 100, 200, 300
print(another_tuple)
print(type(another_tuple)) # Output: <class 'tuple'>
```

#### **1.1 Creating a Single-Item Tuple**

- Khi tạo tuple chỉ có **một phần tử**, **bạn phải thêm dấu phẩy ở cuối**. Nếu không có dấu phẩy, Python sẽ hiểu dấu ngoặc chỉ dùng để nhóm biểu thức, chứ không phải tạo tuple.

```python
# This is NOT a tuple, it's just the integer 5
not_a_tuple = (5)
print(type(not_a_tuple)) # Output: <class 'int'>

# This IS a single-item tuple
single_item_tuple = (5,)
print(type(single_item_tuple)) # Output: <class 'tuple'>
print(single_item_tuple)
```

### **2. Accessing Tuple Elements**

- Truy cập phần tử trong tuple hoạt động giống hệt list. Bạn dùng dấu `[]` cùng index của phần tử muốn lấy.
- Hãy nhớ index bắt đầu từ `0`

```python
rgb_color = (255, 165, 0) # Orange color

# Access the first element (index 0)
red_value = rgb_color[0]
print(f"Red: {red_value}") # Output: Red: 255

# Access the third element (index 2)
blue_value = rgb_color[2]
print(f"Blue: {blue_value}") # Output: Blue: 0
```

#### **2.1 Slicing Tuples**

- Tuple cũng hỗ trợ slicing giống lists.

```python
weekdays = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")

# Get elements from index 1 up to (but not including) index 4
mid_week = weekdays[1:4]
print(mid_week) # Output: ('Tue', 'Wed', 'Thu')

# Get the last two elements
weekend = weekdays[-2:]
print(weekend) # Output: ('Sat', 'Sun')
```

### **3. Immutability: The Defining Feature**

- Đây là điểm khác biệt cốt lõi giữa tuple và list.

- Sau khi tạo tuple, bạn **không thể thay đổi items bên trong nó**. Nếu cố gắng thay đổi giá trị thì Python sẽ báo lỗi **TypeError**

- Và tuple không có các method như list: `append()` , `extend()` , `remove()` , `pop()`. Vì chúng không thể thay đổi


```python
my_tuple = (10, 20, 30)
print(my_tuple)

# Let's try to change the element at index 1
# This will cause an error!
my_tuple[1] = 25 # Uncommenting this line leads to TypeError

# Similarly, you cannot add or remove elements
# Methods like append(), extend(), remove(), pop() do not exist for tuples
```

### **4. Why is immutability useful**


#### **4.1 Data Integrity**


- Nếu dữ liệu không nên thay đổi trong suốt chương trình, tuple giúp đảm bảo điều đó.

- Ví dụ:
  - Tọa độ
  - RGB color
  - Cấu hình hệ thống
  - Dữ liệu cố định

```python
# Configuration data is protected by tuples.
server_config = ("192.168.1.1", 8080)
print(server_config)  # Output: ('192.168.1.1', 8080)
```

#### **4.2 Dictionary Keys**

- Dictionary yêu cầu keys phải immutable.

- Tuple là immutable nên có thể dùng làm key. List thì không.

```python
# Using a tuple as a dictionary key
location_temps = {
    (40.7128, -74.0060): 22,  # New York City coordinates -> temperature
    (34.0522, -118.2437): 28, # Los Angeles coordinates -> temperature
}

print(location_temps[(40.7128, -74.0060)]) # Output: 22
```

```python
# Trying to use a list as a key would cause a TypeError
location_temps[[34.0522, -118.2437]] = 28 # Error!
```

#### **4.3 Performance**

- Tính bất biến (immutability) và kích thước cố định của `tuple` cho phép trình thông dịch Python thực hiện các tối ưu hóa cấp thấp mà `list` không thể làm được:

  1. **Cấp phát bộ nhớ (Memory Allocation)**:
      - Khi tạo một `tuple`, Python biết chính xác kích thước cần thiết và cấp phát một khối bộ nhớ liền kề duy nhất.
      -  Ngược lại, `list` là các mảng động (dynamic arrays) yêu cầu thêm bộ nhớ dự phòng để phục vụ các thao tác append hoặc extend, dẫn đến lãng phí bộ nhớ và **chi phí tái cấp phát (reallocation)** khi mảng đầy.
  2. **Cơ chế lưu trữ tạm thời (Free List/Caching)**:
      - Python duy trì một danh sách các `tuple` nhỏ đã được giải phóng để tái sử dụng ngay lập tức cho các yêu cầu mới, giúp giảm áp lực lên **bộ quản lý bộ nhớ (memory allocator)**.
  3. **Tra cứu và truy cập**:
      - Do cấu trúc tĩnh, việc truy cập các phần tử trong `tuple` có độ trễ thấp hơn. Ngoài ra, vì `tuple` là hashable (nếu các phần tử bên trong đều hashable), nó có thể được sử dụng làm khóa (key) trong dict hoặc phần tử trong set, điều mà list không thể thực hiện do tính chất có thể thay đổi.

### **5. Summary**

- **Tuples** là các **sequence có thứ tự** giống như list, nhưng chúng **immutable (bất biến)**, nghĩa là sau khi được tạo, các item bên trong tuple không thể thay đổi, thêm vào hoặc xóa đi.
- Tuples được tạo bằng cách đặt các item phân tách bằng dấu phẩy trong **dấu ngoặc đơn (`()`)**.
  - Để tạo một tuple có một item, bạn phải thêm dấu phẩy sau item đó (ví dụ: `(5,)`).
- **Truy cập item** trong tuple tương tự như list, sử dụng **indexing** (bắt đầu từ 0) và **slicing**.
- Đặc điểm **bất biến** của tuple là khác biệt chính so với list:
  - Bạn **không thể gán lại giá trị** cho một index cụ thể (sẽ gây ra `TypeError`).
  - Tuples không có các method thay đổi nội dung như `append()`, `insert()`, `remove()`, `pop()`.
- **Tính bất biến hữu ích vì**:
  - **Đảm bảo tính toàn vẹn dữ liệu**: Khi dữ liệu không nên thay đổi (ví dụ: tọa độ, cấu hình).
  - **Dùng làm khóa dictionary**: Vì tuples là hashable (nếu các phần tử của nó hashable), chúng có thể dùng làm khóa trong dictionary, điều mà list không thể.
  - **Hiệu suất**: Python có thể tối ưu hóa bộ nhớ và truy cập cho tuples tốt hơn list do kích thước cố định và tính bất biến.

## **#4: Key Value Pairs: Dictionaries**

- Kiểu dữ liệu **Dict** (viết tắt của Dictionary) trong Python là một kiểu dữ liệu lưu trữ giống như **List** và **Tuple**.

- Dict có chức năng **lưu trữ các giá trị chứa key và value**, và các dữ liệu này **không được sắp xếp theo một trật tự nào cả**.

- Dictionaries rất hữu ích khi bạn cần truy xuất dữ liệu dựa trên một **identifier (định danh)** cụ thể thay vì dựa trên **index** trong một sequence.

### **1. Creating Dictionaries**

- Bạn tạo một dictionary bằng cách đặt các **key-value pairs** phân tách bởi dấu (`,`) bên trong `{}`.

- Mỗi **pair** gồm một `key`, theo sau là dấu `:`, rồi đến `value` tương ứng.

```python
# An empty dictionary
empty_dict = {}

# A dictionary storing student grades
student_grades = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 78
}

print(student_grades)
# Output: {'Alice': 85, 'Bob': 92, 'Charlie': 78}

# Keys can be different immutable types (strings, numbers, tuples)
mixed_keys = {
    "name": "Example Item",
    101: "Item ID",
    ("x", "y"): "Coordinates"
}
print(mixed_keys)
# Output: {'name': 'Example Item', 101: 'Item ID', ('x', 'y'): 'Coordinates'}
```

#### **1.1 Important Rules for Keys**

1. **Uniqueness (tính duy nhất)**: các keys trong một dictionary phải là duy nhất. Nếu bạn thêm một key đã tồn tại, value cũ gắn với key đó sẽ bị ghi đè.


```python
# Initialize dictionary
user_data = {
    "username": "apx_learner",
    "level": 1
}

# add new key
user_data["email"] = "learner@apx.ai"

# Update existing key (overwrite old value)
# "level" from 1 will be overwritten to 2
user_data["level"] = 2

print(user_data)
# output: {'username': 'apx_learner', 'level': 2, 'email': 'learner@apx.ai'}
```

2. **Immutability (tính bất biến)**:
- keys phải là kiểu dữ liệu bất biến. Điều đó có nghĩa là bạn có thể dùng strings, numbers (integers, floats), hoặc tuples làm keys, miễn là tuple đó chỉ chứa các phần tử bất biến.

- Bạn không thể dùng lists hoặc dictionaries khác làm keys vì chúng là **mutable (có thể thay đổi)**.

### **2. Accessing Values**

- Để truy cập `value` tương ứng với một `key`, bạn sẽ dùng đến `[]`, tương tự với list truy cập dựa vào `index`, nhưng ở dây chúng ta sẽ dùng `key` thay vì `index`.

```python
student_grades = {"Alice": 85, "Bob": 92, "Charlie": 78}

# Get Bob's grade
bobs_grade = student_grades["Bob"]
print(f"Bob's grade is: {bobs_grade}")
# Output: Bob's grade is: 92
```

- Nếu bạn truy cập một `key` không tồn tại trong dictionary, Python sẽ báo lỗi **KeyError**.

```python
# Trying to access a non-existent key causes an error
print(student_grades["David"]) # This would raise a KeyError
```

- Một cách an toàn hơn để lấy `value` nếu bạn không chắc `key` có tồn tại hay không là dùng method `get()`

- Method này trả về `None` (hoặc một giá trị mặc định do bạn chỉ định) nếu `key` không được tìm thấy, thay vì báo lỗi.

```python
student_grades = {"Alice": 85, "Bob": 92, "Charlie": 78}

# Get Alice's grade using get()
alice_grade = student_grades.get("Alice")
print(f"Alice's grade (using get): {alice_grade}")
# Output: Alice's grade (using get): 85

# Try to get David's grade (doesn't exist)
david_grade = student_grades.get("David")
print(f"David's grade (using get): {david_grade}")
# Output: David's grade (using get): None

# Provide a default value if the key is missing
eve_grade = student_grades.get("Eve", "Not Found")
print(f"Eve's grade (using get with default): {eve_grade}")
# Output: Eve's grade (using get with default): Not Found
```

### **3. Adding and Modifying Pairs**

- Việc thêm một **key-value pair** mới hoặc sửa `value` của một `key` đã có là rất đơn giản. Bạn dùng dấu `[]` ở bên trái của một phép gán.

```python
student_grades = {"Alice": 85, "Bob": 92}
print(f"Original grades: {student_grades}")

# Add a new student
student_grades["Charlie"] = 78
print(f"After adding Charlie: {student_grades}")

# Update Bob's grade
student_grades["Bob"] = 95
print(f"After updating Bob: {student_grades}")
```

- Nếu key `"Charlie"` chưa tồn tại, một pair mới sẽ được thêm vào.
- Nếu key `"Bob"` đã tồn tại, value tương ứng của nó sẽ được cập nhật.

### **4. Removing Pairs**

- Bạn có thể xóa **key-value pairs** bằng từ khóa `del` hoặc method `pop()`.

```python
student_grades = {"Alice": 85, "Bob": 95, "Charlie": 78}
print(f"Initial dictionary: {student_grades}")

# Remove Charlie using del
del student_grades["Charlie"]
print(f"After deleting Charlie: {student_grades}")

# Remove Bob using pop() and get the removed value
bobs_removed_grade = student_grades.pop("Bob")
print(f"Removed Bob's grade: {bobs_removed_grade}")
print(f"Dictionary after popping Bob: {student_grades}")
```

- `del` chỉ đơn giản là xóa item

- Method `pop()` vừa xóa item vừa trả về value của nó, điều này có thể rất hữu ích.

- Nếu bạn dùng `del` hoặc `pop()` với một key không tồn tại, bạn sẽ gặp lỗi **KeyError**

#### **4.1 Removing the Last Inserted Item: `popitem()`**

- Kể từ Python 3.7, `popitem()` được đảm bảo sẽ loại bỏ và trả về một cặp key-value theo thứ tự **LIFO (Last In, First Out)**, nghĩa là nó xóa item được thêm vào cuối cùng.
- Nếu dictionary rỗng, nó sẽ báo lỗi **KeyError**.

```python
student_grades = {"Alice": 85, "Bob": 95, "Charlie": 78}
print(f"Initial dictionary: {student_grades}")

# Remove the last inserted item using popitem()
removed_item = student_grades.popitem()
print(f"Removed item: {removed_item}") # Output: ('Charlie', 78)
print(f"Dictionary after popitem(): {student_grades}")

# Try to popitem from an empty dictionary (will cause an error)
empty_dict = {}
# empty_dict.popitem() # Uncommenting this line causes a KeyError
```

### **5. Common Dictionary Operations**

- Dictionaries đi kèm một số method rất hữu ích:

  - `keys()`: trả về một view object hiển thị danh sách tất cả keys.
  - `values()`: trả về một view object hiển thị danh sách tất cả values.
  - `items()`: trả về một view object hiển thị danh sách các cặp key-value dưới dạng tuple.
  - `len()`: không phải method mà là một **built-in function (hàm dựng sẵn)**, trả về **số lượng key-value pairs**.
  - `in`: một operator dùng để kiểm tra xem một key có tồn tại trong dictionary hay không.

```python
student_grades = {"Alice": 85, "Bob": 95, "Charlie": 78}

print(f"Keys: {student_grades.keys()}")
print(f"Values: {student_grades.values()}")
print(f"Items: {student_grades.items()}")
print(f"Number of students: {len(student_grades)}")

# Check if 'Alice' is in the dictionary
if "Alice" in student_grades:
    print("Alice is in the grades dictionary.")

# Check if 'David' is in the dictionary
if "David" not in student_grades:
    print("David is not in the grades dictionary.")
```

### **6. Summary**

- **Dictionaries** (Dict) là kiểu dữ liệu lưu trữ các **cặp key-value** và **không có thứ tự**. Chúng rất hữu ích khi cần truy xuất dữ liệu bằng một **định danh (key)** cụ thể.
- Bạn tạo một dictionary bằng cách đặt các cặp key-value phân tách bằng dấu phẩy bên trong **dấu ngoặc nhọn (`{}`)**. Mỗi cặp gồm một `key`, theo sau là dấu `:`, rồi đến `value`.
- **Quy tắc quan trọng cho Keys**:
  - **Duy nhất**: Keys phải là duy nhất. Nếu thêm một key đã tồn tại, value cũ sẽ bị ghi đè.
  - **Bất biến (Immutable)**: Keys phải là kiểu dữ liệu bất biến (strings, numbers, tuples chứa các phần tử bất biến). List hoặc dictionary khác không thể làm keys.

  - Dùng method `get(key, default_value)` để truy cập an toàn hơn; nó trả về `None` hoặc `default_value` nếu key không tồn tại.

    - Nếu dictionary rỗng, nó sẽ báo lỗi **KeyError**.
- **Các Thao tác Phổ biến**:
  - `keys()`: Trả về một view object của tất cả keys.
  - `values()`: Trả về một view object của tất cả values.
  - `items()`: Trả về một view object của tất cả các cặp key-value dưới dạng tuple.
  - `len(my_dict)`: Trả về số lượng cặp key-value trong dictionary.
  - Operator `in` hoặc `not in`: Dùng để kiểm tra sự tồn tại của một key trong dictionary.

## **#5: Unique Items: Sets**

- Kiểu **set (tập hợp)** là một collection, được thiết kế riêng để lưu các **unique items (phần tử duy nhất)**, trong đó thứ tự của các phần tử không quan trọng.

- Hãy nghĩ về **set** như một **mathematical set (tập hợp toán học)**:
  - Một tập hợp các đối tượng khác nhau. Nếu bạn cố thêm một item đã tồn tại vào trong **set**, sẽ không có gì xảy ra, **set** vẫn giữa nguyên vì nó chỉ lưu một bản sao của
  mỗi item duy nhất.

  - Tính chất này khiến **set** đặc biệt hữu ích cho các tác vụ như **removing duplicate (xóa phần tử trùng lặp)**

- Một đặc điểm quan trọng khác của **set** là chúng **unordered (không có thứ tự)**. Khác với **list, tuple**. **Set** không giữa các phân tử theo một **sequence** cố định nào. Điều đó có nghĩa là bạn không thể truy cập phần tử bằng **index** như `my_list[0]`, vì không có khái niệm phần tử “đầu tiên” hay “thứ hai” trong **set**.

### **1. Creating Sets**

- Bạn có thể tạo một set bằng cách đặt các phần tử phân tách bởi dấu phẩy bên trong **curly braces (dấu ngoặc nhọn) `{}`**

```python
# Creating a set with unique elements
unique_numbers = {1, 2, 3, 4, 5}
print(unique_numbers)
# Output might be {1, 2, 3, 4, 5} or some other order

# Creating a set with duplicates - duplicates are automatically removed
colors = {'red', 'green', 'blue', 'red'}
print(colors)
# Output: {'blue', 'red', 'green'} (order may vary)
```

> Lưu ý rằng `'red'` chỉ xuất hiện một lần trong **set** colors, dù ban đầu ta đã thêm nó hai lần.

#### **1.1 Create a empty set**

- - Để tạo một **empty set (set rỗng)**, bạn phải dùng hàm `set()` mà không truyền đối số nào.
- Nếu dùng `{}`, bạn sẽ tạo ra một **empty dictionary (dictionary rỗng)**.

```python
# Correct way to create an empty set
empty_s = set()
print(type(empty_s))
# Output: <class 'set'>
print(empty_s)
# Output: set()

# This creates an empty dictionary, NOT a set
empty_d = {}
print(type(empty_d))
# Output: <class 'dict'>
```

- Bạn cũng có thể tạo **set** từ bất kỳ **iterable object (đối tượng có thể lặp)** nào như list, tuple, hoặc string bằng hàm `set()`

- Đây là một kỹ thuật rất phổ biến để **xóa phần tử trùng lặp (remove duplicate items)** khỏi một sequence.

```python
# Create a set from a list with duplicates
numbers_list = [1, 2, 2, 3, 4, 4, 4, 5]
unique_numbers_set = set(numbers_list)
print(unique_numbers_set)
# Output: {1, 2, 3, 4, 5} (order may vary)

# Create a set from a string
letters_set = set('hello')
print(letters_set)
# Output: {'h', 'l', 'o', 'e'} (order may vary)
```

### **2. Modifying Sets**

- Giống như lists, sets là **mutable (có thể thay đổi)**, nghĩa là bạn có thể sửa nội dung của chúng sau khi đã tạo.

#### **2.1 Adding Elements**

- Để thêm một phần tử vào set, dùng method `.add()`.


```python
permissions = {'read', 'write'}
print(permissions)
# Output: {'write', 'read'} (order may vary)

# Add a new permission
permissions.add('execute')
print(permissions)
# Output: {'write', 'read', 'execute'} (order may vary)

# Try adding an existing element - no change
permissions.add('read')
print(permissions)
# Output: {'write', 'read', 'execute'} (order may vary)
```

#### **2.2 Removing Elements**

- Có hai cách chính để xóa phần tử:

  - **`.remove(element)`**: Xóa phần tử được chỉ định. Tuy nhiên, nếu phần tử không tồn tại trong **set**, nó sẽ báo lỗi **KeyError**.

  - **`.discard(element)`**: Cũng xóa phần tử được chỉ định, nhưng không báo lỗi nếu phần tử không tồn tại.

```python
data_points = {10, 20, 30, 40}

# Remove an existing element using remove()
data_points.remove(20)
print(data_points)
# Output: {10, 40, 30} (order may vary)


# Remove an existing element using discard()
data_points.discard(30)
print(data_points)
# Output: {10, 40} (order may vary)
```

```python
# Try removing a non-existent element using remove() - This will cause an error
data_points.remove(99) # Raises KeyError: 99

# Try removing a non-existent element using discard() - No error
data_points.discard(99)
print(data_points)
```

### **3. Checking for Membership**

- Một trong những thao tác hiệu quả nhất với **sets** là kiểm tra xem một phần tử có tồn tại hay không bằng từ khóa `in`.

```python
allowed_users = {'alice', 'bob', 'charlie'}

# Check if a user is allowed
print('bob' in allowed_users)
# Output: True

print('eve' in allowed_users)
# Output: False

# Check if a user is NOT allowed
print('david' not in allowed_users)
# Output: True
```

### **4. When to Use Sets**

- Sets là lựa chọn lý tưởng khi:

  - **Uniqueness is Required (cần tính duy nhất)**: bạn cần lưu collection mà không cho phép trùng lặp, hoặc cần tự động loại bỏ phần tử trùng.
  - **Membership Testing is Frequent (kiểm tra sự tồn tại thường xuyên)**: chương trình của bạn thường xuyên cần xem một item có nằm trong collection hay không. Sets cho tốc độ tra cứu rất nhanh.
  - **Order Doesn't Matter (thứ tự không quan trọng)**: sequence hay vị trí của phần tử không có ý nghĩa trong ứng dụng.
  - **Mathematical Set Operations Needed (cần các phép toán tập hợp)**: bạn muốn làm các thao tác như tìm phần tử chung **(intersection – giao)**, **gộp phần tử (union – hợp)**, hoặc tìm phần tử khác nhau giữa các collection. Chúng ta sẽ xem các phép toán này ở phần tiếp theo.

## **#6: Operations on Sets**

- **Sets** lưu các phần tử **unique (duy nhất)** và **unordered (không có thứ tự)**. Hãy xem một số phép toán mạnh mẽ bạn có thể thực hiện với chúng.

- Những phép toán này tương tự như **set theory (lý thuyết tập hợp)** trong toán học và rất hữu ích khi muốn so sánh hoặc kết hợp các collections dựa trên nội dung của chúng.

- Python cung cấp các **operators (toán tử)** trực quan và các methods tương ứng để thực hiện những thao tác này.

- Chúng ta sẽ dùng hai **set** ví dụ sau:

```python
programmers = {'Alice', 'Bob', 'Charlie', 'David'}
data_scientists = {'Charlie', 'David', 'Eve', 'Frank'}

print(f"Programmers: {programmers}")
print(f"Data Scientists: {data_scientists}")
```

### **1. Union Combining Sets**

- **Union (phép hợp)** của hai set tạo ra một set mới chứa tất cả các phần tử unique từ cả hai set gốc.

- Bạn có thể tính **union** bằng toán tử dấu `|` hoặc method `union()`.



```python
# Using the | operator
all_staff = programmers | data_scientists
print(f"Union (|): {all_staff}")

# Using the union() method
all_staff_method = programmers.union(data_scientists)
print(f"Union (method): {all_staff_method}")

# Note: The order doesn't matter for union
print(f"Is programmers | data_scientists == data_scientists | programmers? {programmers | data_scientists == data_scientists | programmers}")
```

- Cả hai cách đều cho cùng một kết quả: `{'Frank', 'Alice', 'Eve', 'Charlie', 'Bob', 'David'}`. Set này chứa tất cả những người là `programmer`, data `scientist`, hoặc cả hai.

### **2. Intersection: Finding Common Elements**

- **Intersection (phép giao)** của hai set tạo ra một set mới chỉ chứa các phần tử xuất hiện trong cả hai set gốc.

- Bạn có thể tính **intersection** bằng toán tử dấu `&` hoặc method `intersection()`.

```python
# Using the & operator
common_roles = programmers & data_scientists
print(f"Intersection (&): {common_roles}")

# Using the intersection() method
common_roles_method = programmers.intersection(data_scientists)
print(f"Intersection (method): {common_roles_method}")
```

- Kết quả trong cả hai trường hợp là `{'Charlie', 'David'}`, vì đây là những người xuất hiện trong cả set `programmers` và `data_scientists`.

### **3. Difference: Finding Unique Elements in One Set**

- **Difference (phép hiệu)** giữa hai set, giả sử **set A** và **set B**, tạo ra một set mới chứa các phần tử có trong A nhưng không có trong B. Điều quan trọng là thứ tự có ý nghĩa ở đây: A - B khác với B - A.

- Bạn có thể tính **difference** bằng toán tử trừ `-` hoặc method `difference()`.

```python
# Elements in programmers but NOT in data_scientists
only_programmers = programmers - data_scientists
print(f"Difference (programmers - data_scientists): {only_programmers}")

# Using the difference() method
only_programmers_method = programmers.difference(data_scientists)
print(f"Difference (method): {only_programmers_method}")

# Elements in data_scientists but NOT in programmers
only_data_scientists = data_scientists - programmers
print(f"Difference (data_scientists - programmers): {only_data_scientists}")
```

- Kết quả là:

  - `programmers` - `data_scientists`: `{'Alice', 'Bob'}`: Những programmer không đồng thời là data scientist
  - `data_scientists` - `programmers`: `{'Eve', 'Frank'}`: Nnhững data scientist không đồng thời là programmer

### **4. Symmetric Difference: Finding Elements in Exactly One Set**

- **Symmetric difference (phép hiệu đối xứng)** của hai set tạo ra một set mới chứa các phần tử có trong một trong hai set, nhưng không có trong cả hai.

- Về cơ bản, nó là phần ngược lại của **intersection**; nó bao gồm các phần tử duy nhất của từng set.

- Bạn có thể tính **symmetric difference** bằng toán tử dấu mũ `^` hoặc method `symmetric_difference()`

```python
# Using the ^ operator
exclusive_roles = programmers ^ data_scientists
print(f"Symmetric Difference (^): {exclusive_roles}")

# Using the symmetric_difference() method
exclusive_roles_method = programmers.symmetric_difference(data_scientists)
print(f"Symmetric Difference (method): {exclusive_roles_method}")
```

- Cả hai method đều cho kết quả `{'Frank', 'Alice', 'Eve', 'Bob'}`.

- Set này chứa những người chỉ là programmer hoặc chỉ là data scientist, không bao gồm những người làm cả hai vai trò.

### **5. Summary chi tiết các phép toán trên Set**

- **Sets** là cấu trúc dữ liệu mạnh mẽ cho phép thực hiện các phép toán logic và tập hợp tương tự như trong toán học. Các phép toán này rất hữu ích để xử lý, kết hợp, so sánh và lọc các tập hợp dữ liệu dựa trên nội dung của chúng.

- **1. Union (Phép hợp)**:
  - **Mục đích**: Kết hợp tất cả các phần tử duy nhất từ hai hoặc nhiều set thành một set mới. Nó trả về một set chứa mọi phần tử có mặt trong bất kỳ set nào đã cho.
  - **Toán tử**: `|`
  - **Method**: `.union()`
  - **Ví dụ khái niệm**: Nếu `Set A = {1, 2, 3}` và `Set B = {3, 4, 5}`, thì `A | B` (hoặc `A.union(B)`) sẽ là `{1, 2, 3, 4, 5}`. Phần tử trùng lặp (số 3) chỉ xuất hiện một lần.

- **2. Intersection (Phép giao)**:
  - **Mục đích**: Tìm kiếm các phần tử chung, tức là chỉ trả về những phần tử xuất hiện trong **tất cả** các set được so sánh.
  - **Toán tử**: `&`
  - **Method**: `.intersection()`
  - **Ví dụ khái niệm**: Nếu `Set A = {1, 2, 3}` và `Set B = {3, 4, 5}`, thì `A & B` (hoặc `A.intersection(B)`) sẽ là `{3}`. Đây là phần tử duy nhất có mặt trong cả hai set.

- **3. Difference (Phép hiệu)**:
  - **Mục đích**: Tìm các phần tử chỉ có trong set thứ nhất nhưng không có trong set thứ hai. Nó giúp xác định những yếu tố độc quyền của một set so với set khác.
  - **Toán tử**: `-`
  - **Method**: `.difference()`
  - **Lưu ý**: Thứ tự của các set trong phép hiệu là rất quan trọng. `A - B` không giống với `B - A`.
  - **Ví dụ khái niệm**: Nếu `Set A = {1, 2, 3}` và `Set B = {3, 4, 5}`,
    - `A - B` (hoặc `A.difference(B)`) sẽ là `{1, 2}` (các phần tử trong A không có trong B).
    - `B - A` (hoặc `B.difference(A)`) sẽ là `{4, 5}` (các phần tử trong B không có trong A).

- **4. Symmetric Difference (Phép hiệu đối xứng)**:
  - **Mục đích**: Trả về một set mới chứa tất cả các phần tử có mặt trong một trong hai set, nhưng **không phải trong cả hai**. Nó là tập hợp các phần tử độc quyền của mỗi set, loại bỏ đi những phần tử chung.
  - **Toán tử**: `^`
  - **Method**: `.symmetric_difference()`
  - **Ví dụ khái niệm**: Nếu `Set A = {1, 2, 3}` và `Set B = {3, 4, 5}`, thì `A ^ B` (hoặc `A.symmetric_difference(B)`) sẽ là `{1, 2, 4, 5}`. Đây là tổng hợp các phần tử độc nhất của A và B, không bao gồm phần tử chung là 3.
