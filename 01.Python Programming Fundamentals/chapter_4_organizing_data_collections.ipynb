- Chương này giới thiệu các **built-in collection types (kiểu dữ liệu tập hợp dựng sẵn**) của Python, cung cấp cách hiệu quả để lưu trữ và quản lý nhiều item dữ liệu. Bạn sẽ học về:

- **Lists (danh sách)**: các sequence có thứ tự và có thể thay đổi, dùng để lưu collections mà các item có thể được sửa, thêm, hoặc xóa.

- **Sets (tập hợp)**: các collection không có thứ tự, chỉ lưu các item unique (duy nhất), rất hữu ích cho việc kiểm tra sự tồn tại và loại bỏ dữ liệu trùng lặp.

## **#1: Introduction to Sequences: Lists**

- Trong lập trình, chúng ta thường làm việc với **collections of items (tập hợp nhiều phần tử)**. Hãy tưởng tượng, bạn cần theo dõi tên học sinh trong một lớp, hay danh sách còn mua, điểm số cao nhất trong một trò chơi.

- Mặc dù bạn có thể tạo một variabie riêng cho từng item (`item1 = "milk"`, `item2 = eggs`), cách này sẽ nhanh chóng trở nên khó quản lý khi số lương items tăng lên.

- Python cung cấp một số **built-in data structures (cấu trúc dữ liệu dựng sẵn)**, thường được gọi là **collections**, được thiết kế dành riêng cho việc lưu trữ nhiều data. Một trong những câu trúc cơ bản và linh hoạt nhất **list**

- Hãy xem **list** như một **ordered sequence (dãy có thứ tự)** các item, giống như một shopping list (danh sách mua sắm) hoặc một numbered to-do list (danh sách việc cần làm có đánh số). Mỗi item trong list có một vị trí cụ thể.

- Điều quan trọng là list trong Python là **mutable (có thể thay đổi)**, nghĩa là bạn có thể sửa nội dung của nó sau khi tạo ra — bạn có thể thêm item mới, xóa item hiện có, hoặc thay đổi giá trị của item đã nằm trong list.

### **1. Creating a List**

- Để tạo list trong Python bằng cách đặt item phân tách bằng dấu (`,`) bên trong **square brackets (`[]`)**

```python
# An empty list
empty_list = []
print(empty_list)

# A list of integers
numbers = [1, 2, 3, 5, 8]
print(numbers)

# A list of strings
fruits = ["apple", "banana", "cherry"]
print(fruits)

# A list with mixed data types
mixed_list = [10, "hello", 3.14, True]
print(mixed_list)
```

- Như bạn thấy, list rất linh hoạt và có thể chứa các items thuộc nhiều kiểu dữ liệu khác nhau, bao gồm **numbers, strings, booleans** và **thậm chí cả các list khác (nested list)**

### **2. Accessing Items: Indexing**

- Vì list có thứ tự, bạn có thể truy cập từng item riêng lẻ bằng thứ gọi là **index (chỉ mục)**.

- Python dùng **zero-based indexing (đánh số bắt đầu từ 0)**, nghĩa là item đầu tiên ở index `0`, item thứ hai ở index `1`, và tiếp tục như vậy đến item cuối.

- Để truy cập được một item, bằng cần đặt **index vào trong `[]`**


```python
fruits = ["apple", "banana", "cherry", "date"]

# Access the first item (index 0)
first_fruit = fruits[0]
print(first_fruit)  # Output: apple

# Access the third item (index 2)
third_fruit = fruits[2]
print(third_fruit)  # Output: cherry
```

- Nếu bạn cố gắng truy cập một **index** không tồn tại, Python sẽ báo lỗi **IndexError**

### **3. Negative Indexing**

- Python cũng hỗ trợ việc **negative indexing (chỉ mục số âm)**, đây là một cách khác để truy cập item, nhưng nó sẽ tính từ cuối list. Index `-1` là item cuối cùng, `-2` là thứ 2 từ cuối tính lên, và cứ tiếp tục như vậy

```python
fruits = ["apple", "banana", "cherry", "date"]

# Access the last item
last_fruit = fruits[-1]
print(last_fruit)  # Output: date

# Access the second-to-last item
second_last_fruit = fruits[-2]
print(second_last_fruit) # Output: cherry
```

### 4. **Finding the Length of a List**

- Bạn có thể dễ dàng biết có bao nhiêu items trong list bằng cách dùng **built-in functio** đó là `len()`

```python
numbers = [10, 20, 30, 40, 50]
fruits = ["apple", "banana"]
empty_list = []

print(len(numbers))      # Output: 5
print(len(fruits))       # Output: 2
print(len(empty_list))   # Output: 0
```

### 5. Summary

- **List** là một **ordered sequence (dãy có thứ tự)** các item, và là **mutable (có thể thay đổi)**, cho phép thêm, xóa hoặc sửa đổi các item sau khi tạo.
- Bạn tạo một list bằng cách đặt các item phân tách bằng dấu phẩy trong **dấu ngoặc vuông (`[]`)**. List có thể chứa các kiểu dữ liệu khác nhau (số, chuỗi, boolean, thậm chí cả các list khác).
- Bạn truy cập các item trong list bằng **indexing (chỉ mục)**. Python sử dụng **zero-based indexing** (item đầu tiên là `0`).
- **Negative indexing** cho phép truy cập các item từ cuối list (ví dụ: `-1` cho item cuối cùng).
- Hàm `len()` tích hợp sẵn được dùng để tìm **độ dài (số lượng item)** của một list.

## **#2: Modifying Lists: Adding, Removing, Changing Items**

- Một trong những đặc điểm quan trọng của list là tính **mutable (có thể thay đổi)**. Khác với strings và tuples, các items trong list có thể thay đổi, thêm vào, xóa đi sau khi list được khởi tạo. Sự linh hoạt này khiến list trở thành công cụ hữu ích nhất để lưu các data collections có tính thay đổi khi chương trình chạy.

### **1. Changing an Item at a Specific Index**

- Nếu bạn biết index của item muốn thay đổi, bạn có thể gán trực tiếp một giá trị mới vào index đó.

- Hãy nhớ rằng list indices bắt đầu từ 0.

```python
# A list of colors
colors = ["red", "green", "blue"]
print(f"Original list: {colors}")

# Change the item at index 1 (the second item) from "green" to "yellow"
colors[1] = "yellow"
print(f"After changing index 1: {colors}")

# Change the last item using a negative index
colors[-1] = "purple"
print(f"After changing index -1: {colors}")
```

- Việc gán trực tiếp này sẽ thay thế giá trị cũ tại index được chỉ định bằng giá trị mới.

- Nếu bạn dùng một index không tồn tại trong list, Python sẽ báo lỗi **IndexError**.

### **2. Adding Items to a List**

- Python cung cấp một số method để thêm item mới vào list.

#### **2.1 Adding One Item to the End: `append()`**

- Cách phổ biến nhất để thêm một item là dùng `append()`.
- Method này sẽ thêm item mới vào vị trí cuối của list

```python
# A list of tasks for the day
tasks = ["email team", "review report"]
print(f"Tasks to do: {tasks}")

# Add a new task to the end
tasks.append("buy groceries")
print(f"After append: {tasks}")
```

#### **2.2 Adding One Item at a Specific Position: `insert()`**

- Nếu bạn cần thêm item ở một vị trí cụ thể không phải cuối list, hãy dùng `insert()`

- Method này nhận hai đối số:
  1. index nơi bạn muốn chèn item
  2. Là item muốn thêm

```python
# Insert a high-priority task at the beginning (index 0)
tasks.insert(0, "call client")
print(f"After insert at index 0: {tasks}")

# Insert another task before 'buy groceries' (which is now at index 3)
tasks.insert(3, "prepare meeting notes")
print(f"After insert at index 3: {tasks}")
```

- Việc dùng `insert()` có thể kém hiệu quả hơn `append()` đối với list lớn, vì việc dời các phần tử cần thời gian.

#### **2.3 Adding Multiple Items from Another Sequence: `extend()`**

- Để thêm toàn bộ items từ một **iterable (đối tượng có thể lặp)** khác, như list, tuple, hoặc string, vào cuối list hiện tại, hãy dùng `extend()`.

```python
# Add more tasks from another list
more_tasks = ["schedule follow-up", "submit expenses"]
tasks.extend(more_tasks)
print(f"After extend: {tasks}")
```

##### **2.3.1 `extend()` and `append()`**

- Nếu bạn dùng `append()` với một list, toàn bộ list đó sẽ được thêm vào như một phần tử duy nhất. Còn `extend()` sẽ thêm từng phần tử riêng lẻ của iterable đó.

```python
numbers = [1, 2, 3]
extra_numbers = [4, 5]

# Using extend (correct way to add individual items)
numbers.extend(extra_numbers)
print(f"After extend: {numbers}") # Output: [1, 2, 3, 4, 5]

# Reset numbers and try append
numbers = [1, 2, 3]
numbers.append(extra_numbers)
print(f"After append: {numbers}") # Output: [1, 2, 3, [4, 5]] - Note the nested list!
```

### **3. Removing Items from a List**

#### **3.1 Removing the First Occurrence of a Value: `remove()`**

- Nếu bạn biết value của item muốn xóa nhưng không nhất thiết biết index, hãy dùng `remove()`.
- Method này tìm trong list và xóa lần xuất hiện đầu tiên của value đó.
- Nếu value bạn muốn xóa không có trong list, Python sẽ báo lỗi **ValueError**.

```python
# Let's go back to our colors list
colors = ["red", "yellow", "purple", "yellow"]
print(f"Current colors: {colors}")

# Remove the first instance of "yellow"
colors.remove("yellow")
print(f"After remove('yellow'): {colors}")
```

#### **3.2 Removing an Item by Index: `pop()`**

- Nếu bạn muốn xóa item dựa trên index, hãy dùng `pop()`.

- Method này còn trả về item đã bị xóa, điều này rất hữu ích.

- Nếu bạn không chỉ định index, `pop()` sẽ xóa và trả về item cuối cùng trong list.

- Dùng `pop()` với một index không hợp lệ sẽ gây ra **IndexError**.

```python
colors = ["red", "yellow", "purple"]
print(f"Current colors: {colors}") # Should be ['red', 'purple', 'yellow']

# Remove and get the item at index 1 ("purple")
removed_color = colors.pop(1)
print(f"Removed color: {removed_color}")
print(f"After pop(1): {colors}")

# Remove and get the last item (default behavior)
last_color = colors.pop()
print(f"Removed last color: {last_color}")
print(f"After pop(): {colors}")
```

#### **3.3 Removing an Item by Index or Slice: `del` statement**

- `del` statement là cách tổng quát hơn để xóa item hoặc một slice của list bằng index.

- Khác với `pop()`, `del` không trả về giá trị đã bị xóa.

```python
numbers = [10, 20, 30, 40, 50, 60]
print(f"Original numbers: {numbers}")

# Remove the item at index 0
del numbers[0]
print(f"After del numbers[0]: {numbers}")

# Remove items from index 2 up to (but not including) index 4
# This removes the items currently at index 2 and 3 (30 and 40 originally, now 40 and 50)
del numbers[2:4]
print(f"After del numbers[2:4]: {numbers}")

```

```python
# del can also delete the entire list variable, but that's different from clearing it
del numbers
print(numbers) # This would now cause a NameError
```

#### **3.4 Removing All Items: `clear()`**

- Nếu bạn muốn xóa toàn bộ items khỏi list và biến nó thành list rỗng, hãy dùng `clear()`.

```python
numbers = [10, 60] # From the previous example
print(f"Numbers before clear: {numbers}")

numbers.clear()
print(f"Numbers after clear: {numbers}")
```

### 4. Summary

- List là **mutable (có thể thay đổi)**, nghĩa là bạn có thể sửa đổi nội dung của chúng (thêm, xóa, thay đổi item) sau khi khởi tạo.
- Bạn có thể **thay đổi một item** tại một index cụ thể bằng cách gán một giá trị mới (ví dụ: `my_list[index] = new_value`).
- Để **thêm item**:
  - `append(item)`: thêm một item vào cuối list.
  - `insert(index, item)`: thêm một item vào vị trí index cụ thể.
  - `extend(iterable)`: thêm tất cả các item từ một iterable (như list, tuple) vào cuối list.
- Để **xóa item**:
  - `remove(value)`: xóa lần xuất hiện đầu tiên của một value cụ thể. Nếu value không tồn tại sẽ báo `ValueError`.
  - `pop(index)`: xóa item tại index đã cho và trả về item đó. Nếu không có index, nó sẽ xóa và trả về item cuối cùng. Nếu index không hợp lệ sẽ báo `IndexError`.
  - `del my_list[index]` hoặc `del my_list[start:end]`: xóa item hoặc một slice của list bằng index hoặc slice. Không trả về giá trị đã xóa. Cũng có thể xóa toàn bộ biến list.
  - `clear()`: xóa tất cả các item khỏi list, biến nó thành một list rỗng.
