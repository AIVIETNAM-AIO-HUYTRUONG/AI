## **#1: Storing Information: Variables**

- Các chương trình luôn làm việc với những mẫu thông tin như **numbers, text hoặc kết quả trả về từ các phép tính**. Thay vì sử dụng trực tiếp **dữ liệu thô (raw data)** cho chương trình, cách tốt hơn đó là **đặt tên cho data đó và tham chiếu đến nó thông qua tên**. Đây chính là vai trò **variable (variable)**

- Một **variable** giống như **một hộp có nhãn** hoặc **một vị trí có tên trong bộ nhớ máy tính**, nơi bạn có thể lưu trữ một giá trị.

### **What is a Variable?**

- Trong Python, **variable** là một **tên ký hiệu (symbolic name)** đại diện cho một giá trị. Bạn sử dụng **variable** để **lưu trữ data**, sau đó có thể **tham chiếu và xử lý data đó** trong chương trình

    > `age = 25`: `age` chính là **symbolic name**, tên tượng trưng mà bạn đặt cho địa chỉ bộ nhớ `0x001`.
- Thuật ngữ **variable** ngụ ý rằng **giá trị được lưu trữ có thể thay đổi** trong quá trình chương trình chạy.

- Hãy tưởng tượng bạn đang sắp xếp đồ vào các hộp. Bạn sẽ không đặt đồ vào các hộp không có nhãn. Thay vào đó, bạn sẽ dán nhãn như: `"Quần áo mùa đông"`,`"Sách"`,`"Dụng cụ nhà bếp"`.

- **Variable** trong lập trình cũng tương tự như vậy: **chúng cung cấp những cái tên có ý nghĩa cho các vị trí trong bộ nhớ chứa data**.

### **Creating Variables: The Assignment Statement**

- Tạo **variable** trong Python rất đơn giản bằng cách sử dụng **toán tử gán `(=)`**.
    ```python
    variable_name = value
    ```
    - cách hoạt động: Bạn chọn một tên cho **variable** của mình - Sử dụng `=` - Cung cấp giá trị cần lưu.


````python
message = "Hello, Python Learner!"
count = 10
price = 49.99
is_active = True
````

- Python là **ngôn ngữ định kiểu động (dynamically typed)**. Điều này có nghĩa là bạn **không cần khai báo kiểu dữ liệu trước**. Python sẽ tự động xác định kiểu dựa trên giá trị bạn gán.

````python
message = "Hello, Python Learner!"
````

### **Using Variables**

- Sau khi tạo **variable** bạn có thể **sử dụng tên của variable** để truy cập giá trị mà nó lưu trữ.

````python
message = ""
print(message) # Output:

items = 5
cost_per_item = 2.50
total_cost = items * cost_per_item # Use variables in calculations
print(total_cost) # Output: 12.5
````

### **Naming Variables: Rules and Conventions**

- Việc chọn tên **variable** tốt rất quan trọng để viết mã rõ ràng. Trong Python, việc đặt tên **variable** không chỉ cần tuân thủ các quy tắc bắt buộc (để mã chạy được) mà còn cần tuân theo các quy ước (để mã dễ đọc và chuyên nghiệp)

- **Rules**:
    - Tên **variable** phải bắt đầu bằng chữ cái `(a-z, A-Z)` hoặc dấu gạch dưới `(_)`:
        - `user_name` = "Alice"`,
        - `Total_Score = 100`,
        - `_internal_data = "secret"`
    - Phần còn lại của tên có thể gồm **chữ cái, số (0-9) và dấu gạch dưới**.

    - **phân biệt chữ hoa chữ thường. `myVariable, myvariable` và `MYVARIABLE`** là ba **variable khác nhau.

    - Tên **variable** không được trùng với **từ khóa của Python**. Từ khóa là những từ dành riêng, có ý nghĩa đặc biệt trong Python, như `if, else, for, while, def, class, True, False, None`.
- **Conventions**:
    
    - Dùng tên mô tả rõ mục đích của biến, ví dụ `user_name` thay vì `un`, `total_price` thay vì `tp`.
    
    - Dùng kiểu `snake_case` cho tên biến nhiều từ. Nghĩa là tất cả chữ đều viết thường, các từ được nối bằng dấu gạch dưới, ví dụ `first_name, number_of_items`.

    - Tránh dùng các tên một ký tự như `l, O` hoặc `I` vì chúng dễ bị nhầm với số `1` và `0`. Những tên ngắn như `x, y`, hoặc `i` thường vẫn chấp nhận được trong một số trường hợp như biến đếm trong vòng lặp hoặc tọa độ toán học, nhưng nhìn chung tên mô tả vẫn tốt hơn.
- Một số ví dụ về tên biến hợp lệ và không hợp lệ:

    - **Valid**: `name, age, user_input, _internal_counter, price_2024`
    - **Invalid**: `2nd_place` (bắt đầu bằng số), `user-name` (có dấu gạch ngang), `class (là từ khóa)`

### **Changing Variable Values (Reassignment)**

- Bạn có thể thay đổi giá trị đang lưu trong variable bằng cách gán một giá trị mới cho nó bằng dấu `=` một lần nữa.

````python
# Assign an initial value
score = 100
print("Initial score:", score) # Output: Initial score: 100

# Update the value
score = 150
print("Updated score:", score) # Output: Updated score: 150

# Assign a value based on the current value
score = score + 20 # Read the current value (150), add 20, store the result (170) back
print("Final score:", score)   # Output: Final score: 170
````

- Mỗi lần bạn gán giá trị mới, giá trị cũ của **variable** đó sẽ bị thay thế.

- **variable** là một trong những khối xây dựng cơ bản nhất trong Python. Chúng giúp bạn đặt tên và lưu dữ liệu, làm cho chương trình linh hoạt hơn và dễ hiểu hơn

## **#2: Fundamental Data Types: Numbers (Integers, Floats)**

- Cốt lõi của mọi phép tính là khả năng làm việc với số. Dù bạn đang đếm số lượng, tính khoảng cách, hay đo các đại lượng vật lý, thì số luôn là thành phần không thể thiếu. Python cung cấp nhiều kiểu dữ liệu số được tích hợp sẵn, nhưng hai kiểu cơ bản và được dùng thường xuyên nhất là
    - **Số nguyên (integer)**
    - Và **số thực (floating-point number)**.

### **Integers (int)**


- **Integers (int)** là các **số không có phần thập phân**. Chúng có thể là **số dương**, **số âm** hoặc là **số 0**.

- Trong Python, bạn tạo một số nguyên đơn giản bằng cách **viết số đó mà không có dấu chấm thập phân**


````python
# Examples of integers
user_count = 150
temperature_celsius = -5
year = 2024
zero_value = 0

print(user_count)
print(temperature_celsius)
````

- Trong Python, **integers** có thể lớn tùy ý, chỉ bị **giới hạn bởi bộ nhớ của hệ thống**. Bạn không cần lo lắng về các kiểu kích thước **integers** như `short, int`, hay `long` như trong một số ngôn ngữ lập trình khác.

- Bạn có thể kiểm tra kiểu dữ liệu của một giá trị bằng hàm `type()`:


````python
age = 30
print(type(age))  # Output: <class 'int'>
````

### **Floating-Point Numbers (float)**

- **Float** dùng để biểu diễn **các số thực**, bao gồm cả những **số có phần thập phân**.

- Chúng rất cần thiết trong làm việc với:
    - Các phép **đo phần trăm**
    - Các giá trị không phải **integers**
- Chúng ta tạo **float** bằng cách:
    - Thêm `.` hoặc dùng **ký hiệu khoa học (scientific notation)**

````python
# Examples of floating-point numbers
pi_approx = 3.14159
price = 49.99
temperature_fahrenheit = -20.5
account_balance = 0.0 # Using a decimal makes it a float
distance_meters = 1.5e3 # Scientific notation for 1.5 * 10^3, which is 1500.0

print(pi_approx)
print(distance_meters)
````

- Ngay cả khi một số trông giống **integers**, chỉ cần có **dấu `.` thập phân** thì Python sẽ xem nó là **float**:

````python
whole_number_float = 10.0
print(whole_number_float)      # Output: 10.0
print(type(whole_number_float)) # Output: <class 'float'>
````

### **Choosing Between int and float**
- Dùng `int` khi bạn cần các số nguyên: số đếm, chỉ số, mã định danh.
- Dùng `float` khi làm việc với phép đo, phân số, xác suất, hoặc khi kết quả tính toán có thể.

#### **0.1 + 0.2 = ?**

-  Trong máy tính, phép tính với **float không phải lúc nào cũng chính xác tuyệt đối**. Điều này xảy ra vì các **float** được lưu trữ bằng **dạng nhị phân (binary)** theo chuẩn như **IEEE 754**.

````python
0.1 + 0.2
````

- Kết quả là `0.30000000000000004` thay vì đúng `0.3`

- Trong hầu hết ứng dụng, sai số nhỏ như này nhưng ảnh hưởng đáng kể. Tuy nhiên, đối với các lĩnh vực **cần độ chính xác cao** như: tài chính, mô phỏng khoa học. Python cung cấp các công cụ khác như **Decimal** để xử lý chính xác hơn.

## **#3: Fundamental Data Types: Text (Strings)**

- **String** là **một chuỗi ký tự có thứ tự (ordered sequence of characters)**. Bất cứ thứ gì bạn đặt trong dấu ngoặc kép — chữ cái, số, ký hiệu, khoảng trắng — đều trở thành một **string**.

### **Creating Strings**

- Python khá linh hoạt trong cách định nghĩa string. Bạn có thể dùng **single quotes (')** hoặc **double quotes (")**. Sự linh hoạt này rất hữu ích khi bản thân string chứa dấu ngoặc.

````python
# Using single quotes
message_single = 'Hello, Python learner!'
print(message_single)

# Using double quotes
message_double = "This also works just fine."
print(message_double)

# What if your string contains a single quote? Use double quotes to enclose it.
quote = "He said, 'Python is fun!'"
print(quote)

# What if your string contains a double quote? Use single quotes.
reply = 'She responded, "Indeed it is."'
print(reply)
````

- Nếu bạn cần một **string nhiều dòng** hoặc chứa cả hai loại dấu ngoặc mà không muốn xử lý phức tạp, Python cung cấp **triple quotes: `'''` hoặc `"""`**.

````python
# Triple double quotes cho multi-line string
multi_line_doc = """This is the first line.
This is the second line.
And "quotes" or 'apostrophes' can be used freely inside."""
print(multi_line_doc)

# Triple single quotes cũng tương tự
multi_line_alt = '''Another way to write
multi-line text,
very convenient.'''
print(multi_line_alt)
````

### **Basic String Operations**

- Bạn có thể kết hợp string bằng toán tử `+`. Phép toán này gọi là **concatenation**.

- Bạn cũng có thể **lặp lại string** bằng toán tử `*`.

````python
first_name = "Ada"
last_name = "Lovelace"
space = " "

# Concatenation
full_name = first_name + space + last_name
print(full_name) # Output: Ada Lovelace

# Repetition
separator = "-" * 10 # Repeat the hyphen 10 times
print(separator)   # Output: ----------
print(first_name * 3) # Output: AdaAdaAda
````

> **Lưu ý**:
    - bạn chỉ có thể nối **string** với **string**. Nếu cố cộng trực tiếp một chuỗi với một số, chương trình sẽ báo lỗi.
    - Khi đó, bạn cần chuyển số sang **string** trước bằng `str()`.

### **Accessing Characters: Indexing**

- Vì string là **sequence**, bạn có thể **truy cập từng ký tự bằng index**.
- Python sử dụng **zero-based indexing**, điều này có nghĩa là ký tự đầu tiên có **index `0`** - Thứ hai là **index `1`**

````python
language = "Python"

# Accessing characters by positive index
print(language[0]) # Output: P (the first character)
print(language[1]) # Output: y (the second character)
print(language[5]) # Output: n (the sixth character)

# Accessing characters using negative index (starts from the end)
print(language[-1]) # Output: n (the last character)
print(language[-2]) # Output: o (the second-to-last character)
````

- Nếu truy cập **index** không tồn tại (ví dụ `language[6]`) → Python sẽ báo `IndexError`

### **Extracting Substrings: Slicing**
- Nếu bạn cần nhiều ký tự, bạn có thể lấy **substring** hoặc **slice** bằng slicing.
. Cú pháp `[start:stop:step]`
    - `start` (Bắt đầu): Vị trí chỉ mục (index) nơi bạn bắt đầu cắt - Mặc định: Nếu bỏ qua, nó sẽ bắt đầu từ 0 (phần tử đầu tiên).
    - `stop` (Kết thúc): Vị trí chỉ mục nơi bạn dừng cắt (không bao gồm phần tử ở vị trí này). - Mặc định: Nếu bỏ qua, nó sẽ cắt đến cuối chuỗi.
    - `step` (Bước nhảy): Khoảng cách giữa các phần tử được lấy.- Mặc định: Nếu bỏ qua, giá trị là 1 (lấy lần lượt từng phần tử).

````python
language = "Python"

# Get characters from index 1 up to (but not including) index 4
print(language[1:4]) # Output: yth

# Get characters from the beginning up to index 3 (exclusive)
print(language[:3])  # Output: Pyt

# Get characters from index 2 to the end
print(language[2:])  # Output: thon

# Get the whole string (less common, but shows the defaults)
print(language[:])   # Output: Python

# Get every second character
print(language[::2]) # Output: Pto

# Get the string in reverse
print(language[::-1]) # Output: nohtyP
````

- **Slicing luôn tạo string mới** và **không gây IndexError**, ngay cả khi index vượt phạm vi.

### **String Immutability**

- String trong Python là **immutable**. Điều này có nghĩa là **sau khi tạo, string không thể thay đổi trực tiếp**

````python
greeting = "hello"
greeting[0] = "H" # This will cause a TypeError: 'str' object does not support item assignment
````

- Để **thay đổi string**, bạn cần tạo một **string mới**.

````python
# To "change" the greeting, create a new string
new_greeting = "H" + greeting[1:]
print(new_greeting) # Output: Hello
print(greeting)     # Output: hello (original string is unchanged)
````

- Tính **immutability** giúp string an toàn và ổn định, ví dụ khi dùng làm **dictionary key**.

### **Common String Methods**

- String có nhiều **method** tích hợp sẵn để thực hiện các thao tác thường gặp. **Method** giống như function, nhưng được gọi trên một object (ở đây là string) bằng dấu chấm `.`. Dưới đây là một số **method** hữu ích:

    - `len(string)`: Mặc dù về mặt kỹ thuật là **một hàm built-in chứ không phải method**, `len()` rất quan trọng để lấy số lượng ký tự trong string.
    - `.lower() / .upper()`: Trả về string mới với toàn bộ ký tự được chuyển sang chữ thường hoặc chữ hoa.
    - `.strip() / .lstrip() / .rstrip()`: Trả về string mới đã loại bỏ khoảng trắng ở đầu và/hoặc cuối.
    - `.find(substring)`: Trả về **index** bắt đầu của lần xuất hiện đầu tiên của **substring**. Trả về `-1` nếu không tìm thấy.
    - `.replace(old, new)`: Trả về string mới trong đó tất cả **old** được thay bằng **new**.
    - `.startswith(prefix) / .endswith(suffix)`: Trả về `True` hoặc `False` tùy string có bắt đầu hoặc kết thúc bằng chuỗi đã cho hay không.
    - `.split(separator)`: Trả về **một list các substring**, tách ra tại vị trí **separator**. Nếu không truyền **separator**, nó tách theo khoảng trắng.
    - `separator.join(iterable)`: Ghép các phần tử của một **iterable (như list) thành một string**, với **separator** đặt giữa các phần tử.

````python
text = "  Learning Python is Fun!  "

# Length
print(len(text)) # Output: 27

# Case conversion
print(text.lower()) # Output:   learning python is fun!
print(text.upper()) # Output:   LEARNING PYTHON IS FUN!

# Stripping whitespace
print(text.strip())  # Output: Learning Python is Fun!
print(text.lstrip()) # Output: Learning Python is Fun!
print(text.rstrip()) # Output:   Learning Python is Fun!

# Finding substrings
print(text.find("Python")) # Output: 11 (index where 'Python' starts)
print(text.find("Java"))   # Output: -1 (not found)

# Replacing substrings
print(text.replace("Fun", "Awesome")) # Output:   Learning Python is Awesome!

# Checking start/end (use strip first for accuracy here)
clean_text = text.strip()
print(clean_text.startswith("Learning")) # Output: True
print(clean_text.endswith("!"))        # Output: True

# Splitting a string into a list
words = clean_text.split(" ")
print(words) # Output: ['Learning', 'Python', 'is', 'Fun!']

# Joining a list into a string
joined_string = "---".join(words)
print(joined_string) # Output: Learning---Python---is---Fun!

# Original string remains unchanged due to immutability
print(text) # Output:   Learning Python is Fun!
````

### **Formatting Strings**

- Nhiều lúc bạn cần tạo string có chèn giá trị của biến vào bên trong. Bạn có thể dùng phép nối +, nhưng cách này nhanh chóng trở nên rườm rà và dễ lỗi, đặc biệt khi làm việc với kiểu dữ liệu không phải string. Python có các cách định dạng string tốt hơn

- Cách hiện đại và thường được ưu tiên nhất là dùng **f-string (formatted string literals)**, được giới thiệu từ Python 3.6. Bạn tạo **f-string** bằng cách thêm tiền tố `f` hoặc `F` trước chuỗi. Bên trong chuỗi, bạn có thể đặt biến hoặc biểu thức trực tiếp trong dấu ngoặc nhọn `{}`.

````python
name = "Alice"
age = 30
city = "New York"

# Using f-string for clean formatting
intro_fstring = f"My name is {name}, I am {age} years old, and I live in {city}."
print(intro_fstring)
# Output: My name is Alice, I am 30 years old, and I live in New York.

# You can also include expressions inside the braces
radius = 5
area = 3.14159 * radius**2
print(f"A circle with radius {radius} has an area of {area:.2f}.")
# Output: A circle with radius 5 has an area of 31.42.
# Note: :.2f formats the float to 2 decimal places.
````

- **F-string** dễ đọc, ngắn gọn, và thường nhanh hơn các cách định dạng cũ như `.format()` hoặc toán tử kiểu C `%`, những cách này bạn vẫn có thể gặp trong code cũ.

#### **Escape Sequences**

- Đôi khi bạn cần chèn vào string những ký tự khó hoặc không thể gõ trực tiếp, hoặc những ký tự có ý nghĩa đặc biệt.

- Ví dụ: Làm sao để chèn xuống dòng, tab, hoặc một dấu nháy thật sự bên trong chuỗi dùng cùng loại dấu nháy để bao ngoài? Việc này được thực hiện bằng **escape sequences**, bắt đầu bằng dấu gạch chéo ngược `\`:

    - `\n`: Xuống dòng
    - `\t`: Tab
    - `\\`: Dấu gạch chéo ngược thật
    - `\'`: Dấu nháy đơn thật
    - `\"`: Dấu nháy kép thật

````python
# Newline
print("First line.\nSecond line.")
# Output:
# First line.
# Second line.

# Tab
print("Column1\tColumn2")
# Output: Column1    Column2

# Literal backslash
print("This is a path: C:\\Users\\Name")
# Output: This is a path: C:\Users\Name

# Literal quotes within string defined by same quotes
print('He said, \'Hello!\'') # Output: He said, 'Hello!'
print("She replied, \"Hi!\"") # Output: She replied, "Hi!"
````

- **String** là nền tảng để xử lý mọi loại dữ liệu văn bản, từ input của người dùng, nội dung file, đến tin nhắn và nhãn trong chương trình.

## **#4: Fundamental Data Types: Booleans**

### **The Boolean Type: `True` and `False`**

- Kiểu **boolean** được đặt theo tên của nhà toán học **George Boole**, người phát triển **Boolean algebra**. Kiểu dữ liệu này chỉ có hai giá trị: `True` và `False`

    > Lưu ý về **chữ hoa**. `True` và `False` là **từ khóa (reserved keywords)** trong Python, đại diện cho hai giá trị boolean. Bạn không thể dùng `true` hoặc `false` (viết thường).

````python
is_learning = True
is_finished = False

print(is_learning)
print(is_finished)
````

- Các biến `is_learning` và `is_finished` hiện đang lưu **giá trị boolean**. Chúng hoạt động giống như **các cờ (flags)** hoặc **công tắc (switches)** đơn giản trong chương trình.

### **Booleans from Comparisons**

- Mặc dù bạn có thể gán trực tiếp `True` hoặc `False`, nhưng trong **thực tế giá trị boolean thường xuất hiện từ các phép so sánh**.

- Khi sử dụng **comparison operators**:
    - `==` (bằng)
    - `!=` (không bằng)
    - `>` (lớn hơn)
    - `<` (nhỏ hơn)
    - `>=` (lớn hơn hoặc bằng)
    - `<=` (nhỏ hơn hoặc bằng)
- Kết quả của phép so sánh **luôn luôn là một giá trị boolean**.

````python
age = 20
is_adult = age >= 18  # Is age greater than or equal to 18?
print(is_adult)      # Output: True

temperature = 15.5
is_cold = temperature < 10.0 # Is temperature less than 10.0?
print(is_cold)       # Output: False

name = "Alice"
is_bob = name == "Bob" # Is the name equal to "Bob"?
print(is_bob)          # Output: False
````

- Trong mỗi ví dụ trên:

    1.  Biểu thức bên phải của toán tử gán `=` được tính toán trước.
    2. Phép so sánh trả về `True` hoặc `False`.
    3. Giá trị đó được lưu vào **variable**.

- Mối liên hệ trực tiếp **giữa phép so sánh và kết quả boolean** là nền tảng để điều khiển hành vi của chương trình.

### **Booleans and Logical Operations**

- Giá trị boolean là nền tảng cho **việc ra quyết định trong Python**. Chúng thường được dùng với **logical operators**:
    - `and` - `or` - `not`
    - Để kết hợp nhiều điều kiện
- Ví dụ: Kiểm tra xem người dùng **đã đăng nhập** và **có quyền quản trị viên** hay chưa.

````python
logged_in = True
is_admin = False

# 'and' requires both sides to be True for the result to be True
can_access_admin_panel = logged_in and is_admin
print(can_access_admin_panel) # Output: False

# 'or' requires at least one side to be True for the result to be True
can_view_content = logged_in or is_admin
print(can_view_content)       # Output: True

# 'not' inverts the boolean value
is_guest = not logged_in
print(is_guest)               # Output: False
````

### **Truthiness: When Other Types Act Like Boolean**

- Python có một khái niệm hữu ích gọi là **truthiness**.
- Trong các ngữ cảnh cần giá trị boolean (ví dụ trong câu lệnh `if`), Python có thể đánh giá **các kiểu dữ liệu khác** như:
    - **truthy** (hoạt động như `True`)
    - **falsy** (hoạt động như `False`)

#### **falsy/truthy In Python**

- Những giá trị sau được coi là **falsy**:
    - `False`
    - `None` (biểu thị không có giá trị)
    - Số `0` của bất kỳ kiểu số nào `(0, 0.0)`
    - Các sequence hoặc collection rỗng:
        - String rỗng: `""`
        - List rỗng: `[]`
        - Tuple rỗng: `()`
        - Dictionary rỗng: `{}`
        - Set rỗng: `set()`
- Phần còn lại là **truthy**

- Tại sao điều này hữu ích? Vì nó giúp viết các kiểm tra ngắn gọn hơn.

- Ví dụ: Thay vì kiểm tra list `my_list` có nhiều hơn `0` phần tử bằng `len(my_list) > 0`, bạn thường chỉ cần viết `if my_list:`, vì list không rỗng là **truthy**, còn list rỗng là **falsy**

````python
items = []
if items: # items is empty, so it's falsy
    print("There are items in the list.")
else:
    print("The list is empty.") # This will be printed

user_name = "Charlie"
if user_name: # user_name is not empty, so it's truthy
    print(f"Hello, {user_name}!") # This will be printed
else:
    print("User name is missing.")

count = 0
if count: # count is 0, so it's falsy
    print("Count is positive.")
else:
    print("Count is zero.") # This will be printed
````

- Mặc dù `True` và `False` là các **giá trị boolean tường minh**, việc hiểu **truthiness** giúp bạn đọc được cách Python đưa ra quyết định trong logic điều kiện.

- Kiểu đánh giá boolean ngầm định này khiến code Python thường ngắn gọn và dễ đọc hơn.

## **#5: Working with Operators: Arithmetic**

- Lập trình Python thường liên quan đến việc làm việc với dữ liệu số. Các số này có thể được lưu trong biến dưới dạng **integer** (như `10, -5, 0`) hoặc **floating-point numbers** (như `3.14, -0.5, 2.0`).

- Để thực hiện các phép tính với các số này, Python cung cấp một tập hợp **arithmetic operators**. Những **operator** này là các công cụ cơ bản, giống như một máy tính, để thực hiện các phép toán trong chương trình của bạn.

### **Basic Arithmetic**



#### **Addition (`+`)**
- Cộng hai số lại với nhau.

````python
apples = 5
oranges = 10
total_fruit = apples + oranges
print(total_fruit) # Output: 15

price = 9.99
tax = 0.80
total_cost = price + tax
print(total_cost) # Output: 10.79
````

#### **Subtraction (`-`)**
- Trừ hai số với nhau

````python
budget = 100
spent = 35.50
remaining = budget - spent
print(remaining) # Output: 64.5
````

#### **Multiplication (`*`)**

- Nhân 2 số

````python
width = 8
height = 12
area = width * height
print(area) # Output: 96
````

#### **Division (`/`)**
- Chia 2 số với nhau.

    > Điều quan trọng cần lưu ý là **phép chia chuẩn trong Python luôn trả về số floating-point**, ngay cả khi kết quả là số nguyên.

````python
total_distance = 100 # kilometers
time_hours = 2 # hours
speed = total_distance / time_hours
print(speed) # Output: 50.0

items = 10
people = 4
items_per_person = items / people
print(items_per_person) # Output: 2.5
````

- Trong Python 3, sự khác biệt giữa 50 (số nguyên - int) và 50.0 (số thực - float) là rất quan trọng. Việc phép chia luôn trả về số thực (float) mang lại sự nhất quán trong lập trình vì các lý do sau:

    1. **Tránh mất mát dữ liệu (Data Loss)**:
    - Trong nhiều ngôn ngữ lập trình cũ, phép chia hai số nguyên thường trả về một số nguyên (ví dụ: `5 / 2 = 2`), làm mất đi phần thập phân.
    - Python chọn cách trả về `2.5` để đảm bảo độ chính xác ngay cả khi bạn không yêu cầu.

    2. **Sự nhất quán trong các phép toán tiếp theo**:
    - Hãy tưởng tượng bạn đang viết một chương trình tính toán phức tạp:

        - Nếu `100 / 2` trả về `50` (int), nhưng `101 / 2` trả về `50.5` (float), kiểu dữ liệu của kết quả sẽ thay đổi tùy thuộc vào con số đầu vào.
        
        - Việc luôn trả về float (ví dụ `50.0` thay vì `50`) giúp các hàm hoặc công thức phía sau luôn nhận được cùng một kiểu dữ liệu, tránh được các lỗi "bug" ngầm định khi chương trình mong đợi một con số có thể chứa phần thập phân.

### **Integer Division and Remainder**
- Python cung cấp thêm hai toán tử chia đặc biệt, rất hữu ích khi làm việc với số nguyên

#### **Floor Division (`//`)**:

- Thực hiện phép chia giống /, nhưng **loại bỏ phần thập phân và làm tròn xuống số nguyên gần nhất**. Đây còn được gọi là **integer division**.

````python
cookies = 10
friends = 3
cookies_per_friend = cookies // friends
print(cookies_per_friend) # Output: 3 (Each friend gets 3 whole cookies)

value = 7
divisor = 2
result = value // divisor
print(result) # Output: 3 (7 divided by 2 is 3.5, rounded down is 3)

# Compare with standard division
print(10 / 3)  # Output: 3.3333333333333335
print(10 // 3) # Output: 3
````

#### **Modulo (`%`)**

- Toán tử này không trả về kết quả chia, mà trả về **phần dư của phép chia**. Vì vậy nó thường được gọi là **remainder operator**.

````python
total_cookies = 10
friends = 3
leftover_cookies = total_cookies % friends
print(leftover_cookies) # Output: 1 (After giving 3 cookies each, 1 is left)

number = 17
divisor = 5
remainder = number % divisor
print(remainder) # Output: 2 (17 divided by 5 is 3 with a remainder of 2)
````

- Toán tử modulo thường được dùng để kiểm tra số **chẵn hay lẻ**, nếu `number %2` bằng `0`.

````python
num = 6
print(num % 2) # Output: 0 (So, 6 is even)

num = 7
print(num % 2) # Output: 1 (So, 7 is odd)
````

### **Exponentiation (`**`)**
- Để nâng một số lên lũy thừa của số khác, Python dùng toán tử **double asterisk (**)**: `x ** y` nghĩa là $x^y$

````python
base = 2
exponent = 3
power_result = base ** exponent # Equivalent to 2 * 2 * 2
print(power_result) # Output: 8

side = 5
area_square = side ** 2 # Equivalent to 5 * 5 (5 squared)
print(area_square) # Output: 25
````

### **Order of Operations**
- Khi một biểu thức có nhiều operator, Python tuân theo **thứ tự phép toán** giống toán học **(PEMDAS/BODMAS)**:

    1. Parentheses `()`
    2. Exponentiation `**`
    3. Multiplication `*`, Division `/`, Floor Division `//`, Modulo `%`
    4. Addition `+`, Subtraction `-`

````python
calculation = 100 - 20 / 5 * 2 + 3**2
# Step-by-step evaluation:
# 1. Exponentiation: 3**2 becomes 9
#    Expression: 100 - 20 / 5 * 2 + 9
# 2. Division/Multiplication (from left to right):
#    20 / 5 becomes 4.0
#    Expression: 100 - 4.0 * 2 + 9
#    4.0 * 2 becomes 8.0
#    Expression: 100 - 8.0 + 9
# 3. Addition/Subtraction (from left to right):
#    100 - 8.0 becomes 92.0
#    Expression: 92.0 + 9
#    92.0 + 9 becomes 101.0
print(calculation) # Output: 101.0
````

### **Shorthand: Augmented Assignment Operators**

- Trong lập trình, bạn thường cần cập nhật giá trị của một biến dựa trên giá trị hiện tại của nó.

````python
count = 0
count = count + 1 # Increment count by 1
print(count) # Output: 1

total_cost = 50
item_price = 15.50
total_cost = total_cost + item_price # Add item_price to total_cost
print(total_cost) # Output: 65.5
````

- Python cung cấp các **augmented assignment operators** để viết ngắn gọn hơn.

````python
count = 0
count += 1 # Equivalent to count = count + 1
print(count) # Output: 1

score = 100
score -= 10 # Equivalent to score = score - 10
print(score) # Output: 90

multiplier = 5
multiplier *= 2 # Equivalent to multiplier = multiplier * 2
print(multiplier) # Output: 10

total = 50.0
total /= 4 # Equivalent to total = total / 4
print(total) # Output: 12.5
````

- Các toán tử viết tắt này (`+=, -=, *=, /=, //=, %=, **=`) tồn tại cho tất cả các toán tử số học và kết hợp phép toán với phép gán. Chúng rất tiện và thường giúp code dễ đọc hơn khi bạn đã quen dùng.

## **#6: Working with Operators: Comparison**

## **#7: Working with Operators: Logical**

- Các toán tử như `==` (bằng), `!=` (khác), `<` (nhỏ hơn), và `>` (lớn hơn) được dùng để so sánh giá trị và luôn trả về một giá trị boolean: `True` hoặc `False`.

- **Logical operators (toán tử logic)** cho phép **kết hợp các kết quả boolean này**, từ đó tạo ra các biểu thức điều kiện phức tạp hơn. Trong nhiều trường hợp, bạn cần **kiểm tra nhiều điều kiện cùng lúc**. Ví dụ:
    - Một người dùng **đã đăng nhập và có quyền admin**
    - Hôm nay **là thứ Bảy hoặc Chủ nhật**

- Đây chính là lúc **logical operators** phát huy tác dụng. Chúng cho phép bạn **kết hợp nhiều giá trị boolean** để tạo ra nhiều điều kiện phức tạp.

- Python cung cấp **ba toán tử logic**:
    - `and`- `or` - `not`


### **The `and` Operator**

- Toán tử `and` đánh giá **hai biểu thức boolean**.
- Nó chỉ trả về `True` khi **cả hai biểu thức đều là** `True`. Nếu **một trong hai hoặc cả hai biểu thức `False`**, kết quả là `False`

````python
age = 25
has_ticket = True

# Is the person old enough AND has a ticket?
can_enter = (age >= 18) and (has_ticket == True)
print(f"Age is 25, has ticket: {can_enter}") # Output: True

age = 16
# Now the first condition (age >= 18) is False
can_enter = (age >= 18) and (has_ticket == True)
print(f"Age is 16, has ticket: {can_enter}") # Output: False

age = 30
has_ticket = False
# Now the second condition (has_ticket == True) is False
can_enter = (age >= 18) and (has_ticket == True)
print(f"Age is 30, no ticket: {can_enter}") # Output: False
````

| Condition 1 | Condition 2 | Condition 1 and Condition 2 |
|:------------|:------------|:----------------------------|
| True        | True        | True                        |
| True        | False       | False                       |
| False       | True        | False                       |
| False       | False       | False                       |

### **The `or` Operator**

- Toán tử `or` cũng đánh giá **hai biểu thức boolean**.

- Nó trả về `True` nếu **ít nhất một biểu thức là `True`**.

- Chỉ khi **cả hai đều** `False` thì kết quả mới là `False`.


````python
is_logged_in = False
print(f"Is logged in: {is_logged_in}") # Output: False

# Check if the user is NOT logged in
needs_login = not is_logged_in
print(f"Needs login: {needs_login}") # Output: True

is_raining = True
print(f"Is raining: {is_raining}") # Output: True

# Check if it is NOT raining
can_go_outside = not is_raining
print(f"Can go outside: {can_go_outside}") # Output: False
````

| Condition 1 | Condition 2 | Condition 1 or Condition 2 |
|:------------|:------------|:---------------------------|
| True        | True        | True                       |
| True        | False       | True                       |
| False       | True        | True                       |
| False       | False       | False                      |

### **The `not` Operator**

- Toán tử `not` hoạt động với **một biểu thức boolean** và **đảo ngược giá trị của nó**:

    - `not True` → `False`
    - `not False` → `True`

````python
is_logged_in = False
print(f"Is logged in: {is_logged_in}") # Output: False

# Check if the user is NOT logged in
needs_login = not is_logged_in
print(f"Needs login: {needs_login}") # Output: True

is_raining = True
print(f"Is raining: {is_raining}") # Output: True

# Check if it is NOT raining
can_go_outside = not is_raining
print(f"Can go outside: {can_go_outside}") # Output: False
````

| Condition | not Condition |
|:----------|:--------------|
| True      | False         |
| False     | True          |

### **Combining Logical Operators**

- Bạn có thể **kết hợp nhiều toán tử logic lại trong một biểu thức**.
- Thứ tự ưu tiên của chúng trong Python: `not` - `and` - `or`
- Tuy nhiên, bạn có thể dùng dấu ngoặc `()` để:
    - Kiểm soát thứ tự, và làm code dễ đọc hơn

````python
age = 22
is_student = True
has_coupon = False

# Eligible if (age < 25 AND is a student) OR has a coupon
is_eligible_discount = (age < 25 and is_student) or has_coupon
# Evaluates as: (True and True) or False
# -> True or False
# -> True
print(f"Eligible for discount: {is_eligible_discount}") # Output: True

# Change coupon status
has_coupon = True
age = 30
is_student = False

is_eligible_discount = (age < 25 and is_student) or has_coupon
# Evaluates as: (False and False) or True
# -> False or True
# -> True
print(f"Eligible for discount (different criteria): {is_eligible_discount}") # Output: True
````

```py
(False and False) or True
→ False or True
→ True
```

## **#8: Type Conversion Explained**

- Các biến trong Python có thể lưu trữ nhiều kiểu dữ liệu khác nhau, nhưng kiểu dữ liệu cụ thể rất quan trọng khi thực hiện các phép toán.

- Ví dụ, bạn không thể cộng trực tiếp một số với một chuỗi biểu diễn số:
    ```
    42 + "10"
    ```

- Điều này không hoạt động về mặt toán học vì Python hiểu `"10"` là **văn bản (text)** chứ không phải **giá trị số 10**. Để thực hiện phép tính, bạn phải **chuyển `"10"` thành số**.

- Quá trình thay đổi **kiểu dữ liệu của một giá trị** được gọi là: **Type conversion** hoặc **Type Casting**.

- Python cung cấp nhiều **hàm built-in** để thực hiện các chuyển đổi phổ biến:

    - `int()`: Chuyển sang số nguyên.
    - `float()`: Chuyển sang số thực.
    - `str()`: Chuyển sang chuỗi.
    - `bool()`: Chuyển sang giá trị boolean

### **Converting to Integers: `int()`**

- Hàm `int()` cố gắng tạo một số nguyên từ giá trị đầu vào:

#### **From Floats**
- Khi chuyển **float → int**, Python **cắt bỏ phần thập phân** - Python không làm tròn, mà chỉ **loại bỏ phần thập phân**.

````python
count_float = 9.8
count_int = int(count_float)
print(count_int)  # Output: 9

price_float = 15.2
price_int = int(price_float)
print(price_int) # Output: 15
````

#### **From Strings**
- Bạn có thể chuyển **string** chứa chữ số thành số nguyên.

````python
num_string = "101"
num_int = int(num_string)
print(num_int * 2) # Output: 202
````

#### **Potential Errors**
- Nếu **string** không thể diễn giải thành số nguyên, Python sẽ tạo lỗi: `ValueError`

````python
# This will cause an error:
value_error_int = int("hello")
# ValueError: invalid literal for int() with base 10: 'hello'

# This will also cause an error:
value_error_float_string = int("3.14")
# ValueError: invalid literal for int() with base 10: '3.14'
````

### **Converting to Floats: `float()`**

- Hàm `float()` chuyển giá trị thành **số thực (floating-point number)**

#### **From Integers**
- Khi chuyển **int → float**, Python chỉ thêm phần `.0`.

````python
whole_number = 50
float_number = float(whole_number)
print(float_number) # Output: 50.0
````

#### **From Strings:**
- Bạn có thể chuyển **string** chứa chữ số thành số nguyên.

````python
pi_string = "3.14159"
pi_float = float(pi_string)
print(pi_float) # Output: 3.14159

count_string = "100"
count_float_from_string = float(count_string)
print(count_float_from_string) # Output: 100.0
````

#### **Potential Errors**:
- Tương tự như `int()`, nếu string không biểu diễn một số hợp lệ thì Python sẽ báo lỗi `ValueError`.

````python
# This will cause an error:
value_error_float = float("not a number")
# ValueError: could not convert string to float: 'not a number'
````

### **Converting to Strings: `str()`**

- Hàm `str()` chuyển đối số của nó thành một representation dạng string. Đây là hàm cực kỳ hữu ích khi bạn muốn tạo thông báo đầu ra kết hợp text với giá trị của biến.

#### **From Numbers**

- Integer và float đều có thể được chuyển dễ dàng thành string.

````python
item_count = 5
message = "You have " + str(item_count) + " items in your cart."
print(message) # Output: You have 5 items in your cart.

temperature = 22.5
weather_report = "The temperature is " + str(temperature) + " degrees Celsius."
print(weather_report) # Output: The temperature is 22.5 degrees Celsius.
````

#### **From Booleans**

- Giá trị boolean sẽ được chuyển thành string `"True"` hoặc "False".

````python
is_active = True
status_message = "User active status: " + str(is_active)
print(status_message) # Output: User active status: True
````

### **Converting to Booleans: `bool()`**

- Hàm `bool()` chuyển một giá trị sang tương đương boolean của nó (`True` hoặc `False`). Python có khái niệm **truthiness**, nghĩa là một số giá trị vốn dĩ được xem là **false** trong ngữ cảnh boolean, còn hầu hết giá trị khác là **true**.

````python
print(bool(100))        # Output: True
print(bool(-5.5))       # Output: True
print(bool("Python"))     # Output: True
print(bool([1, 2, 3]))  # Output: True

print(bool(0))          # Output: False
print(bool(0.0))        # Output: False
print(bool(""))          # Output: False
print(bool([]))         # Output: False
print(bool({}))         # Output: False
print(bool(None))       # Output: False
````

### **Why is This Important Right Now?**

- **Type conversion** rất quan trọng ngay từ đầu vì cách Python xử lý dữ liệu từ người dùng. Khi bạn dùng hàm `input()` để nhận dữ liệu từ bàn phím, nó luôn trả về **string**, bất kể người dùng nhập gì.

- Nếu bạn hỏi tuổi của người dùng và họ nhập `30`, hàm `input()` sẽ trả về string `"30"`, không phải **integer 30**. Nếu sau đó bạn muốn thực hiện tính toán, chẳng hạn tính tuổi sau 5 năm, bạn sẽ cần chuyển **string "30"** đó thành **integer 30** bằng `int()` trước khi cộng thêm `5`.

Nếu cố tính `age_input + 1`, chương trình sẽ báo lỗi `TypeError`, vì Python không biết cách cộng một integer với một string trong ngữ cảnh này.

- Nếu bạn cần xử lý dữ liệu người dùng như một số nguyên hoặc số thực, bạn phải **chủ động chuyển đổi** string do `input()` trả về. Bạn có thể dùng các hàm chuyển kiểu đã học như `int()` hoặc `float()`.[văn bản liên kết](https://)

## **#10: Adding Comments to Your Code**

- Viết code chạy được là một kỹ năng, nhưng viết code **dễ hiểu cho bản thân và người khác trong tương lai** lại là một kỹ năng khác. **Comments (chú thích)** giúp đạt được mục tiêu này.

- Comment là những ghi chú trong code mà Python **bỏ qua khi thực thi chương trình**. Mục đích duy nhất của chúng là **giải thích code cho con người đọc**.
- Bạn có thể xem comment như những ghi chú để lại cho chính mình hoặc cho đồng nghiệp nhằm làm rõ mục đích, logic hoặc giả định của một đoạn code.

### **Writing Comments in Python**

- Trong Python, bất kỳ nội dung nào **sau dấu # trên một dòng** đều được xem là comment và sẽ bị **interpreter** bỏ qua.

````python
# This is a full-line comment. It explains the general purpose of the code below.
principal = 1000  # This is an inline comment. It explains this specific variable.
rate = 0.05       # Annual interest rate
years = 3         # Investment duration in years

# Calculate simple interest
interest = principal * rate * years

# Print the result (we'll learn more about formatted printing later)
print("The calculated simple interest is:")
print(interest)
````

- Trong ví dụ này:

    - Dòng đầu tiên là **comment toàn dòng**, giải thích mục tiêu chung của đoạn code.
    - Các comment sau `principal, rate, years` là **inline comments**, giải thích từng biến.
    - Comment trước phép tính giải thích **công thức được sử dụng**.
    - Comment trước `print` giải thích các dòng tiếp theo sẽ làm gì.

## **Commenting Out Code**

- Nếu nghi ngờ một dòng code gây lỗi, bạn có thể **comment nó lại** thay vì xóa hoàn toàn.

    - Dễ quản lý hơn
