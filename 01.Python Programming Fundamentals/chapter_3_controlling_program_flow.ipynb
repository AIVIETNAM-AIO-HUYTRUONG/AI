Chương này giới thiệu các **cấu trúc cốt lõi** giúp bạn **điều khiển luồng thực thi (execution path) của code python**.

   - Hiểu cách sử dụng **conditional statements**: `if, elif, else`- để chạy những **khối mã (block of code)** - khi thoải mã **điều kiện (conditions)**

## **#1: Making Decisions: The if Statement**

- Trong lập trình Python, khi bạn muốn một đoạn code chỉ **thực hiện logic** khi nó **thỏa mãn một điều kiện nào đó**.
- Ví dụ: Ở ngoài đời thực - Nếu trời mưa, tôi sẽ mang ô. Chương trình bạn viết code cũng cần cách để ra quyết định tương tự. Và Python cung cấp **câu lệnh `if`** để làm điều này.

- **Câu lệnh `if`** cho phép bạn thực thi một **khối mã (block of code)** có **điều kiện**. Nó kiểm tra xem một điều kiện nào đó có `True` hay `False`.
    - Nếu **điều kiện** là `True` -> **khối mã** của `if` sẽ được thực thi.
    - Nếu **điều kiện** là `False` -> **khối mã** bị bỏ qua, và chương trình tiếp tục các lệnh tiếp theo.

### **1. The Structure of an if Statement**

- Cú pháp cơ bản như sau:

```python
if condition:
    # Code to execute if the condition is True
    statement1
    statement2
    # ... more statements ...

# Code here runs regardless of the condition, after the if block
next_statement
```

- Ta hãy phân tích:
    1. **`if` Keyword**: Câu lệnh bắt đầu bằng từ khóa `if`.
    2. **`condition`**: Đây là một biểu thức trả về `True` hoặc `False`. Nó thường được gọi là **boolean expression** và dùng:
        - Các **comparison operators** như `==`,`!=`,`<`,`>`,`<=`,`>=`
        - Hoặc các **logical operators** như `and`, `or`, `not`
    3. **Colon `:`**: Sau điều kiện phải có dấu hai chấm. Nó đánh dấu bắt đầu của **khối code** phụ thuộc vào điều kiện.
    4. **Indented Block**:
        - Các dòng code ngay sau `if condition`: và được thụt vào bên phải sẽ tạo thành **if block**.
        - Python dùng **indentation (thường là 4 dấu cách)** để xác định các **khối code**. Tất cả **các dòng có cùng mức thụt lề dưới `if` đều thuộc cùng một block**

### **2. How it Works**

- Hãy tưởng tượng bạn đang viết một chương trình kiểm tra xem người dùng đã đủ tuổi để bỏ phiếu chưa. Độ tuổi bỏ phiếu thường là 18

````python
age = 20

print("Checking voting eligibility...")

if age >= 18:
    print("You are old enough to vote.")
    print("Please register if you haven't already.")

print("Eligibility check complete.")
````

- Biến `age` được gán giá trị `20`.
- Câu lệnh `print("Checking voting eligibility...")` được thực thi.
- **Câu lệnh `if age >= 18:`** kiểm tra điều kiện. Vì `20 >= 18` là `True`, **điều kiện được thỏa mãn**.
- Do **điều kiện đúng**, **khối mã bên trong `if`** được thực thi:
    - `print("You are old enough to vote.")`
    - `print("Please register if you haven't already.")`
- Sau khi **khối mã `if`** kết thúc, chương trình tiếp tục thực thi **`print("Eligibility check complete.")`**.

- Nhưng trong trường hợp tuổi người dùng dưới 18 thì sẽ như nào?

````python
age = 15

print("Checking voting eligibility...")

if age >= 18:
    # This block will be skipped
    print("You are old enough to vote.")
    print("Please register if you haven't already.")

print("Eligibility check complete.")
````

- Biến `age` được gán giá trị `15`.
- Câu lệnh `print("Checking voting eligibility...")` được thực thi.
- **Câu lệnh `if age >= 18:`** kiểm tra điều kiện. Vì `15 >= 18` là `False`, **điều kiện không được thỏa mãn**.
- Do **điều kiện sai**, **khối mã bên trong `if`** bị bỏ qua.
- Chương trình tiếp tục thực thi **`print("Eligibility check complete.")`**.

### **3. The Importance of Indentation**
- **Indentation trong Python** không chỉ để dễ đọc; nó là **một phần của cú pháp**. **Nó cho interpreter biết câu lệnh nào thuộc về khối if**. **Thụt lề sai sẽ dẫn đến lỗi hoặc hành vi không mong muốn**.

````python
# Correct indentation
temperature = 30
if temperature > 25:
    print("It's a warm day.") # Indented correctly
    print("Wear light clothes.") # Indented correctly

# Incorrect indentation (would cause an IndentationError if uncommented)
if temperature > 25:
print("It's a warm day.") # Missing indentation
````

### **5. Visualizing the Flow**

Để dễ hình dung hơn về cách **Python** điều khiển luồng thực thi, hãy xem sơ đồ tư duy đơn giản dưới đây cho câu lệnh `if`:

1. **Bắt đầu**: Chạy các lệnh tuần tự bên trên.
2. **Điểm quyết định (`if condition:`)**:
    - Nếu **Kết quả là `True`** -> Rẽ nhánh vào **Khối mã thụt lề (Indented Block)**.
    - Nếu **Kết quả là `False`** -> **Nhảy qua (Skip)** toàn bộ khối thụt lề đó.
3. **Hội tụ**: Cả hai nhánh đều quay lại thực thi **Dòng code không thụt lề đầu tiên** ngay sau khối `if`.

```text
[Lệnh trước đó]
       |
       v
  <Điều kiện ?> --(False)---|
       |                    |
    (True)                  |
       |                    |

       |                    |
       |<-------------------|
       v

```

### **6. Summary**

- **Mục đích**: Cho phép chương trình **đưa ra quyết định** bằng cách chỉ thực thi một khối mã khi một **điều kiện (condition)** là `True`.
- **Cú pháp**: Bắt đầu bằng từ khóa `if`, theo sau là **điều kiện** và dấu hai chấm `:`.
- **Khối lệnh (Indentation)**: Các câu lệnh bên trong `if` **bắt buộc phải thụt lề** (thường là 4 khoảng trắng). Đây là cách Python xác định các lệnh nào thuộc về nhánh `if`.
- **Luồng thực thi**:
    - Nếu điều kiện **Đúng (`True`)**: Thực thi khối mã thụt lề.
    - Nếu điều kiện **Sai (`False`)**: Bỏ qua khối mã thụt lề và nhảy đến lệnh tiếp theo không thụt lề.

## **#2: Handling Alternatives: elif and else**

- Câu lệnh `if` cho phép chương trình thực thi một block code khi một `condition` cụ thể có giá trị **`True`**. Tuy nhiên, trong nhiều trường hợp, chương trình cần kiểm tra **nhiều điều kiện** hoặc thực hiện một **hành động mặc định** khi các điều kiện trước đó không thỏa mãn.
- **`elif`** và **`else`** được dùng để mở rộng `if` cho những tình huống đưa ra quyết định phức tạp hơn.

### **1. Checking Multiple Conditions with `elif`**

- Câu lệnh **`elif`** (viết tắt của **else if**) cho phép ta kiểm tra thêm các conditions khác **chỉ khi** các điều kiện trước đó trong `if` hoặc `elif` đều là **`False`**.

- Python sẽ đánh giá các conditions theo **thứ tự từ trên xuống dưới**. Ngay khi gặp **condition đầu tiên** có giá trị **`True`**, nó sẽ thực thi block mã tương ứng và sau đó **bỏ qua toàn bộ phần còn lại** của cấu trúc `if-elif-else`.

```python
if condition_1:
    # Block 1: Executes if condition_1 is True
    statement(s)
elif condition_2:
    # Block 2: Executes if condition_1 is False AND condition_2 is True
    statement(s)
elif condition_3:
    # Block 3: Executes if condition_1 AND condition_2 are False,
    # AND condition_3 is True
    statement(s)
# ... you can have more elif blocks
```

````python
number = -10

if number > 0:
    print("The number is positive.")
elif number < 0:
    print("The number is negative.")

# If number is 0, neither condition above is True, so nothing happens yet.
````

- Trong trường hợp này, nếu **`number = -10`**:
    - `number > 0` là **`False`**, nên block đầu tiên bị bỏ qua.
    - `number < 0` là **`True`**, nên thông báo `"The number is negative."` được in ra.
    - Sau đó chương trình **bỏ qua** mọi `elif` hoặc `else` phía sau trong cùng một cấu trúc này.

- Nếu **`number = 0`** thì sao? Khi đó cả hai điều kiện đều là **`False`**. Để xử lý trường hợp "không có điều kiện nào đúng", ta dùng **`else`**.

### **2. The Default Case: `else`**

- **`else`** cung cấp một block **mặc định** chỉ chạy khi **tất cả** các conditions trước đó đều là **`False`**.
- Nếu dùng `else`, nó luôn phải **đứng cuối cùng**, không đi kèm với bất kỳ điều kiện nào.
```python
if condition_1:
    # Block 1: Executes if condition_1 is True
    statement(s)
elif condition_2:
    # Block 2: Executes if condition_1 is False AND condition_2 is True
    statement(s)
# ... more elif blocks (optional) ...
else:
    # Block N: Executes if ALL preceding conditions (condition_1, condition_2, ...) are False
    statement(s)
```

````python
number = 0

print(f"Checking the number: {number}")

if number > 0:
    print("The number is positive.")
elif number < 0:
    print("The number is negative.")
else:
    # This block runs only if number is not > 0 AND number is not < 0
    print("The number is zero.")

print("Finished checking.")
````

- Nếu chạy với **`number = 0`**:
    - `number > 0` là **`False`**.
    - `number < 0` là **`False`**.
    - Vì tất cả các conditions trước đó đều **`False`**, block **`else`** được thực thi và in ra `"The number is zero."`

### **3. Order and Exclusivity**

- Điều quan trọng cần nhớ trong cấu trúc `if-elif-else` là:
    - **Order matters (Thứ tự quan trọng)**: Các conditions được kiểm tra theo đúng thứ tự xuất hiện. Nếu có nhiều điều kiện cùng đúng, **chỉ block đầu tiên thỏa mãn** sẽ được chạy.
    - **Exclusivity (Tính loại trừ)**: Trong một chuỗi `if-elif-else`, **tối đa chỉ có một block** được thực thi. Các phần còn lại sẽ bị bỏ qua ngay lập tức sau khi một nhánh đã chạy.

### **4. Summary**

- **`elif` (else if)**: Dùng để kiểm tra thêm các điều kiện mới nếu các điều kiện phía trên đều **`False`**. Bạn có thể có nhiều khối `elif`.
- **`else`**: Là nhánh **mặc định**, không có điều kiện đi kèm và luôn nằm ở cuối cùng. Nó chỉ chạy khi **tất cả** các điều kiện `if` và `elif` trước đó đều sai.
- **Tính loại trừ**: Trong một chuỗi `if-elif-else`, Python chỉ thực thi **duy nhất một khối mã** đầu tiên thỏa mãn điều kiện và bỏ qua tất cả các khối còn lại.
- **Thứ tự**: Luôn bắt đầu bằng `if`, tiếp theo là các `elif` (nếu có) và kết thúc bằng `else` (nếu có).

## **#:3 Repeating Actions: The while Loop**

- Vòng lặp được sử dụng trong lập trình để lặp đi lặp lại một đoạn code xử lý logic cụ thể nào đó,

- Python cung cập `while` để thực hiện điều này với **điều kiên kiểm tra luôn đúng**. `while` dùng trong những trường hợp mà chúng ta **không thể dự đoán trước được số lần cần lặp là bao nhiều**

### **1. The Structure of a `while` Loop**

- Cú pháp cơ bản của `while` loop như sau:

  ```python
  while condition:
      # Code block to be executed repeatedly
      # This block MUST be indented
      statement1
      statement2
      # ... potentially update variables affecting the condition ...
  ```

- Hãy phân tích:

  - **`while` keyword**: Báo hiệu bắt đầu một loop.
  - `condition`: Một **boolean expression (biểu thức boolean)**, tức biểu thức trả về `True` hoặc `False`, giống như **condition** bạn dùng với **`if` statement**. Ví dụ:
    - `count < 5`, `user_input != 'quit'`, `is_valid == False`.

  - **Colon** `:` : Đánh dấu kết thúc phần header của câu lệnh `while`.
  - **Indented Code Block (khối code thụt lề)**:
    - Là phần thân của loop. Tất cả statements thụt vào dưới dòng `while` đều thuộc loop.
    - Block này sẽ được chạy lặp lại. Điều cực kỳ quan trọng là phải có thứ gì đó trong block làm cho **condition** cuối cùng trở thành `False`, nếu không loop sẽ chạy mãi mãi.

### **2. How Python Executes a `while` Loop**

- Luồng thực thi của `while` loop khá đơn giản:
  1. **Check the Condition (kiểm tra condition)**: Python đánh giá condition trước.
  2. **Execute or Skip (thực thi hoặc bỏ qua)**:
    - Nếu **condition** là `True`, Python chạy toàn bộ **indented code block**.
    - Nếu **condition** là `False`, Python bỏ qua block đó và chương trình tiếp tục ở statement đầu tiên sau loop.
  3. **Repeat (lặp lại)**: Nếu block đã được chạy, Python quay lại bước 1 và kiểm tra **condition** lại.

- Chu trình **check-execute-repeat** này tiếp tục cho đến khi condition trở thành `False`.

### **3. Example: A Simple Counter**

- VD: Hãy tạo một loop in ra các số từ 1 đến trước 5.

````python
# Initialize a counter variable BEFORE the loop
count = 1

# The loop continues as long as 'count' is less than 5
while count < 5:
    print(f"Current count is: {count}")
    # IMPORTANT: Update the counter inside the loop!
    count = count + 1 # Or use the shorthand: count += 1

print("Loop finished!")
````

- Hãy chú ý 3 phần quan trọng để điều khiển loop này:

  - **Initialization (khởi tạo)**: `count = 1` thiết lập trạng thái ban đầu trước khi loop bắt đầu.
  - **Condition (điều kiện)**: `count < 5` được kiểm tra trước mỗi lần lặp.
  - **Update (cập nhật)**:
    - `count = count + 1` thay đổi variable được dùng trong **condition** bên trong loop.
    - Nếu không cập nhật, `count` sẽ luôn là `1`, `count < 5` sẽ luôn là `True`, và loop sẽ không bao giờ kết thúc.

### **4. The Danger: Infinite Loops**

- Nếu condition trong `while` loop không bao giờ trở thành `False`, loop sẽ chạy mãi. Đây gọi là **infinite loop (vòng lặp vô hạn)**. Đây là lỗi phổ biến với người mới, thường xảy ra khi quên viết logic thoát ra bên trong `while` loop.

- VD: Đoạn code sau đây để tái hiện một **infinite loop**

````python
# WARNING: This creates an infinite loop!
counter = 0
while counter >= 0:
    print(f"Counter is {counter}. Still positive!")
    # Oops! We forgot to change 'counter' in a way
    # that makes the condition (counter >= 0) false.
    # Maybe we meant to decrement it? Or check a different condition?
    # Without a change, it will run forever.
    counter = counter + 1 # This actually makes it *more* likely to stay >= 0!
````

- Nếu bạn lỡ chạy code có **infinite loop** trong **ternimal** hoặc **interactive interpreter**, ta sử dùng `ctrl + c` để dừng chương trình lại.

## **5. Example: Waiting for Specific User Input**

- `while` loops rất hữu ích khi bạn cần lặp cho đến khi một sự kiện cụ thể xảy ra, như nhận input hợp lệ từ người dùng

````python
# Initialize response to something that ensures the loop runs at least once
response = ""

# Keep asking until the user types 'quit'
while response.lower() != 'quit':
    response = input("Enter some text (or type 'quit' to exit): ")
    print(f"You entered: {response}")

print("Okay, quitting now. Goodbye!")
````

- Trong trường hợp này, ta không biết user sẽ nhập bao nhiêu lần trước khi gõ `quit`.
- `while` loop xử lý sự không chắc chắn này rất tốt, vì nó kiểm tra response sau mỗi lần nhập.

## **#4: Iterating Over Sequences: The `for` Loop**

- **`for` loop (vòng lặp for)** cung cấp một cách trực tiếp và hiệu quả để thực hiện một hành động cho mỗi item trong một **sequence** hoặc **collection**.

- Nó đặc biệt hữu ích khi xử lý một **collection** đã biết trước, từng phần tử một. Các trường hợp dùng phổ biến gồm lặp qua tất cả phần tử trong một list, tất cả ký tự trong một string, hoặc lặp lại một hành động một số lần xác định.

### **1. Basic for `Loop` Syntax**

- Cấu trúc nền tảng của một `for` loop trông như sau:


  ```python
  for variable_name in sequence:
      # Code block to execute for each item
      # Use variable_name to access the current item
      statement1
      statement2
      # ...
  ```

- Hãy phân tích:

  - `for`: từ khóa bắt đầu loop.
  - `variable_name`: một biến do bạn đặt tên, nên dùng tên có ý nghĩa, dùng để chứa item hiện tại từ sequence trong mỗi vòng lặp **(iteration / pass)**.
  - `in`: Từ khóa phân tách loop variable và sequence.
  - `sequence`: Collection các item bạn muốn lặp qua. Đây có thể là list, string, hoặc các kiểu iterable khác mà ta sẽ gặp sau.
  - `:` : Dấu hai chấm đánh dấu kết thúc dòng khai báo for.
  - **Indented Code Block (khối code thụt lề)**: giống như `if` và `while`, code muốn chạy trong loop phải được thụt lề. Block này chạy một lần cho mỗi item trong sequence.

### **2. Iterating Over Lists**

- Lists là các collection có thứ tự, nên rất phù hợp cho `for` loop. Hãy tưởng tượng bạn có một list nhiệt độ và muốn in từng giá trị.

````python
temperatures = [19.5, 22.1, 18.0, 25.3]

print("Daily temperatures:")
for temp in temperatures:
    print(f"Temperature recorded: {temp}°C")

print("\nLoop finished.")
````

- Trong ví dụ này:

  - `temperatures` là sequence (một list).
  - `temp` là loop variable.
  - Ở lần lặp đầu tiên, `temp` nhận giá trị `19.5`. Lệnh `print` thụt lề được thực thi.
  - Ở lần lặp thứ hai, `temp` nhận giá trị `22.1`. Lệnh `print` chạy lại.
  - Cứ như vậy cho đến khi phần tử cuối cùng (`25.3`) được xử lý.
  - Sau khi loop kết thúc, chương trình tiếp tục với dòng không thụt lề tiếp theo (`print("\nLoop finished.")`).

### **3. Iterating Over Strings**

- **Strings** là các sequence của ký tự. `for` loop có thể tự động lặp qua từng ký tự trong một string.

````python
user_name = "Alice"

print(f"Characters in the name {user_name}:")
for character in user_name:
    print(f"- {character}")
````

- Ở đây, **loop variable `character`** lần lượt nhận từng ký tự trong string `user_name` trong mỗi lượt lặp: `'A','l', 'i'`, ...

### **4. Repeating Code a Specific Number of Times with `range()`**

- Đôi khi bạn không có sẵn một list hay string để lặp qua, mà chỉ đơn giản muốn lặp một hành động một số lần cố định. Hàm dựng sẵn `range()` của Python rất hữu ích trong trường hợp này. Nó tạo ra một sequence các số để `for` loop có thể lặp qua.


#### **4.1 `range(stop)`**

- `range(stop)`: Giúp tạo ra các số từ 0 đến trước `stop`:

````python
print("Counting from 0 to 4:")
for i in range(5): # Generates 0, 1, 2, 3, 4
    print(i)
````

- **Lưu ý**:
  -  `range(5)` tạo ra các số từ `0` đến `4`, tức là tổng cộng 5 con số. Cách đánh số bắt đầu từ 0 này rất phổ biến trong lập trình.
  - Biến `i` thường được dùng làm **loop counter** đơn giản, nhưng bạn có thể dùng bất kỳ tên biến hợp lệ nào.

#### **4.2 `range(start, stop)`**

- `range(start, stop)`: Tạo ra các số bắt đầu từ `start` đến trước `stop`

````python
print("\nNumbers from 2 to 5:")
for num in range(2, 6): # Generates 2, 3, 4, 5
    print(num)
````

#### **4.3 `range(start, stop, step)`**

- Tạo ra các số bắt đầu từ `start`, đến trước `stop`, tăng thêm `step` sau mỗi lần.

````python
print("\nOdd numbers less than 10:")
for odd_num in range(1, 10, 2): # Generates 1, 3, 5, 7, 9
    print(odd_num)
````

- Dùng `range()` với `for` loop là cách chuẩn để thực hiện các việc như chạy mô phỏng trong 100 bước, xử lý 10 phần tử đầu tiên của một dataset, hoặc đơn giản là in một thông báo nhiều lần.

### **5. Practical Example: Summing List Elements**

- Hãy kết hợp `for` loop với những gì ta đã học về **variables** và **operators** để tính tổng các số trong một list.

````python
scores = [88, 92, 75, 98, 85]
total_score = 0 # Initialize accumulator variable

for score in scores:
    total_score = total_score + score # Add current score to total

print(f"The list of scores is: {scores}")
print(f"The total score is: {total_score}")
````

- Trong ví dụ này, `total_score` đóng vai trò là một **accumulator (biến tích lũy)**. Nó bắt đầu từ `0`, và ở mỗi lần lặp, giá trị `score` hiện tại từ **list scores** được cộng vào.

### **6. When to Use `for` vs. `while`**

- Dùng **`for` loop**:
  - Khi bạn muốn lặp qua mọi item trong một **sequence** đã biết trước (như `list`, `string`, hoặc `range()`)
  - Hoặc khi bạn cần lặp một hành động một số lần xác định. for sẽ tự quản lý quá trình lặp cho bạn.

- Dùng **`while` loop**:
  - Khi bạn cần lặp một hành động chừng nào một **condition** còn đúng, và bạn không nhất thiết biết trước loop sẽ chạy bao nhiêu lần

## **#5: Controlling Loops: `break` and `continue`**

- Trong khi `for` và `while` loops cung cấp cách mạnh mẽ để lặp lại các block code, đôi khi bạn cần kiểm soát chi tiết hơn đối với cách chúng chạy.

- Bạn có thể muốn thoát khỏi loop hoàn toàn trước khi nó chạy xong, hoặc bỏ qua phần còn lại của lần lặp hiện tại và chuyển thẳng sang lần lặp tiếp theo.

- Python cung cấp hai câu lệnh cho đúng những tình huống này: `break` và `continue`.

### **1. Exiting Early: The `break` Statement**

- Câu lệnh `break` kết thúc **current loop (vòng lặp hiện tại)** sớm.

- Ngay khi Python gặp `break` bên trong một `for` hoặc `while` loop, nó lập tức thoát khỏi loop đó, và chương trình tiếp tục chạy từ câu lệnh ngay sau **loop block**.

- Điều này đặc biệt hữu ích khi bạn đang tìm kiếm một thứ gì đó và muốn dừng ngay khi tìm thấy nó. VD:
  - Tìm vị trí xuất hiện đầu tiên của một số cụ thể trong list:

````python
numbers = [1, 5, 12, 8, 9, 21, 5, 3]
target = 9
found_at_index = -1 # Initialize with a value indicating not found

print(f"Searching for {target} in {numbers}...")

for index, num in enumerate(numbers):
    print(f"Checking index {index}: value {num}")
    if num == target:
        found_at_index = index
        print(f"Found {target}!")
        break # Exit the loop immediately

# Code here runs after the loop finishes or breaks
if found_at_index != -1:
    print(f"Target {target} first found at index {found_at_index}.")
else:
    print(f"Target {target} not found in the list.")

print("Search complete.")
````

- **Lưu ý**: `loop` đã dừng ở **index `4`**. Các giá trị `21`, `5`, và `3` không bao giờ được kiểm tra vì câu lệnh `break` gây ra việc thoát ngay lập tức khi `num == target` trở thành đúng.

### **2. Skipping an Iteration: The `continue` Statement**

- Câu lệnh `continue` khác với `break`:
  - Thay vì kết thúc toàn bộ loop, `continue` bỏ qua phần code còn lại bên trong **current iteration (lần lặp hiện tại)** và ngay lập tức chuyển sang lần lặp tiếp theo của loop:
    - Với **`while` loop**, điều này có nghĩa là nhảy về lại bước kiểm tra **loop condition (điều kiện vòng lặp)**.
    - Với **`for` loop**, điều này có nghĩa là chuyển sang **item tiếp theo trong sequence**.

- Điều này rất hữu ích khi bạn gặp dữ liệu hoặc condition trong một lần lặp mà bạn muốn bỏ qua hoặc xử lý khác đi, nhưng vẫn muốn tiếp tục xử lý các item sau đó. VD:
  - Hãy tưởng tượng bạn muốn tính tổng chỉ các số dương trong list, bỏ qua các số âm hoặc số 0:

````python
data = [10, -5, 20, 0, -15, 30, 5]
positive_sum = 0

print(f"Processing data: {data}")

for num in data:
    print(f"Current number: {num}")
    if num <= 0:
        print("  Skipping non-positive number.")
        continue # Skip the rest of this iteration
    # This code only runs if continue was NOT executed
    print(f"  Adding {num} to sum.")
    positive_sum += num

print(f"\nSum of positive numbers: {positive_sum}")
````

### **3. `break` vs. `continue`**

- `break`: **Thoát khỏi loop hoàn toàn**. Việc thực thi tiếp tục sau loop.
- `continue`: **Bỏ qua iteration hiện tại**. Việc thực thi tiếp tục ở đầu iteration tiếp theo, hoặc bước kiểm tra điều kiện của loop.

### **4. Nested Loops**

- Điều quan trọng cần nhớ là `break` và `continue` chỉ ảnh hưởng đến **innermost loop (vòng lặp trong cùng)** nơi chúng xuất hiện. Nếu bạn có **nested loops (vòng lặp lồng nhau**), một `break` hoặc `continue` trong **inner loop** sẽ không ảnh hưởng đến **outer loop**.

````python
for i in range(1, 4): # Outer loop
    print(f"Outer loop iteration {i}:")
    for j in ['a', 'b', 'c', 'STOP', 'd']: # Inner loop
        if j == 'STOP':
            print("  Inner loop breaking!")
            break # Breaks only the inner loop
        if j == 'b':
            print(f"  Skipping '{j}' in inner loop...")
            continue # Continues to next item in inner loop
        print(f"  Processing inner item: {j}")
    # This line is reached after the inner loop completes or breaks
    print(f"Outer loop iteration {i} finished.")
print("All loops complete.")
````

- **Lưu ý**:
  - `break` chỉ dừng việc xử lý `['a', 'b', 'c', 'STOP', 'd']` trong **inner loop**, còn **outer loop** vẫn tiếp tục các lần lặp của nó.
  - Tương tự, `continue` chỉ bỏ qua việc in `b` trong mỗi lần chạy **inner loop**.
