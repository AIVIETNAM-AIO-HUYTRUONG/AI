### **1. Creating an Object**

- Tạo một **instance** của **class** trong Python rất đơn giản. Bạn chỉ cần gọi tên class kèm theo `()`:

## **#4: Attributes: Storing Data in Objects**

- Objects là những cấu trúc cơ bản trong lập trình, được định nghĩa bởi một **blueprint (bản thiết kế)** bằng từ khóa `class`, từ đó ta tạo ra các **instances (thể hiện)** riêng lẻ. Điều làm cho objects trở nên hữu ích là khả năng lưu **trữ unique data (dữ liệu riêng biệt)**.

- Ví dụ, một **customer object** trong một online store là một thực thể cụ thể. **Customer object** này cần lưu các thông tin riêng của nó:
    - Tên
    - Địa chỉ
    - Email
    - Lịch sử mua hàng
    - Và các chi tiết khác

- Dữ liệu được lưu bên trong object này được gọi là **attributes (thuộc tính)**

- **Attributes** thực chất là các **variables (biến)** thuộc về một **instance** cụ thể của `class`. Chúng biểu diễn **state (trạng thái)** hoặc đặc điểm của object đó.

### **1. Assigning & Accessing Attributes**

- Bạn có thể thêm **attributes** cho một object sau khi nó đã được tạo bằng cách dùng **dot notation**:

- Sau khi **attributes** đã được gán, bạn có thể truy cập giá trị của chúng bằng cùng **dot notation**
    ```python
    object.attribute_name = value
    ```

````python
# Define a simple class (blueprint)
class Dog:
    # For now, this class definition is empty
    pass

# Create an instance (an actual dog object)
my_dog = Dog()

# Now, let's give this specific dog some attributes
my_dog.name = "Fido"
my_dog.breed = "German Shepherd"
my_dog.age = 4

# Create another instance
another_dog = Dog()
another_dog.name = "Buddy"
another_dog.breed = "Golden Retriever"
another_dog.age = 2

# Accessing attributes of my_dog
print(f"My dog's name is: {my_dog.name}")
print(f"Its breed is: {my_dog.breed}")
print(f"And it is {my_dog.age} years old.")

# Accessing attributes of another_dog
print(f"\nMy other dog's name is: {another_dog.name}")
print(f"Its breed is: {another_dog.breed}")
````

- Trong code này:

    - Ta định nghĩa một **class `Dog`** rỗng. Nó chỉ là một **blueprint** tạm thời.
    - Ta tạo hai **Dog objects** khác nhau: `my_dog` và `another_dog`.
    - Sau đó ta gán giá trị cho các attributes như name, breed, và age cho từng object cụ thể bằng **dot notation** (`my_dog.name = "Fido"`).

## **#5: Methods: Defining Behavior for Objects**

- Hãy xem **methods** như **những functions “thuộc về” một class**. Chúng được định nghĩa bên trong **class block**, giống như **attributes**, nhưng vẫn dùng từ khóa quen thuộc `def` mà ta đã học khi tạo các **standalone functions**.

### **1. Defining a Method**

- VD: Để thêm một method - behavior vào **class `Dog`**. Chúng ta mong muốn bất kỳ `Dog` object nào cũng có thể "bark"

````python
class Dog:
    # The __init__ method initializes the object's attributes
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed
        print(f"Dog initialized: {self.name}")

    # This is a method: a function defined inside a class
    def bark(self):
        print(f"{self.name} says Woof!")

# Create an instance (object) of the Dog class
my_dog = Dog("Buddy", "Golden Retriever")

# Call the bark method on the my_dog object
my_dog.bark()
````

### **2. The `self` Parameter: Referring to the Object Itself**

- Khi bạn tạo một **instance** từ một `class`, `self` cho phép bạn truy cập các **attributes** và **methods** của chính instance này, bên trong định nghĩa của `class`.

- Dưới đây là các điểm chính:

  - **Tham chiếu đối tượng**: Khi bạn gọi một **method** của **instance**:
    - Ví dụ: `my_object.my_method()`
    - Python tự động truyền `my_object` vào tham số đầu tiên của **mothod** đó. Theo quy ước, chúng ta đặt tên tham số này là `self`.

  - **Phân biệt dữ liệu**: `self` giúp phân biệt giữa các biến cục bộ trong **mothod** và các biến thuộc về **instance**.
    - Ví dụ, `self.name = name` -  gán giá trị của tham số name vào thuộc tính name của đối tượng hiện tại.

  - **Truy cập thành viên**: Bạn cần dùng `self`. để gọi các **mothod** hoặc thuộc tính khác của cùng một class bên trong các **mothod** của class đó.

- Mặc dù `self` là quy ước đặt tên bắt buộc trong cộng đồng Python, về mặt kỹ thuật, bạn có thể đặt tên là bất kỳ từ nào khác, nhưng tuyệt đối nên giữ là `self` để mã nguồn dễ đọc và nhất quán.

````python
def bark(self):
    print(f"{self.name} says Woof!")
````

- Khi ta viết `my_dog.bark()`, Python tự động **truyền object `my_dog` như đối số đầu tiên vào method `bark`**. Vì vậy, **bên trong lời gọi method, `self` chính là `my_dog`**. Đó là lý do method biết phải in ra tên của con chó nào.

- Bạn không cần **truyền `self` một cách tường minh khi gọi method**:
  -  (my_dog.bark(), chứ không phải my_dog.bark(my_dog)).

  - Python tự xử lý việc này. Tuy nhiên, **bạn phải có `self` là tham số đầu tiên khi định nghĩa method trong class**.

### **3. Methods Can Access and Modify Attributes**

- **Methods** là cách chính để bạn tương tác với dữ liệu của object (tức các attributes của nó) - Chúng có **thể đọc giá trị attribute**, như trong **method `bark`**, và **cũng có thể thay đổi chúng**.

````python
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed
        print(f"Dog initialized: {self.name}")

    def bark(self):
        print(f"{self.name} says Woof!")

    # Method to change the dog's name attribute
    def set_name(self, new_name):
        if isinstance(new_name, str) and new_name: # Basic check
            self.name = new_name
            print(f"My name has been changed to {self.name}")
        else:
            print("Please provide a valid name.")

# Create instances
dog1 = Dog("Buddy", "Golden Retriever")
dog2 = Dog("Lucy", "Beagle")

# Use the methods
dog1.bark() # Output: Buddy says Woof!
dog2.bark() # Output: Lucy says Woof!

print(f"Dog 1's name is currently: {dog1.name}") # Output: Dog 1's name is currently: Buddy

# Call the set_name method on dog1
dog1.set_name("Max") # Output: My name has been changed to Max

print(f"Dog 1's name is now: {dog1.name}") # Output: Dog 1's name is now: Max
dog1.bark() # Output: Max says Woof!

print(f"Dog 2's name is: {dog2.name}") # Output: Dog 2's name is: Lucy
````

- Trong method set_name:
  - Nó nhận `self` và `new_name`.
  - Nó thực hiện một kiểm tra đơn giản để xem `new_name` có phải là một string không rỗng.
  - Nếu hợp lệ, nó cập nhật **attribute name** của object bằng `self.name = new_name.`

- Lưu ý rằng khi gọi `dog1.set_name("Max")`, chỉ tên của `dog1` thay đổi. Object `dog2` không bị ảnh hưởng, vì `self` trong lần gọi method đó đã trỏ đến `dog1`

### **4. Methods vs. Functions**

- Nên nghĩ **methods như các functions gắn với objects**.

- Điểm khác biệt chính là:
  - **Methods hoạt động** trong ngữ cảnh của một object cụ thể, thông qua `self`và có thể truy cập trực tiếp cũng như thay đổi các **attributes** của object đó.

  - **Standalone functions** thì không có liên kết tự nhiên như vậy với state của một object.


## **#6: The __init__ Method (Constructor)**

- Khi bạn tạo một object từ một class, bạn thường muốn thiết lập trạng thái ban đầu của object đó ngay lập tức.

- Ví dụ, nếu bạn có **class `Dog`**, bạn có thể muốn xác định tên và giống chó ngay khi tạo **Dog object** thay vì tạo một “chú chó chung chung” rồi mới gán attributes sau đó.

- Python cung cấp một method đặc biệt cho mục đích này: `__init__`. Tên `__init__` có **double underscores (`__`)** ở trước và sau, cho biết đây là một **special method** được Python nhận diện đặc biệt.

- **Method** này sẽ được **gọi tự động mỗi khi bạn tạo một instance (object) mới từ class**. Vì **nó dùng để khởi tạo trạng thái ban đầu cho object** nên nó thường **được gọi là constructor của class**,

````python
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

    def bark(self):
        print(f"{self.name} says: Woof!")

# Tạo object và gán trạng thái ban đầu ngay lập tức
dog1 = Dog("Milo", "Poodle")
dog2 = Dog("Buddy", "Golden Retriever")

print(dog1.name)   # Milo
print(dog1.breed)  # Poodle
dog1.bark()        # Milo says: Woof!

print(dog2.name)   # Buddy
print(dog2.breed)  # Golden Retriever
dog2.bark()        # Buddy says: Woof!
````

- Ý nghĩa là: khi tạo `Dog("Milo", "Poodle")`, Python gọi `__init__` để gán ngay `name` và `breed` cho object đó. Nhờ vậy, object được khởi tạo với dữ liệu đầy đủ, thay vì tạo ra một object rỗng rồi mới gán sau.

````python
class Dog:
    pass

dog = Dog()
dog.name = "Milo"
dog.breed = "Poodle"

print(dog.name)   # Milo
print(dog.breed)  # Poodle
````

- Cách này vẫn chạy, nhưng dễ quên gán thuộc tính, khiến object ở trạng thái không hoàn chỉnh và kém an toàn hơn.

### **1. Defining and Using `__init__`**

- Bạn định nghĩa `__init__` giống như bất kỳ method nào khác bên trong class. Tham số đầu tiên luôn phải là `self`, đại diện cho **instance** cụ thể của class đang được tạo. Sau `self`, bạn có thể định nghĩa thêm các tham số khác để nhận dữ liệu cần dùng cho việc khởi tạo object.

- Bên trong `__init__`, bạn thường gán các giá trị nhận được từ các tham số đó vào attributes của object bằng cú pháp `self.attribute_name = value`.

````python
class Dog:
    def __init__(self, name, breed):
        # This method is called automatically when a Dog object is created
        print(f"Creating a new Dog object named {name}...")

        # Assign the provided name and breed to attributes of the 'self' object
        self.name = name
        self.breed = breed
        self.tricks = [] # Initialize an empty list for tricks

    def add_trick(self, trick):
        self.tricks.append(trick)
        print(f"{self.name} learned a new trick: {trick}!")

# Now, when we create Dog objects, we MUST provide the name and breed
dog1 = Dog("Buddy", "Golden Retriever")
dog2 = Dog("Lucy", "Beagle")

# The attributes were set during creation by __init__
print(f"{dog1.name} is a {dog1.breed}.") # Output: Buddy is a Golden Retriever.
print(f"{dog2.name} is a {dog2.breed}.") # Output: Lucy is a Beagle.

# We can still call other methods
dog1.add_trick("fetch")
print(f"{dog1.name}'s tricks: {dog1.tricks}") # Output: Buddy's tricks: ['fetch']
````

### **2. How it Works**

1. **Instantiation (khởi tạo instance)**: Khi bạn viết `dog1 = Dog("Buddy", "Golden Retriever")`, Python đầu tiên tạo ra một **Dog object** mới, rỗng.
2. **Gọi __init__ (init call__)**: Sau đó, Python tự động gọi method `__init__` trên object vừa tạo.
3. **Truyền đối số (Argument Passing)**: Chính object đó được truyền vào như đối số đầu tiên (`self`), và các đối số bạn cung cấp trong dấu ngoặc đơn (`"Buddy", "Golden Retriever"`) được truyền vào các tham số tiếp theo (`name, breed`).
4. **Gán attributes (Attribute Assignment)**: Bên trong `__init__`,
    - Dòng `self.name = name` gán giá trị **"Buddy" vào attribute name của object dog1**.
    - Tương tự, `self.breed = breed` gán **"Golden Retriever" vào attribute breed**. Dòng `self.tricks = []` khởi tạo attribute tricks là một list rỗng cho riêng instance này.
5. **Return**: Method `__init__` không trả về giá trị tường minh nào (nó ngầm trả về None). Nhiệm vụ của nó là sửa đổi object `self`. Quá trình **instantiation (Dog(...))** trả về object đã được khởi tạo đầy đủ, rồi object đó được gán cho biến `dog1`.
