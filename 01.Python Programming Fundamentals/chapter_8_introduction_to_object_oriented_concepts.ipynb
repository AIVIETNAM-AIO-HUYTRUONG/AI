## **#1: Thinking in Objects: Classes and Objects**

- **Procedural programming (lập trình thủ tục)** tổ chức code bằng các **functions** để thực hiện nhiệm vụ, và dùng các **control structures (cấu trúc điều khiển)** như loops và conditionals để quản lý luồng thực thi. Cách tiếp cận này nhấn mạnh việc viết các **procedures** hoặc **functions** thao tác trên dữ liệu.

- **Object-Oriented Programming - OOP (lập trình hướng đối tượng)** đưa ra một góc nhìn khác. Thay vì chủ yếu nghĩ về các bước **procedures** mà chương trình sẽ đi qua. Bạn nên nghĩ về các **“things”** (đối tượng/sự vật/khái niệm) mà chương trình làm việc với. Những “things” này được gọi là objects (đối tượng). VD:
  - Nếu bạn đang xây dựng một ứng dụng quản lý cửa hàng thú cưng, các **“things”** sẽ là:
    - **Một con chó** Nó có tên, giống, tuổi (dữ liệu - attributes) và nó có thể sủa, chạy, ăn (hành động - methods).
    - **Một con mèo**: Nó cũng có tên, giống, tuổi và hành động (kêu meo meo, vồ chuột).
  - Trong code, "con chó" hay "con mèo" này chính là các **“things”** / **Objects**.

- Mỗi object này có hai đặc điểm chính:
  1. **State (trạng thái / dữ liệu)**: object là gì hoặc có gì:
      - Một con chó có tên, giống chó, và tuổi. Một tài khoản ngân hàng có số tài khoản và số dư. Những thứ này thường được gọi là **attributes (thuộc tính)** hoặc **properties (tính chất)**.

  2. **Behavior (hành vi)**: object có thể làm gi?
      - Một con chó có thể sủa, vẫy đuôi, hoặc ăn. Một tài khoản ngân hàng có thể được nạp tiền hoặc rút tiền. Những hành động này được cài đặt dưới dạng functions gắn với object, gọi là **methods (phương thức)**.
  


#### **1.1 What is Class?**

- Một class hoạt động như một **blueprint (bản thiết kế)**, **template (khuôn mẫu)**, hoặc **recipe (công thức)** để tạo ra objects. Nó định nghĩa cấu trúc chung và **behavior chung** mà tất cả objects thuộc cùng một loại sẽ có.

- VD: Ta có thể định nghĩa một **class `Dog`**.
  - Định nghĩa **class** này sẽ chỉ ra rằng tất cả những con chó mà ta tạo trong chương trình sẽ có các **attributes** như `name` và `breed`, và các **methods** như `bark()`:
  - Bản thân class không phải là một con chó; nó chỉ là bản thiết kế để tạo ra chó

#### **1.2. What is Object?**

- Một **object (đối tượng)** hay gọi là **instance (thể hiện)** là một thực thể cụ thể được tạo ra từ một class.

- Dùng class `Dog` làm **blueprint**, ta có thể tạo ra từng **`dog` object** riêng lẻ.

- VD: Ta có thể tạo một **dog object** tên `"Buddy"` thuộc giống `"Golden Retriever"` và một **dog object** khác tên `"Lucy"` thuộc giống `"Poodle"`. Cả `Buddy` và `Lucy` đều là **objects** được tạo từ **class `Dog`**. Chúng chia sẻ cùng một bộ **attributes (name, breed)** và **methods (bark)**, nhưng giá trị của các **attributes** đó thì riêng biệt cho từng object.

  - Nó giúp tổ chức code tốt hơn, đặc biệt trong các chương trình lớn, bằng cách nhóm những thông tin và chức năng có liên quan với nhau.

## **#2: Defining a Class**

- **Object-oriented programming** được tổ chức code xung quanh các **objects** - tức là kết hợp giữa data và các functions thao tác trên data đó.
- Để tạo ra các object này trước tiên ta cần định nghĩa một **template** mô tả object sẽ chứa loại data nào và có thể thực hiện hành động gì. Và tất cả điều này gọi là **bản thiết kế - Class** trong Python


- Bạn hãy tưởng tượng một **class** giống một bản thiết kế kiến trúc cho một ngôi nhà, nó chứa toàn bộ thông tin cần thiết để xây dựng một hoặc nhiều ngôi nhà giống nhau.

- Một **class** không trức tiếp tạo ra các **objects**, mà nó định nghĩa cấu trúc và behavior cho tất cả objects đó.

### **1. Syntax for defining classes in Python**

- Để định nghĩa một class trong Python, từ dùng **`class` keyword**, theo sau là **ClassName** và cuối cùng là **`:`**
- Phần code tạo nên class sẽ được **thụt lề (indented)** giống như **functions** và các **control flow statements**

- Quy ước trong Python là đặt **ClassName** theo kiểu **CamelCase** (vd: `MyClass` , `NetwrokConnection`)

    ``` python
    class ClassName:
        pass
    ```
- Hãy phân tích:

    - `class` : từ khóa này báo hiệu bắt đầu một **class**.
    - `ClassName` : thay bằng tên thật bạn chọn cho class của mình; nên theo quy ước **CamelCase**.
    - `:` : dấu hai chấm cho biết bắt đầu một khối code thụt lề chứa **class body**.
    - `pass` : đây là một **Python statement** không làm gì cả. Nó đóng vai trò như một **placeholder (chỗ giữ tạm)**


### **2. What will be inside the class?**

- Ở phần bên trong class sẽ có:
    - **Attributes (thuộc tính)**: Là các variables lưu dữ liệu liên quan đến object - VD: `tên mèo`, `giống mèo` , `tuổi`, ...
    - **Methods (phương thức)**: Là các functions được định nghĩa bên trong class để mô tả hành vi của object - VD: `mèo kêu`, `mèo chạy`, ....

## **#4: Creating Instances (Objects)**

- Trong lập trình Python, quá trình tạo ra một "thing" từ **template** của **class** được gọi là **instantation** và "thing" bạn tạo ra được gọi là **object** hoặc **instance** của class đó.

### **1. Creating an Object**

- Tạo một **instance** của **class** trong Python rất đơn giản. Bạn chỉ cần gọi tên class kèm theo `()`:

````python
# Let's assume we have a simple class defined like this:
class Dog:
  # For now, this class doesn't define any specific
  # attributes or behaviors. We'll add those soon.
  pass # The 'pass' keyword means "do nothing"

# Now, let's create instances (objects) of the Dog class
fido = Dog()
buddy = Dog()

# fido and buddy are now two distinct Dog objects
print(fido)
print(buddy)
print(type(fido))
````

- Khi bạn chạy code này, `fido` và `buddy` sẽ trở thành các biến **tham chiếu đến hai Dog objects khác nhau trong bộ nhớ của máy tính**:

- `<__main__.Dog object at 0x....>`, trong đó `0x....` là một **memory address (địa chỉ bộ nhớ) riêng cho mỗi object.**. Điều này xác nhận rằng chúng là các **instances** khác nhau được tạo từ **class `Dog`**.

- `type()` xác nhận rằng fido đúng là có kiểu `Dog`.

### **2. The Role of Parentheses `()`**

- Có thể bạn sẽ thắc mắc vì sao cần dấu `()` sau tên class, như `Dog()`?
    - Khi bạn gọi một class như vậy, thực chất bạn đang kích hoạt quá trình tạo object.
    - Python sẽ tìm một **special initialization method (phương thức khởi tạo đặc biệt)** bên trong class — chính là `__init__` — để thiết lập trạng thái ban đầu cho object mới.
