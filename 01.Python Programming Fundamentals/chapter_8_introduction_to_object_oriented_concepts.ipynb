## **#1: Thinking in Objects: Classes and Objects**

- **Procedural programming (lập trình thủ tục)** tổ chức code bằng các **functions** để thực hiện nhiệm vụ, và dùng các **control structures (cấu trúc điều khiển)** như loops và conditionals để quản lý luồng thực thi. Cách tiếp cận này nhấn mạnh việc viết các **procedures** hoặc **functions** thao tác trên dữ liệu.

- **Object-Oriented Programming - OOP (lập trình hướng đối tượng)** đưa ra một góc nhìn khác. Thay vì chủ yếu nghĩ về các bước **procedures** mà chương trình sẽ đi qua. Bạn nên nghĩ về các **“things”** (đối tượng/sự vật/khái niệm) mà chương trình làm việc với. Những “things” này được gọi là objects (đối tượng). VD:
  - Nếu bạn đang xây dựng một ứng dụng quản lý cửa hàng thú cưng, các **“things”** sẽ là:
    - **Một con chó** Nó có tên, giống, tuổi (dữ liệu - attributes) và nó có thể sủa, chạy, ăn (hành động - methods).
    - **Một con mèo**: Nó cũng có tên, giống, tuổi và hành động (kêu meo meo, vồ chuột).
  - Trong code, "con chó" hay "con mèo" này chính là các **“things”** / **Objects**.

- Mỗi object này có hai đặc điểm chính:
  1. **State (trạng thái / dữ liệu)**: object là gì hoặc có gì:
      - Một con chó có tên, giống chó, và tuổi. Một tài khoản ngân hàng có số tài khoản và số dư. Những thứ này thường được gọi là **attributes (thuộc tính)** hoặc **properties (tính chất)**.

  2. **Behavior (hành vi)**: object có thể làm gi?
      - Một con chó có thể sủa, vẫy đuôi, hoặc ăn. Một tài khoản ngân hàng có thể được nạp tiền hoặc rút tiền. Những hành động này được cài đặt dưới dạng functions gắn với object, gọi là **methods (phương thức)**.
  


#### **1.1 What is Class?**

- Một class hoạt động như một **blueprint (bản thiết kế)**, **template (khuôn mẫu)**, hoặc **recipe (công thức)** để tạo ra objects. Nó định nghĩa cấu trúc chung và **behavior chung** mà tất cả objects thuộc cùng một loại sẽ có.

- VD: Ta có thể định nghĩa một **class `Dog`**.
  - Định nghĩa **class** này sẽ chỉ ra rằng tất cả những con chó mà ta tạo trong chương trình sẽ có các **attributes** như `name` và `breed`, và các **methods** như `bark()`:
  - Bản thân class không phải là một con chó; nó chỉ là bản thiết kế để tạo ra chó

#### **1.2. What is Object?**

- Một **object (đối tượng)** hay gọi là **instance (thể hiện)** là một thực thể cụ thể được tạo ra từ một class.

- Dùng class `Dog` làm **blueprint**, ta có thể tạo ra từng **`dog` object** riêng lẻ.

- VD: Ta có thể tạo một **dog object** tên `"Buddy"` thuộc giống `"Golden Retriever"` và một **dog object** khác tên `"Lucy"` thuộc giống `"Poodle"`. Cả `Buddy` và `Lucy` đều là **objects** được tạo từ **class `Dog`**. Chúng chia sẻ cùng một bộ **attributes (name, breed)** và **methods (bark)**, nhưng giá trị của các **attributes** đó thì riêng biệt cho từng object.
