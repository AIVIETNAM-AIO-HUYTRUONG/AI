## **#1: What are Modules?**

- Khi chương trình Python của bạn trở nên lớn và phức tạp, việc đặt toàn bộ code vào một file duy nhất sẽ nhanh chóng trở nên khó quản lý.

- Hãy tưởng tượng bạn phải **tạo** một **function giải quyết logic toán học đơn giản** trong một file dài hàng nghìn dòng, hoặc muốn tái sử dụng một function hữu ích bạn đã viết từ project này sang một project khác mà không cần **copy & paste**.
- Đây là lúc **modules của Python** phát huy tác dụng.

- Một module trong Python đơn giản là một file (đuôi `.py`) chứa các **function, class, hoặc variables** được định nghĩa sẵn.

- Modules có vài mục đích rất quan trọng:
  1. **Organization (tổ chức)**: Chúng cho phép bạn **nhóm các code liên quan với nhau một cách logic**.
    - VD: Bạn có thể **đặt tất cả các function hỗ trợ toán học vào một module** -> Điều này làm cho **codebase dễ hiểu và dễ điều hướng hơn**.
  2.  **Reusability (khả năng tái sử dụng)**: khi bạn **đã viết và kiểm thử code trong một module**, bạn có thể **dùng lại nó trong project khác mà không cần viết lại từ đầu**. -> Điều này **giúp phát triển hiệu quả hơn** và **giảm lỗi**.

  3. **Namespace Separation (phân tách namespace)**:
    - Modules tạo ra các **separate namespaces (namespace riêng biệt)**, tức là scope của các tên như variable và function có phạm vi riêng -> Nghĩa là một function tên `calculate` trong module `utilities` sẽ không xung đột một function khác tên `calculate` trong một module khác hoặc file chính của bạn.  
    - Mỗi module tự quản lý các tên của nó, giúp tránh **naming conflicts (xung đột tên)**

### **1. Modular thinking**
- Thay vì viết một chương trình lớn, bạn:
  - Chia thành nhiều module nhỏ
  - Mỗi module xử lý một nhiệm vụ cụ thể
  - Sau đó kết hợp lại

- Ví dụ, module `utilities.py`

```python
# utilities.py

def calculate_area(length, width):
  """Calculates the area of a rectangle."""
  return length * width

def format_greeting(name):
  """Creates a simple greeting string."""
  return f"Hello, {name}!"

# ... other utility functions ...
```

- File `utilities.py` này là một module.
- Nó chứa các function liên quan với nhau:
  - `calculate_area`
  - `format_greeting`.
