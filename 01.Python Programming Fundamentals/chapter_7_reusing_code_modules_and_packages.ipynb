## **#1: What are Modules?**

- Khi chương trình Python của bạn trở nên lớn và phức tạp, việc đặt toàn bộ code vào một file duy nhất sẽ nhanh chóng trở nên khó quản lý.

- Hãy tưởng tượng bạn phải **tạo** một **function giải quyết logic toán học đơn giản** trong một file dài hàng nghìn dòng, hoặc muốn tái sử dụng một function hữu ích bạn đã viết từ project này sang một project khác mà không cần **copy & paste**.
- Đây là lúc **modules của Python** phát huy tác dụng.

- Một module trong Python đơn giản là một file (đuôi `.py`) chứa các **function, class, hoặc variables** được định nghĩa sẵn.

- Modules có vài mục đích rất quan trọng:
  1. **Organization (tổ chức)**: Chúng cho phép bạn **nhóm các code liên quan với nhau một cách logic**.
    - VD: Bạn có thể **đặt tất cả các function hỗ trợ toán học vào một module** -> Điều này làm cho **codebase dễ hiểu và dễ điều hướng hơn**.
  2.  **Reusability (khả năng tái sử dụng)**: khi bạn **đã viết và kiểm thử code trong một module**, bạn có thể **dùng lại nó trong project khác mà không cần viết lại từ đầu**. -> Điều này **giúp phát triển hiệu quả hơn** và **giảm lỗi**.

  3. **Namespace Separation (phân tách namespace)**:
    - Modules tạo ra các **separate namespaces (namespace riêng biệt)**, tức là scope của các tên như variable và function có phạm vi riêng -> Nghĩa là một function tên `calculate` trong module `utilities` sẽ không xung đột một function khác tên `calculate` trong một module khác hoặc file chính của bạn.  
    - Mỗi module tự quản lý các tên của nó, giúp tránh **naming conflicts (xung đột tên)**

### **1. Modular thinking**
- Thay vì viết một chương trình lớn, bạn:

## **#2: Importing Modules: import Statement**

- Để dùng code từ một file trong file khác, Python cung cấp **`import` statement**. Đây là cơ chế cơ bản để tái sử dụng và sử dụng code được định nghĩa trong các module khác

### **1. The Basic `import` Statement**

- Cách đơn giản nhất để dùng một module là dùng **`import` keyword** theo sau là tên file module.

- VD: Nếu bạn muốn dùng các hàm toán học, Python có sẵn một module dựng sẵn tên là `math`, chứa nhiều **mathematical operations** và **constants**. Để dùng nó, bạn bắt đầu script bằng:
  ```python
  import math
  ```

- Dòng này thực sự làm gì?
  - **Finds the Module (tìm module)**: Python tìm một file tên `math.py` trong danh sách các thư mục mà nó biết. Danh sách này gồm thư mục chứa script hiện tại và các vị trí của **standard library**.

  - **Executes the Module (thực thi module nếu cần)**: Nếu module chưa được nạp trong lần chạy hiện tại, Python sẽ chạy code bên trong `math.py`. Việc này làm cho tất cả **functions (hàm)**, **variables (biến)**, và **classes (lớp)** được định nghĩa trong `math.py` sẵn sàng để dùng.

  - **Creates a Namespace (tạo namespace)**: Đây là điểm rất quan trọng. `import math` không copy trực tiếp tất cả các hàm như `sqrt` hay hằng số như `pi` vào workspace chính của script. Thay vào đó, nó tạo ra một tên duy nhất trong namespace của script: `math`. Tên `math` này bây giờ trỏ tới chính module object.

### **2. Accessing Module Contents: Dot Notation**

- Vì `import math` tạo một namespace tên `math`, bạn cần nói rõ cho Python biết nó phải tìm function hoặc variables ở đâu trong modules đó. Bạn làm điều này bằng **dot notation**: `module_name.item_name`.

- Để dùng `sqrt` từ module `math`:
  ```python
  math.sqrt(16)
  ```
- Để truy cập hằng số `pi`, bạn viết:
  ```python
  math.pi
  ```

- Việc thêm tiền tố rõ ràng `module_name`. rất có ích vì nó giúp tránh **naming conflicts**. Nếu bạn tự định nghĩa một biến tên `pi` trong script, nó sẽ không xung đột với `math.pi` vì chúng nằm trong các namespace khác nhau. `pi` của bạn nằm trong namespace của main script, còn hằng số toán học được truy cập thông qua namespace `math`.

### **3. Example: Using the `math` Module**

- Dưới đây là một script hoàn chỉnh, đơn giản, minh họa việc import và sử dụng:

````python
# Import the entire math module
import math

# Calculate the square root of 25
number = 25
square_root = math.sqrt(number)
print(f"The square root of {number} is {square_root}") # Output: The square root of 25 is 5.0

# Calculate the area of a circle with radius 3
radius = 3
area = math.pi * (radius ** 2) # Using math.pi and the power operator
print(f"The area of a circle with radius {radius} is {area}") # Output: The area of a circle with radius 3 is 28.274333882308138
````

- Trong ví dụ này:

  - `import math` làm module `math` có thể sử dụng được.
  - `math.sqrt()` gọi hàm sqrt bên trong module `math`.
  - `math.pi` truy cập hằng số `pi` từ module `math`.

### **4. Importing Multiple Modules**

- Nếu bạn cần hàm hoặc biến từ nhiều module khác nhau, bạn có thể import chúng riêng lẻ. Thực hành chuẩn là đặt mỗi câu lệnh import trên một dòng riêng ở đầu file:

````python
import math
import random # Another standard library module for random numbers
import os     # Module for interacting with the operating system

print(math.sqrt(100))
print(random.randint(1, 10)) # Get a random integer between 1 and 10
print(os.getcwd())           # Get the current working directory
````

- Cách dùng `import module_name` đơn giản này, kết hợp với **dot notation (ký hiệu dấu chấm)**, là cách phổ biến và được khuyến nghị nhất để đưa code bên ngoài vào Python script của bạn, vì nó rõ ràng và giúp tránh **name collisions (xung đột tên)**.

## **#3: Importing Specific Names: `from ... import`**

- Trong khi **`import module_name` (import toàn bộ module)** yêu cầu bạn luôn dùng tiền tố như `math.pi` hoặc `random.randint()`. Cách này là một cách code tốt vì nó giúp làm rõ nguồn gốc và tránh **naming conflicts (xung đột tên)** ngoài ý muốn.

- Bên cạnh đó Python cũng cung cấp một cách khác đó là **`from ... import`(import cụ thể)** cách này có thể giúp bạn lấy ra những tên **function, variable, class,...** cụ thể mà không cần phải dùng đến tiền tố module.

### **1. The `from ... import` Syntax**.

- Cấu trúc cơ bản như sau:
    ```python
    from module_name import name1, name2, ...
    ```

- VD: Nếu bạn dùng `import module` thì code sẽ như sau

````python
import math

radius = 5
area = math.pi * (radius ** 2)
hypotenuse = math.sqrt(3**2 + 4**2)

print(f"Area: {area}")
print(f"Hypotenuse: {hypotenuse}")
````

- Với `from ... import`, ta có thể viết:

````python
from math import pi, sqrt

radius = 5
# Notice no 'math.' prefix is needed now
area = pi * (radius ** 2)
hypotenuse = sqrt(3**2 + 4**2)

print(f"Area: {area}")
print(f"Hypotenuse: {hypotenuse}")
# Output:
# Area: 78.53981633974483
# Hypotenuse: 5.0
````

### **2. The Trade-off: Namespaces and Clarity**

- Sự tiện lợi của `from ... import` đi kèm với một cái giá tiềm năng là **namespace pollution** và giảm sự rõ ràng.

- **Namespace (không gian tên)** giống như một dictionary nơi Python lưu các tên hiện có và có thể truy cập, như variables, functions, classes.
  - Khi bạn dùng `import math`, Python tạo một entry là `math`, và mọi nội dung của module được truy cập qua entry đó, ví dụ `math.pi`
  - Các tên bên trong module `math` không trộn trực tiếp với các tên do bạn định nghĩa trong script.

- Tuy nhiên, khi dùng `from math import pi`, chính tên `pi` được thêm trực tiếp vào **main namespace (namespace chính)** của script. Nếu trong chương trình Python hiện tại có sẵn một `pi` variable,  thì nó sẽ bị ghi đè bởi `import math`, dẫn đến hành vi không mong muốn hoặc lỗi.

````python
# Define our own 'sqrt' function (perhaps poorly)
def sqrt(number):
  print(f"Maybe finding the square root of {number}?")
  return number / 2 # Incorrect calculation!

# Let's calculate something else unrelated first
result1 = sqrt(100)
print(f"Our initial sqrt(100) result: {result1}")

# Now, import the real sqrt from the math module
from math import sqrt

# Try using sqrt again - which one runs?
result2 = sqrt(100)
print(f"Result after importing math.sqrt: {result2}")

# Output:
# Maybe finding the square root of 100?
# Our initial sqrt(100) result: 50.0
# Result after importing math.sqrt: 10.0
````

- Câu lệnh `from math import sqrt` đã ghi đè function `sqrt` tự tạo của chúng ta bằng **function** từ **module math**. Ví dụ này hơi đơn giản hóa, nhưng trong các project lớn với nhiều variables và functions đây là một điều cấm kỵ.

### **3. Importing Everything: `from module import *` (Use with Caution!)**


- Python có một biến thể khác: `from module_name import *`
- Dấu `*` là **wildcard (ký tự đại diện)**, có nghĩa là **import tất cả**.

````python
# Caution: Generally discouraged!
from math import *

# Now all names from math are directly available
print(pi)
print(sqrt(16))
print(cos(0))
# Output:
# 3.141592653589793
# 4.0
# 1.0
````

- Cách này là **import toàn bộ public names (tên công khai)** từ **module** đó vào **namespace hiện tại** của bạn. Nó có thể **trông tiện khi dùng trong môi trường thử nghiệm nhanh chóng hoặc script rất nhỏ** - Nhưng bị **strongly discouraged (rất không khuyến nghị) trong chương trình lớn**:
  1. **High Risk of Namespace Collisions (rủi ro xung đột tên cao)**: nó đưa hàng chục hoặc hàng trăm tên vào namespace của bạn, khiến xung đột tên rất dễ xảy ra.
  2. **Reduced Readability (giảm khả năng đọc hiểu)**:
    - Rất khó biết một function hay variable đến từ đâu.
    - Nhìn `sqrt(100)` thì không rõ đó là **built-in function, function bạn tự định nghĩa** hay **function import từ math (hoặc module khác dùng `*`)**

### **4. Import với alias: `from ... import ... as ...`**

- Nếu bạn muốn **import một tên cụ thể nhưng nó xung đột với tên đã tồn tại** trong script, hoặc **bạn chỉ đơn giản muốn một tên khác dễ hiểu hơn** thì hãy dùng `as` (trong đó `as` hoạt động như một **alias**).

````python
from math import pi as mathematical_pi

# Use our own variable named 'pi'
pi = 3.14 # A less precise version, maybe for specific needs

radius = 5
# Use the aliased import for the precise value
area = mathematical_pi * (radius ** 2)

print(f"Our pi: {pi}")
print(f"Imported pi: {mathematical_pi}")
print(f"Calculated area: {area}")

# Output:
# Our pi: 3.14
# Imported pi: 3.141592653589793
# Calculated area: 78.53981633974483
````

### **5. When to Use Which Import Style?**

- `import module_name`: đây là cách phổ biến và thường được khuyến nghị nhất.

  - Giữ namespace riêng và sạch.
  - Code rõ ràng, dễ đọc hơn vì luôn biết `module_name.function()` đến từ đâu.
  - Phù hợp khi bạn cần nhiều item từ một module hoặc muốn độ rõ ràng tối đa.

- `from module_name import name1, name2`: dùng khi:

  - Bạn chỉ cần vài item cụ thể từ module.
  - Bạn sẽ dùng các item đó thường xuyên, và việc phải gõ `module_name`. quá nhiều gây rối mắt.
  - Bạn chắc chắn không có xung đột tên, hoặc bạn xử lý bằng `as`.

- `from module_name import name as alias`: hữu ích khi:

  - Muốn tránh xung đột tên.
  - Muốn đổi tên ngắn hơn hoặc dễ hiểu hơn khi dùng `from ... import`.

- `from module_name import *`:
  - nên tránh trong code production hoặc bất kỳ script nào có kích thước đáng kể, vì vấn đề namespace pollution và readability.
  - Nó chỉ có thể tạm chấp nhận trong một số thử nghiệm nhanh khi làm việc tương tác với Python.

## **#4: Python's Standard Library Overview**

- Một trong những điểm mạnh lớn của Python là nó **batteries included (đã có sẵn rất nhiều công cụ)**. Điều này có nghĩa là **khi bạn cài Python**, bạn cũng **nhận được một bộ lớn các modules (mô-đun) hữu ích đi kèm**. **Tập hợp này** được gọi là **Python Standard Library (thư viện chuẩn của Python)**

- Thư viện này giúp bạn tiếp kiệm rất nhiều công sức. Thay vì phải tự viết code lại từ đầu cho các thao tác phổ biến như việc **tính toán, xử lý text** hoặc tương tác với **operating system**,... Các modules này được viết bởi **nhóm phát triển cốt lõi của Python** và các cộng tác viên khác

- **Standard library** rất lớn và bao phủ nhiều loại công việc khác nhau. Bạn không cần phải học hết mà chỉ biết rằng nó có tồn tại. Khi gặp một bài toán lập trình mới, bạn hãy tự hỏi mình: "Liệu có module nào trong standard library đã làm sẵn rồi việc này rồi không"

- Một số module phổ biến trong standard library được cung cấp:
  1. **Mathematical Operations (phép toán toán học) - module `math`**:
       - Cung cấp các hàm cho **trigonometry (lượng giác), logarithms (logarit), exponentiation (lũy thừa)** và các phép toán phổ biến khác. VD:
      - `math.sqrt()` dùng để tính **square root (căn bậc hai)**.
  2. **Random Number Generation (tạo số ngẫu nhiên) - module `random`**:
      - Cho phép bạn tạo pseudo-random numbers (số giả ngẫu nhiên), xáo trộn chuỗi phần tử, hoặc chọn ngẫu nhiên.
      - Rất hữu ích cho mô phỏng, game, hoặc statistical sampling (lấy mẫu thống kê).
  3. **Operating System Interaction (tương tác với hệ điều hành) - module `os`**:
      - Cung cấp các classes để làm việc với hệ điều hành bên dưới, như duyệt **file systems (hệ thông tệp) - `os.path`**, **quản lý thư mục - `os.mkdir`**, và truy cập **environment variable (biến môi trường)**.
  4. **Reading and Writing Data Formats (đọc và ghi các định dạng dữ liệu)**: các module như `csv` và `json` giúp bạn làm việc với các định dạng file dữ liệu phổ biến.

## **#5: Finding and Installing External Packages with pip**

- **Python standard library (thư viện chuẩn)**: là tập hợp các module và gói được cài đặt sẵn cùng với Python khi bạn tải xuống ngôn ngữ này. Bạn không cần cài đặt thêm bất cứ thứ gì để sử dụng chúng. Chúng cung cấp các chức năng cơ bản như xử lý file, làm việc với thời gian, toán học, hay kết nối mạng cơ bản. Ví dụ: `math`, `os`, `sys`, `json`. để xử lý mọi thứ:
  - web development
  - data analysis
  - machine learning
  - game
  - ....

- **Third-party packages (gói bên thứ ba)**: một kho lưu trữ trực tuyến khổng lồ (thường được gọi là "Repo") chứa hàng trăm nghìn gói phần mềm do cộng đồng phát triển. Đây là nơi bạn tìm kiếm các thư viện chuyên sâu mà Python không cài sẵn.

- Vậy làm thế nào để bạn tìm và dùng những **Third-party packages** này trong project của mình? Đây là lúc **Python Package Index (PyPI – chỉ mục gói Python)** và công cụ `pip` xuất hiện.

### **1. Python Package Index (PyPI)**

- Hãy xem **Python Package Index (PyPI)**, như một **central repository (kho lưu trữ trung tâm)** chính thức cho Python. Nó giống như một **app store (cửa hàng ứng dụng)** dành riêng cho Python libraries.
- **Developer** có thể **upload packages của họ lên PyPI để người khác tải về và sử dụng**.

### **2. Introducing `pip`: The Package Installer**

- `pip` là **package manager (trình quản lý gói)**. Nó là một **command-line tool (công cụ dòng lệnh)**

- Khi bạn cài một phiên bản Python, `pip` thường đã được cài đặt theo. ## **#6. Organizing Code into Packages (Basic Structure)**

- Khi project của bạn phát triển quá lớn có vài trăm functions hoặc classes, việc nhét mọi thứ vào một file `.py` sẽ nhanh chóng trở nên khó quản lý. Bạn sẽ gặp khó khăn khi nhớ chính xác một function được định nghĩa ở đâu, hoặc tệ hơn là vô tình dùng cùng một tên cho những logic khác nhau.

- **Modules** tổ chức code bên trong từng file riêng lẻ, còn **packages (gói)** mở rộng cách tổ chức đó sang chính các modules.

- Hãy xem một **package** như một **directory/folder (thư mục)** chứa các modules có liên quan. Cấu trúc phân cấp này giúp các project lớn dễ quản lý hơn nhiều. Python sẽ nhận diện một directory là package nếu nó chứa một file đặc biệt tên là `__init__.py`.

### **1. The Basic Package Structure**

- Ở mức đơn giản nhất, một **Python package** là **một thư mục chứa các Python modules** và **một file `__init__.py`**.

- Giả sử bạn đang xây dựng một **utility library (thư viện tiện ích)** với các **function cho string manipulation (xử lý chuỗi)** và **mathematical calculations (tính toán toán học)**. Bạn **có thể tổ chức nó** thành một **package** tên là `myutils`

    ```python
    myproject/
    ├── main.py
    └── myutils/
        ├── __init__.py
        ├── string_ops.py
        └── math_ops.py
    ```

- Trong cấu trúc này:

    - `myproject/` là **root directory (thư mục gốc)** của project.
    - `main.py` là một script nơi bạn có thể sử dụng **utility package** của mình.
    - `myutils/` là **directory đại diện cho package**.
    - `__init__.py` là file quan trọng cho Python biết: Hãy xem thư mục `myutils` như một **package**"
    - `string_ops.py` và `math_ops.py` là **các modules nằm bên trong package `myutils`**.

### **2. The Role of `__init__.py`**

- File `__init__.py` là phần rất quan trọng. Khi file này có mặt trong một directory, **Python interpreter** sẽ hiểu rằng **directory** đó nên được **xem như một package**. Điều này cho phép bạn import các modules từ directory đó bằng dot notation.

- Với các package cơ bản, `__init__.py` thường có thể rỗng hoàn toàn. Chỉ cần nó tồn tại là đủ.

- Trong các tình huống nâng cao hơn, bạn có thể viết code bên trong `__init__.py`. Code này sẽ tự động chạy lần đầu tiên khi package hoặc một trong các modules của nó được import. Điều này có thể hữu ích để:
    - Thiết lập các package-level variables (biến ở cấp package)
    - Tự động import các submodules (module con)
    - Định nghĩa các symbols (ký hiệu/tên) có vẻ như đến trực tiếp từ package

### **3. Subpackages**

- Cũng như bạn đặt modules bên trong một **package directory**, bạn có thể đặt các **directories** khác bên trong **package directory** đó, và mỗi **directory** lại có `__init__.py` riêng.
- Những directory này được gọi là **subpackages (gói con)**. Điều này cho phép tổ chức code ở nhiều cấp độ sâu hơn.

### **4. Why Use Packages?**

- Tổ chức code thành packages có nhiều lợi ích, đặc biệt khi project ngày càng lớn:
    - **Namespace Management (quản lý namespace)**: packages giúp tránh xung đột tên giữa các modules. `myutils.math_ops.add` khác với `another_package.add`.
    - **Logical Grouping (nhóm logic)**: các modules liên quan được đặt cùng nhau, giúp cấu trúc project dễ hiểu và dễ điều hướng hơn.
    - **Maintainability (khả năng bảo trì)**: dễ tìm, cập nhật và debug code hơn khi code được tổ chức hợp lý.
    - **Reusability (khả năng tái sử dụng)**: các packages được định nghĩa tốt có thể dùng lại trong các project khác.
