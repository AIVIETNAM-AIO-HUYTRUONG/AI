# **Hands-on Practical: Reading and Writing Text Files**

- Python cho phép chương trình tương tác với file trên máy tính, vì vậy nó có thể đọc **existing data** và lưu **new information**.
- Trong bài thực hành chúng ta sẽ minh họa cách kết hợp các kỹ năng xử lý file:
  -. **Đọc từ một file**
  -. **Sửa nội dung của nó**
  -. **Ghi kết quả sang một file mới**.

## **Goal**:
  1. Tạo một **text file** bạn đầu chứa một vài **tên**.
  2. Đọc các tên này từ file vào Python script.
  3. Hỏi người dùng nhập tên mới không ?
  4. Thêm tên mới này vào danh sách
  5. Ghi danh sách tên đã đượ cập nhật vào một **text file** mới.

### **Step 1: Creating the Initial Names File**

- Trước tiên, chúng ta hãy tạo file khởi đầu `names.txt`. File này sẽ chứa ba tên có sẵn.

```python
# Định nghĩa danh sách tên ban đầu
initial_names =["Alice","Bob","Charlie"]

# Chỉ định tên file
filename = "names.txt"

# Dùng `with open` ở chế độ **write mode ('w')**
# để tạo vào tránh các lỗi về xử lý file.

try:
  with open(filename, 'w') as file_object:
        for name in initial_names:
          file_object.write(name + "\n")
  print(f"Successfully created and wrote initial names to {filename}")
except IOError:
  print(f"Error: Could not write to file {filename}")
```

## **Step 2: Reading Names from the File**

- B2: hãy viết code để có thể đọc file `names.txt`

```python
names_from_file = []

try:
  with open(filename, 'r') as file_object:
    for line in file_object:
      clean_name = line.strip()
      names_from_file.append(clean_name)

  print(f"Successfully read names from {filename}:")
  print(names_from_file)
except FileNotFoundError:
    print(f"Error: The file {filename} was not found.")
except IOError:
    print(f"Error: Could not read from file {filename}.")
```

### **Step 3: Getting User Input**

- B3: Chúng ta yêu cầu người dùng nhập một tên mới để thêm vào danh sách

```python
# Get a new name from the user
new_name = input("Please enter a new name to add:")
```

### **Step 4: Adding the New Name**
- B4: ta sẽ thêm `new_name` nhận được từ người dùng vào `names_from_file` ở B2

```python
if names_from_file:
  names_from_file.append(new_name)
  print(f"Added: {new_name}. Updated list: {names_from_file}")
else:
  print("Cannot add name as the initial list")
```

### **Step 5: Writing the Updated List to a New File**

- Cuối cùng, ta sẽ ghi toàn bộ `names_from_file` đã update vào một file mới `updated_names.txt`.
- Ghi ra file mới an toàn hơn là ghi đè ngay lên fiel gốc.

```python
# Chỉ định file mới cho danh sách đã cập nhật
updated_filename = "updated_names.txt"

if names_from_file:
  try:
    with open(updated_filename, 'w') as file_object:
      for name in names_from_file:
        file_object.write(name + '\n')
    print(f"Successull wrote updated names to {updated_filename}")
  except IOError:
    print(f"Erorr: Could not write to file {updated_filename}")
```

## **Complete Script**

```python
# --- Cấu hình ---
initial_filename = "names.txt"
updated_filename = "updated_names.txt"
initial_names_data = ["Alice", "Bob", "Charlie"]

# --- Bước 1: Tạo file ban đầu (tùy chọn, nếu chưa có sẵn) ---
try:
    # Thử đọc trước để tránh ghi không cần thiết nếu file đã tồn tại
    with open(initial_filename, 'r') as f:
        print(f"{initial_filename} already exists. Skipping creation.")
except FileNotFoundError:
    print(f"{initial_filename} not found. Creating it...")
    try:
        with open(initial_filename, 'w') as file_object:
            for name in initial_names_data:
                file_object.write(name + "\n")
        print(f"Successfully created {initial_filename}")
    except IOError:
        print(f"Error: Could not write initial file {initial_filename}. Exiting.")
        exit()  # Thoát nếu không thể tạo file cần thiết

# --- Bước 2: Đọc tên từ file ---
names_from_file = []
try:
    print(f"Reading names from {initial_filename}...")
    with open(initial_filename, 'r') as file_object:
        for line in file_object:
            clean_name = line.strip()
            if clean_name:  # Tránh thêm dòng rỗng nếu có
                names_from_file.append(clean_name)
    print(f"Successfully read names: {names_from_file}")

except FileNotFoundError:
    print(f"Error: The file {initial_filename} was not found. Cannot proceed.")
    exit()  # Thoát nếu file nguồn không tồn tại
except IOError:
    print(f"Error: Could not read from file {initial_filename}. Cannot proceed.")
    exit()  # Thoát nếu có lỗi đọc khác

# --- Bước 3: Lấy input từ người dùng ---
new_name = input("Please enter a new name to add: ")

# --- Bước 4: Thêm tên mới ---
# Kiểm tra cơ bản: đảm bảo người dùng đã nhập gì đó
if new_name.strip():
    names_from_file.append(new_name.strip())
    print(f"Added '{new_name.strip()}'. List is now: {names_from_file}")
else:
    print("No valid name entered. List remains unchanged.")

# --- Bước 5: Ghi danh sách đã cập nhật vào file mới ---
print(f"Writing updated list to {updated_filename}...")
try:
    with open(updated_filename, 'w') as file_object:
        for name in names_from_file:
            file_object.write(name + "\n")
    print(f"Successfully wrote updated names to {updated_filename}")
except IOError:
    print(f"Error: Could not write to file {updated_filename}")

print("\nProcess finished. Check names.txt and updated_names.txt.")
```
