### **1. Representing Data as Numbers**

- Trước khi một thuật toán có thể học, chúng ta cần một cách có **hệ thống để tổ chức dữ liệu**.

- VD: Dự đoán giá nhà, thông tin của một căn nhà có thể gồm:
    - Diện tích: `1500 sq.ft`
    - Số phòng ngủ: `3`
    - Tuổi của nhà: `20 năm`
    
    => Chúng ta có thể biểu diễn căn nhà này dưới dạng một **vector**:

<p align="center">
$house_{vector} = \begin{bmatrix}
  1500 ,
  3 ,
  20
\end{bmatrix}$</p>

- Vector này không chỉ ở dạng **danh sách số**. Nó còn có thể được hiểu là **một điểm trong không gian ba chiều** trong đó mỗi chiều tương ứng với **một feature**:

- Ta có thể có một căn nhà khác và cũng có:
    - 2100 sq.ft

    - 5 năm tuổi
    
    => Và ta sẽ có một **vector** của một căn nhà khác:

<p align="center">
$\begin{bmatrix}
  2100 ,
  4 ,
  5

- Khi chúng ta thu thập data của **hàng nghìn căn nhà**, các **vector** này có thể được **xếp chồng lên nhau tạo thành một bảng số**. Và ta gọi bảng này là **matrix**.

<p align="center">
$house_{vector} = \begin{bmatrix}
  1500 & 3 & 20  \\
  2100  & 4 & 5\\
  1800 & 3 & 10 \\
  ... & ... & ..
\end{bmatrix}$
 </p>

- Trong **matrix** này có:
    - Mỗi **hàng (row)** - một **căn nhà (sample)**
    - Mỗi **cột (column)** - một **đặc trưng (feature)**
    

- **Matrix** này trở thành **một đối tượng chính mà mô hình Machine Learning sẽ  làm việc.**

### **2. Computing with Data Efficiently**

- Sau khi data được tổ chức thành **vector và matrix**, chúng ta cần thực hiện các phép tính:
- VD:
    - Điều chỉnh **tham số mô hình (model parameters)**.
    - Tính **similarity** giữa các đối tượng.
    - **Transform dữ liệu** để làm nổi bật các pattern quan trọng

- Nếu **không có Linear algebra**, bạn sẽ phải viết **vòng lặp (loops)** để xử lý từng số một trong **dataset** => Điều này: Rất chậm, không hiệu quả, làm code khó đọc hơn.

```python
dataset = [[8, 7, 9], [6, 5, 7], [9, 9, 10]]

for i in range(len(dataset)):
    for j in range(len(dataset[i])):
        dataset[i][j] = dataset[i][j] + 1

print(f"dataset: {dataset}")
```

- **Linear algebra** cho phép chúng ta **viết một phép toán cho toàn bộ dataset chỉ trong một dòng code**. - VD:
    - **Nhân ma trận (matrix multiplication)** cho phép xử lý **hàng nghìn dữ liệu cùng lúc**

    => Điều này không chỉ là tiện lợi, nó là **lý do khiến Machine Learning hiện đại có thể chạy được trên thực tế.**

```python
import numpy as np

dataset = [[8, 7, 9], [6, 5, 7], [9, 9, 10]]
matrix = np.array(dataset)

matrix = matrix + 1

print(f"{matrix}")
```

- Các thư viện như NumPy sử dụng các **thư viện toán học cấp thấp (như BLAS/LAPACK) được tối ưu hóa bằng ngôn ngữ C hoặc Fortran**. Chúng **thực hiện phép tính này trên toàn bộ khối dữ liệu cùng lúc thay vì từng số một**

### **3. The Blueprint for Machine Learning Models**

## **#2: Scalars: The Simplest Objects**

- **Một giá trị đơn lẻ (A single value represents)** là dạng thông tin cơ bản nhất trong data. Trong **linear algebra** giá trị này được gọi là **scalar**.

- **Scalar** đơn giản chỉ là **một con số bình thường**. Nó có **độ lớn (magnitude)** nhưng **không có hướng (direction)**. Đây là **phần tử dữ liệu đơn giản nhất** dùng trong tính toán.

- VD: Trong đời sống **chúng ta** thường gặp **scalar**:
    - Nhiệt độ trong phòng: $20^\circ\text{C}$
    - Giá của một món đồ: $49.95$

    => Những con số riêng đơn lẻ này chính là **scalar**.

### **1. Scalars in a Machine Learning Context**

- Trong **machine learning**, **scalar** xuất hiện ở khắp nơi. Chúng thường được dùng để biểu diễn những giá trị đơn lẻ như:
    - **Feature value**: Tuổi của một người - `34`
    - **Hyperparameter**: learning rate của mô hình - `0.01`
    - **Measurement**: Độ sáng của một pixel trong ảnh grayscale - `182`
    - **Model** output: Xác suất dự đoán một sự kiện - `0.87`

#### **1.1 Mathematical Symbols**

- Về mặt toán học, **scalar** thường được viết bằng:
    - Chữ cái thường in nghiêng: *$c, k, x, ...$*
    - Hoặc chữ cái Hy Lạp: $𝛂 (alpha), λ (lambda),...$
    - **Ví dụ**: ta có thể định nghĩa một scalar *$s$* như sau:
        -  $s = 5$

### **3. Relationship between Scalars and Vectors**

- **Scalars** là các thành phần cấu tạo nên **Vectors**. Một vector $v$ trong không gian $n$ chiều thực chất là một danh sách gồm $n$ scalars.
- **Scalar Multiplication**: Đây là phép toán quan trọng nhất thể hiện mối quan hệ giữa chúng. Khi nhân một scalar $k$ với một vector $v$, mỗi phần tử trong vector sẽ được nhân với $k$. Điều này làm thay đổi độ lớn (magnitude) của vector nhưng giữ nguyên (hoặc đảo ngược) hướng của nó.

VD: Nếu bạn có một vector vận tốc, và bạn nhân nó với một scalar là 2, bạn đang gấp đôi tốc độ nhưng vẫn đi theo hướng cũ.

```python
import numpy as np

# Define a vector (a set of scalars).
vector = np.array([1, 2, 3])

# Define a scalar
scalar_k = 2

# Scalar multiplication by vector
scaled_vector = scalar_k * vector

print(f"Original Vector: {vector}")
print(f"Scalar: {scalar_k}")
print(f"Scaled Vector (k * v): {scaled_vector}")
```

### **4. Summary**

- Mặc dù **scalar rất đơn giản**, nhưng chúng là **những "nguyên tử" của dữ liệu**.
- Chúng ta sử dụng các scalar này để xây dựng các cấu trúc phức tạp hơn:
    - **Vector**: Một danh sách các scalars.
    - **Matrix**: Một bảng (grid) chứa các scalars.
- Trong Machine Learning, scalar đại diện cho các giá trị đơn lẻ như learning rate, giá trị đặc trưng (feature value), hoặc trọng số (bias).
- Phép nhân scalar với vector cho thấy cách một giá trị đơn lẻ có thể tác động lên toàn bộ tập hợp dữ liệu bằng cách thay đổi độ lớn của chúng.

## **#3: Vectors: Points in Space**

- Một **mẫu thông tin đơn lẻ** thường được gọi là **scalar**, nhưng hầu hết **data không đơn giản như vậy**.
- VD: khi mô tả một căn nhà trên website bất động sản, bạn không chỉ ghi **giá tiền**, mà có **diện tích, số phòng ngủ, số phòng tắm,...**. Một **danh sách có các số thứ tự** như vậy được gọi là **vector**


### **1. What is Vector**

- **Vector** là một cách cơ bản để **lưu trữ dữ liệu có nhiều thuộc tính**. Nếu một căn nhà có:
    - Diện tích: **1500 sq.ft**
    - Phòng ngủ: **3**
    - Phòng tắm: **2**
    
    - Ta có thể biểu diễn thông tin trên bằng **vector**: $x = [1500, 3, 2]$
- **Một con số** trong vector là một **element** hoặc **component**.
- **Thứ tự của chúng rất quan trọng**. **Ví dụ**: **vector**: $x = [3, 2, 1500]$ sẽ biểu diễn một căn nhà hoàn toàn khác:
    - Diện tích: **3 sq.ft**
    - Phòng ngủ: **2**
    - Phòng tắm: **1500**
    
    => Điều này hoàn toàn không hợp lý.

### **2. Geometric Interpretation**

- Ngoài vai trò là một **container** chứa các con số **(element / component)**, vector còn có ý nghĩa về mặt hình học. Hãy tưởng tượng, xét một **vector có hai thành phần**:
<p align="center">
$v = [3, 4]$
</p>

- Ta có thể **hình dung vector này trong hệ tọa độ hai chiều (2D)**:
    - Có hai cách để hiểu về vector này:
        1. **Vector như một điểm (As a Point)**: Vector **$v=[3,4]$** có thể biểu diễn **tọa độ của một điểm trong không gian**:
            - 3 ở (x)
            - 4 ở (y)
        2. **Vector như một mũi tên (As an Arrow)**:
            - Bắt từ **gốc tọa độ**: $(0,0)$ và kết thúc tại $(3,4)$ - cách hiểu này cho **vector** hai thuộc tính quan trọng:
                - **Độ dài (Magnitude)** → mũi tên dài bao nhiêu
                - **Hướng (Direction)** → mũi tên chỉ về đâu

#### **2.1 Vector in Machine Learning**

- Trong Machine Learning, dữ liệu thường có **rất nhiều chiều hơn rất nhiều so với 2 hay 3 chiều** mà ta có thể dễ dàng minh họa ra.
- **Ví dụ**: Một vector biểu diễn **ảnh grayscale** kích thước **28 x 28 pixel** sẽ có:
**28x28=784 component**. Dù ta không thể vẽ không gian **784 chiều**, nhưng các ý tưởng hình học như **length** và **direction** vẫn được áp dụng được.

### **3. Notation**
- Trong sách và tài liệu toán học, bạn sẽ thấy một số cách ký hiệu vector phổ biến.

#### **3.1 Lowercase, Bold Letters**
- **Lowercase, Bold Letters**: Vector thường được viết bằng chữ thường in đậm: **$v,x$**

#### **3.2 Column and Row Vectors**
- **Column, Row Vectors**: Trong đại số tuyến tính, vector thường được **mặc định là column vector**.

- **Column vector** sắp xếp các component theo chiều dọc:
<p align="center">
$v = \begin{bmatrix}
  3  \\
  4  
 \end{bmatrix}$
</p>

-  **Row vector** sắp xếp các component theo chiều ngang:
<p align="center">
$v = [3, 4]$
</p>

- **Sự khác biệt** này sẽ trở nên quan trọng khi chúng ta **bắt đầu nhân vector với matrix**. Còn bây giờ, ta chỉ **cần nhớ rằng cả hai dạng đều tồn tại**. Số lượng component trong một vector quyết định **dimension** của nó:
    - **Vector [3,4]** là **vector 2 chiều**
    - **Vector [1500,3,2]** là **vector 3 chiều**.

- **Vector** cho phép chúng ta **đóng gói các feature của một data point vào một cấu trúc duy nhất**. Tuy nhiên, **một dataset** thường **chứa nhiều data point**.
- Về mặt toán học, **một data point** thường được biểu diễn dưới dạng một **vector đặc trưng (feature vector)**. **Các thành phần của vector** này là các **giá trị số (features) mô tả đặc điểm của thực thể đó** -  Việc hiểu rõ cách biểu diễn các **data point** dưới dạng vector là nền tảng quan trọng để thực hiện các phép toán đại số tuyến tính trong học máy:
    - VD: $v = [3, 4]$ là một ví dụ điển hình về **cách biểu diễn một data point dưới dạng vector trong không gian 2 chiều**. Trong trường hợp này, **data point này có 2 đặc trưng (features)**:
        - Feature thứ nhất: **3**
        - Feature thứ hai: **4**

- Để tổ chức một tập hợp các vector như vậy, chúng ta cần cấu trúc tiếp theo: **matrix**.

### **4. Summary: Vectors**

- **Vector** là một **danh sách có thứ tự các số (scalar)** dùng để biểu diễn một **điểm trong không gian** hoặc một **sample dữ liệu** có nhiều thuộc tính (features).
- Mỗi con số trong vector được gọi là một **component** hoặc **element**.
- Về mặt hình học, vector có thể được hiểu là **một điểm (point)** trong không gian hoặc **một mũi tên (arrow)** từ gốc tọa độ, có **độ lớn (magnitude)** và **hướng (direction)**.
- Trong Machine Learning, vector là cấu trúc cơ bản để tổ chức dữ liệu, cho phép chúng ta làm việc với dữ liệu đa chiều một cách hiệu quả.

## **#4: Matrices: Organizing Data in Grids**

- Trong khi **vector** rất phù hợp để biểu diễn một điểm với nhiều feature, các mô hình machine learning hiếm khi được huấn luyện trên một ví dụ.
- Chúng ta cần một cách để tổ chức cả một tập hợp các data point, và đó là **matrix**


### **1. What is Matrix**

- **Matrix** là một bảng số hình chữ nhật được sắp xếp theo **row và column**.
- Nếu **vector** là **một danh sách số**, thì **matrix** là **nhiều danh sách số được xếp chồng lên nhau**. Cấu trúc này rất quan trọng để **biểu diễn dataset trong machine learning**.

#### **1.1 Matrix and Dataset**
- Sự khác biệt giữa **matrix** và **dataset** nằm ở bản chất của chúng: một cái là **công cụ toán học**, cái còn lại là **nội dung thông tin**.

    1. **Dataset**:
        - Đây là tập hợp các dữ liệu thực tế mà bạn thu thập được từ thế giới bên ngoài (ví dụ: danh sách thông tin 1.000 ngôi nhà, file ảnh của 5.000 khuôn mặt, hoặc lịch sử giao dịch của khách hàng)
        - Nó chứa đựng thông tin về đối tượng, sự kiện hoặc hiện tượng mà bạn muốn nghiên cứu
    2. **Matrix**:
        - Đây là cấu trúc toán học dùng để biểu diễn dataset đó dưới dạng con số để máy tính có thể tính toán được.
        - Nó là khái niệm mang tính cấu trúc. Ma trận không quan tâm dữ liệu đó là "giá nhà" hay "pixel ảnh", nó chỉ quan tâm đến các hàng, cột và giá trị số nằm trong đó

### **2. The Anatomy of a Matrix**

- Giả sử chúng ta đang xây dựng một mô hình để dự đoán **giá nhà**. Chúng ta thu thập dữ liệu của nhiều căn nhà. Mỗi căn nhà là một **sample**, và với **mỗi sample** ta ghi lại **một số feature** như:
    - Diện tích
    - Số phòng ngủ
    - Tuổi của nhà
- Ta có thể biểu diễn mỗi căn nhà bằng một **feature vector**:
    - House 1: `[1500, 3, 10]` (1500 sq ft, 3 bedrooms, 10 năm tuổi)
    - House 2: `[2100, 4, 5]` (2100 sq ft, 4 bedrooms, 5 năm tuổi)
    - House 3: `[1200, 2, 20]` (1200 sq ft, 2 bedrooms, 20 năm tuổi)
- Để **tổ chức toàn bộ dataset** này, ta có thể **xếp các row vectors lại với nhau để tạo thành một matrix**.

<div style="text-align: center;">

|          | Feature 1 (Size sq ft) | Feature 2 (Bedrooms) | Feature 3 (Age years) |
| :------: | :--------------------: | :------------------: | :-------------------: |
| Sample 1 | 1500                   | 3                    | 10                    |
| Sample 2 | 2100                   | 4                    | 5                     |
| Sample 3 | 1200                   | 2                    | 20                    |

</div>

- Trong matrix này:
    - **Rows** chạy theo chiều ngang và đại diện cho từng **data sample** hoặc **observation**. Matrix của chúng ta có 3 rows.
    - **Columns** chạy theo chiều dọc và đại diện cho các **feature** khác nhau của dữ liệu. Matrix của chúng ta có 3 columns.

- Kích thước của matrix được gọi là **dimension** hoặc **shape**.
- **Matrix** được mô tả dưới dạng $m x n$:
    - $m$: số rows
    - $n$: số columns
    - Ma trận dữ liệu nhà ở phía trên là một **matrix $3 × 3$**

### **3. Mathematical Notation for Matrices**

- Trong linear algebra, **matrix** thường được **ký hiệu bằng chữ in hoa**, ví dụ **$A$**.
- Để có thể chỉ định một phần tử cụ thể trong **matrix**, ta viết: **$A_{ij}$** - Trong đó:

    - $𝑖$ = row
    - $𝑗$= column

⚠️ **Quy tắc: Row trước, Column sau**

- VD: **Housing data matrix A** của chúng ta:

<p align="center">
 $A = \begin{bmatrix}
  1500 &3 & 10 \\
  2100 &4 & 5 \\
  1200 &2 & 20
 \end{bmatrix}$
</p>

- Phần tử **$A_{1,1} = 1500$**
- Phần tử **$A_{2,3} = 5$**
- Phần tử **$A_{3,1} = 1200$**

### **4.The Standard Convention in Machine Learning**

- Cách sắp xếp **mỗi sample một row** và **mỗi feature một column** là quy ước tiêu chuẩn trong machine learning. **Data matrix** này thường được **gọi là feature matrix** và thường được biểu diễn bằng biến **$X$**

- Các **giá trị mà ta muốn dự đoán**, như giá của mỗi căn nhà, thường được lưu trong một **column vector riêng**. **Vector** này thường được gọi là **target vector** và được **biểu diễn bằng biến $y$**

- Với ví dụ của chúng ta, **feature matrix $X$** sẽ là matrix đã thấy ở trên, và **target vector $y$** có thể chứa các mức giá tương ứng:

<p align="center">
 $X = \begin{bmatrix}
  1500 &3 & 10 \\
  2100 &4 & 5 \\
  1200 &2 & 20
 \end{bmatrix}$,
 $y = \begin{bmatrix}
  300000 \\
  450000 \\
  220000
 \end{bmatrix}$
</p>

- Cách biểu diễn có cấu trúc này cho **features (X)** và **targets (y)** là **input cho gần như mọi supervised machine learning algorithm**.

- Bằng cách **tổ chức dữ liệu thành matrices và vectors**, chúng ta **có thể dùng các phép toán mạnh mẽ của linear algebra** để:
    - Phân tích mối quan hệ giữa các feature
    - Tối ưu mô hình
    - Xây dựng hệ thống dự đoán.

### **5. Summary: Matrices**

- **Matrix** là một bảng số hình chữ nhật (grid) dùng để **tổ chức nhiều vector dữ liệu thành một dataset**.
- Mỗi **hàng (row)** trong matrix thường đại diện cho một **data sample** (ví dụ: một căn nhà), và mỗi **cột (column)** đại diện cho một **feature** (ví dụ: diện tích, số phòng ngủ).
- Kích thước của matrix được mô tả bằng **dimension** hoặc **shape** ($m \times n$, trong đó $m$ là số hàng và $n$ là số cột).
- Trong Machine Learning, **data matrix** (thường ký hiệu là $X$) và **target vector** (thường ký hiệu là $y$) là cấu trúc cơ bản để biểu diễn dữ liệu đầu vào và đầu ra cho các thuật toán.
- Việc sử dụng matrices cho phép thực hiện các phép toán đại số tuyến tính mạnh mẽ, hiệu quả trên toàn bộ dataset, tạo nền tảng cho hầu hết các mô hình Machine Learning.
